## 1. Convert Negative Values in a DataFrame
Use `df.abs()` to convert negative numbers into positive absolute values.

````python
import pandas as pd

df = pd.DataFrame({'MATH': [-30, 40, -50], 'ENGLISH': [20, -30, 40]})
result = df.abs()
print(result)

# Self-validating assertion
assert result.loc[0, 'MATH'] == 30
assert result.loc[1, 'ENGLISH'] == 30
assert (result >= 0).all().all()
````

**Expected Output:**
```
   MATH  ENGLISH
0    30       20
1    40       30
2    50       40
```

## 2. Absolute Values on Pandas Series
Apply `abs()` to a single Series containing float values.

````python
s = pd.Series([-1.5, 2.0, -3.75, 0.0])
abs_s = s.abs()
print(abs_s)

# Self-validating assertion
assert abs_s.tolist() == [1.5, 2.0, 3.75, 0.0]
````

**Expected Output:**
```
0    1.50
1    2.00
2    3.75
3    0.00
dtype: float64
```
