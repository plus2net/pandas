## 1. Absolute Values on Series
Transform signed float/integer Series values into non-negative magnitudes.

````python
import pandas as pd
import numpy as np

# Exercise 1: Single Series Transformation
s = pd.Series([-1.5, 2.0, -3.75, 0.0, -42.0])
abs_s = s.abs()

print('Absolute Series:')
print(abs_s)

# Self-validating assertion
assert abs_s.tolist() == [1.5, 2.0, 3.75, 0.0, 42.0], 'Assertion Failed: abs_s output mismatch'
````

**Expected Output:**
```
Absolute Series:
0     1.50
1     2.00
2     3.75
3     0.00
4    42.00
dtype: float64
```

## 2. Multi-Column DataFrame & Missing Data Handling
Apply `abs()` across numeric columns while preserving `NaN` values.

````python
df = pd.DataFrame({
    'MATH_VAR': [-30, 40, -50, np.nan],
    'ENG_VAR': [20, -30, 40, -5]
})

df_abs = df.abs()
print('Absolute DataFrame:')
print(df_abs)

# Self-validating assertion
assert df_abs.loc[0, 'MATH_VAR'] == 30
assert pd.isna(df_abs.loc[3, 'MATH_VAR'])
assert df_abs.loc[3, 'ENG_VAR'] == 5
````

**Expected Output:**
```
Absolute DataFrame:
   MATH_VAR  ENG_VAR
0      30.0       20
1      40.0       30
2      50.0       40
3       NaN        5
```

## 3. Mean Absolute Error (MAE) Calculation
Calculate Mean Absolute Error between target and actual readings.

````python
actual = pd.Series([46500, 78200, 54000, 91500])
target = pd.Series([50000, 75000, 60000, 90000])

deviation = (actual - target).abs()
mae = deviation.mean()

print(f'Mean Absolute Error: ${mae:.2f}')

assert mae == 3550.0, 'Assertion Failed: MAE calculation mismatch'
````

**Expected Output:**
```
Mean Absolute Error: $3550.00
```
