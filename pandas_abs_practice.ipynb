## 1. Convert Negative Values in a DataFrame

Use `df.abs()` to convert negative numbers into positive absolute values across multiple columns.

````python
import pandas as pd

df = pd.DataFrame({'MATH': [-30, 40, -50], 'ENGLISH': [20, -30, 40]})
result = df.abs()
print(result)

# Self-validating assertion
assert result.loc[0, 'MATH'] == 30
assert result.loc[1, 'ENGLISH'] == 30
assert (result >= 0).all().all()
````

**Expected output**

```text
   MATH  ENGLISH
0    30       20
1    40       30
2    50       40
```

## 2. Series Single-Column Transformation

Apply `abs()` to a single Pandas Series containing floating-point numbers.

````python
s = pd.Series([-1.5, 2.0, -3.75, 0.0, -42.0])
abs_s = s.abs()
print(abs_s)

# Self-validating assertion
assert abs_s.tolist() == [1.5, 2.0, 3.75, 0.0, 42.0]
````

**Expected output**

```text
0     1.50
1     2.00
2     3.75
3     0.00
4    42.00
dtype: float64
```

## 3. Multi-Column DataFrame & Selective Numeric Types

Safely calculate absolute values on specific numeric columns while retaining original text metadata columns.

````python
student_data = {
    'NAME': ['Ravi', 'Raju', 'Alex', 'Ananya'],
    'MATH_VARIANCE': [-12.5, 8.0, -15.2, 0.0],
    'ENG_VARIANCE': [5.0, -10.0, 14.0, -3.5]
}
df_students = pd.DataFrame(student_data)

df_students[['MATH_ABS', 'ENG_ABS']] = df_students[['MATH_VARIANCE', 'ENG_VARIANCE']].abs()
print(df_students)

# Self-validating assertion
assert df_students.loc[0, 'MATH_ABS'] == 12.5
assert df_students.loc[1, 'ENG_ABS'] == 10.0
````

**Expected output**

```text
     NAME  MATH_VARIANCE  ENG_VARIANCE  MATH_ABS  ENG_ABS
0    Ravi          -12.5           5.0      12.5      5.0
1    Raju            8.0         -10.0       8.0     10.0
2    Alex          -15.2          14.0      15.2     14.0
3  Ananya            0.0          -3.5       0.0      3.5
```

## 4. Handling Missing Data (NaN) and Complex Numbers

Verify that `NaN` values are preserved and complex numbers calculate modulus $\sqrt{a^2 + b^2}$.

````python
import numpy as np

df_complex = pd.DataFrame({
    'REAL': [-10.5, np.nan, -25.0],
    'COMPLEX': [3 + 4j, -5 - 12j, np.nan]
})
result_complex = df_complex.abs()
print(result_complex)

# Self-validating assertion
assert result_complex.loc[0, 'COMPLEX'] == 5.0
assert result_complex.loc[1, 'COMPLEX'] == 13.0
assert pd.isna(result_complex.loc[1, 'REAL'])
````

**Expected output**

```text
   REAL  COMPLEX
0  10.5      5.0
1   NaN     13.0
2  25.0      NaN
```

## 5. Case Study: Calculating Mean Absolute Error (MAE)

Compute MAE between target benchmarks and actual observed figures.

````python
actual = pd.Series([46500, 78200, 54000, 91500])
target = pd.Series([50000, 75000, 60000, 90000])

mae = (actual - target).abs().mean()
print(f"Mean Absolute Error: ${mae:.2f}")

# Self-validating assertion
assert mae == 3550.0
````

**Expected output**

```text
Mean Absolute Error: $3550.00
```
