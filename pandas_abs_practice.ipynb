# Pandas DataFrame `abs()` Method - Interactive Practice
This notebook provides hands-on exercises to master the `abs()` method in Pandas.

```python
import pandas as pd
import numpy as np

# Exercise 1: Convert negative Series to positive
s = pd.Series([-10, 20, -30, 40, -50])
print('Original Series:')
print(s)
print('Absolute Series:')
print(s.abs())
```

```python
# Exercise 2: Multi-column DataFrame abs calculation
df = pd.DataFrame({
    'A': [-1.2, 3.4, -5.6],
    'B': [10, -20, 30],
    'C': [-100, -200, -300]
})
print('DataFrame Absolute Values:')
print(df.abs())
```

```python
# Exercise 3: Calculate Mean Absolute Error (MAE)
actual = pd.Series([120, 140, 190, 210])
predicted = pd.Series([115, 148, 185, 215])
error = actual - predicted
mae = error.abs().mean()
print(f'Mean Absolute Error (MAE): {mae}')
```
