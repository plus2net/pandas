## 1. Basic Working Example

Execute the code below to test core functionality.

````python
import pandas as pd
df = pd.DataFrame({'MATH': [-30, 40], 'ENGLISH': [20, -30]})
result = df.abs()
print(result)
assert result.loc[0, 'MATH'] == 30
assert not result.empty
````

**Expected output**

```text
   MATH  ENGLISH
0    30       20
1    40       30
```
