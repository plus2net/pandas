## 1. Scalar Addition to Columns
Add constant scalar values across DataFrame columns.

````python
import pandas as pd

df = pd.DataFrame({
    'MATH': [30, 40, 50],
    'ENGLISH': [20, 30, 40]
})

df_bonus = df.add(5)
print('DataFrame with 5 Bonus Marks:')
print(df_bonus)

assert df_bonus.loc[0, 'MATH'] == 35
assert df_bonus.loc[2, 'ENGLISH'] == 45
````

**Expected Output:**
```
DataFrame with 5 Bonus Marks:
   MATH  ENGLISH
0    35       25
1    45       35
2    55       45
```

## 2. Axis Alignment with Series
Align Series addition along column (axis=1) and row (axis=0) directions.

````python
df_marks = pd.DataFrame({
    'MATH': [70, 80, 90],
    'ENGLISH': [60, 65, 75]
}, index=['Ravi', 'Raju', 'Alex'])

# Column-wise addition
subject_bonus = pd.Series({'MATH': 10, 'ENGLISH': 5})
col_added = df_marks.add(subject_bonus, axis=1)

print('Column-aligned addition:')
print(col_added)

assert col_added.loc['Ravi', 'MATH'] == 80
assert col_added.loc['Alex', 'ENGLISH'] == 80
````

**Expected Output:**
```
Column-aligned addition:
      MATH  ENGLISH
Ravi    80       65
Raju    90       70
Alex   100       80
```

## 3. Fill Value Parameter (`fill_value=0`)
Prevent `NaN` results during alignment of non-matching DataFrame indices.

````python
df1 = pd.DataFrame({'Q1_SALES': [100, 200]}, index=['Store_A', 'Store_B'])
df2 = pd.DataFrame({'Q1_SALES': [300, 400]}, index=['Store_B', 'Store_C'])

total_sales = df1.add(df2, fill_value=0)
print('Consolidated Sales:')
print(total_sales)

assert total_sales.loc['Store_A', 'Q1_SALES'] == 100
assert total_sales.loc['Store_B', 'Q1_SALES'] == 500
assert total_sales.loc['Store_C', 'Q1_SALES'] == 400
````

**Expected Output:**
```
Consolidated Sales:
         Q1_SALES
Store_A     100.0
Store_B     500.0
Store_C     400.0
```
