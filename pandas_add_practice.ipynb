# Pandas DataFrame `add()` Method - Interactive Practice
Hands-on exercises covering scalar addition, axis alignment, and `fill_value` handling.

```python
import pandas as pd
# Exercise 1: Scalar Addition
df = pd.DataFrame({'A': [10, 20], 'B': [30, 40]})
print(df.add(100))
```

```python
# Exercise 2: Align DataFrames with fill_value=0
df1 = pd.DataFrame({'Sales': [100, 200]}, index=['Jan', 'Feb'])
df2 = pd.DataFrame({'Sales': [150, 250]}, index=['Feb', 'Mar'])
total = df1.add(df2, fill_value=0)
print(total)
```
