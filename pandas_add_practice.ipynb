## 1. Scalar Addition to DataFrame
Add a scalar value of 5 to numeric DataFrame columns.

````python
import pandas as pd

df = pd.DataFrame({'MATH': [30, 40], 'ENGLISH': [20, 30]})
result = df.add(5)
print(result)

# Self-validating assertion
assert result.loc[0, 'MATH'] == 35
assert result.loc[1, 'ENGLISH'] == 35
````

**Expected Output:**
```
   MATH  ENGLISH
0    35       25
1    45       35
```

## 2. DataFrame Addition with fill_value=0
Combine two DataFrames with mismatched index labels without creating NaNs.

````python
df1 = pd.DataFrame({'A': [100, 200]}, index=['x', 'y'])
df2 = pd.DataFrame({'A': [300]}, index=['y'])

result = df1.add(df2, fill_value=0)
print(result)

# Self-validating assertion
assert result.loc['x', 'A'] == 100.0
assert result.loc['y', 'A'] == 500.0
````

**Expected Output:**
```
       A
x  100.0
y  500.0
```
