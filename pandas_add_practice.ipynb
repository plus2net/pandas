## 1. Quick Scalar Addition

Add a scalar constant value of 5 across all numerical columns.

````python
import pandas as pd

df = pd.DataFrame({'MATH': [30, 40], 'ENGLISH': [20, 30]})
result = df.add(5)
print(result)

# Self-validating assertion
assert result.loc[0, 'MATH'] == 35
assert result.loc[1, 'ENGLISH'] == 35
````

**Expected output**

```text
   MATH  ENGLISH
0    35       25
1    45       35
```

## 2. Scalar Addition Across DataFrames

Add grace marks to exam scores.

````python
df_scores = pd.DataFrame({
    'MATH': [75, 82, 90],
    'SCIENCE': [68, 74, 88]
})
df_bonus = df_scores.add(5)
print(df_bonus)

# Self-validating assertion
assert df_bonus.loc[0, 'MATH'] == 80
assert df_bonus.loc[2, 'SCIENCE'] == 93
````

**Expected output**

```text
   MATH  SCIENCE
0    80       73
1    87       79
2    95       93
```

## 3. Series Addition with Axis Alignment (`axis=1`)

Align Series index with DataFrame columns for subject-specific score addition.

````python
df_marks = pd.DataFrame({
    'MATH': [70, 80],
    'ENGLISH': [60, 65]
}, index=['Ravi', 'Raju'])

subject_bonus = pd.Series({'MATH': 10, 'ENGLISH': 5})
df_result = df_marks.add(subject_bonus, axis=1)
print(df_result)

# Self-validating assertion
assert df_result.loc['Ravi', 'MATH'] == 80
assert df_result.loc['Raju', 'ENGLISH'] == 70
````

**Expected output**

```text
      MATH  ENGLISH
Ravi    80       65
Raju    90       70
```

## 4. Missing Data Alignment (`fill_value=0`)

Consolidate store sales figures with non-overlapping index labels.

````python
df1 = pd.DataFrame({'Q1_SALES': [100, 200]}, index=['Store_A', 'Store_B'])
df2 = pd.DataFrame({'Q1_SALES': [300, 400]}, index=['Store_B', 'Store_C'])

df_total = df1.add(df2, fill_value=0)
print(df_total)

# Self-validating assertion
assert df_total.loc['Store_A', 'Q1_SALES'] == 100.0
assert df_total.loc['Store_B', 'Q1_SALES'] == 500.0
assert df_total.loc['Store_C', 'Q1_SALES'] == 400.0
````

**Expected output**

```text
         Q1_SALES
Store_A     100.0
Store_B     500.0
Store_C     400.0
```
