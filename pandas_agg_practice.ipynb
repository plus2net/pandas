## 1. Multiple Function Aggregation
Compute `sum` and `mean` simultaneously across columns.

````python
import pandas as pd

df = pd.DataFrame({'MATH': [30, 40, 50], 'ENGLISH': [20, 30, 40]})
result = df.agg(['sum', 'mean'])
print(result)

# Self-validating assertion
assert result.loc['sum', 'MATH'] == 120
assert result.loc['mean', 'ENGLISH'] == 30.0
````

**Expected Output:**
```
       MATH  ENGLISH
sum   120.0     90.0
mean   40.0     30.0
```

## 2. Column-Specific Dict Aggregations
Apply `sum` to MATH and `mean` to ENGLISH.

````python
result = df.agg({'MATH': 'sum', 'ENGLISH': 'mean'})
print(result)

# Self-validating assertion
assert result['MATH'] == 120
assert result['ENGLISH'] == 30.0
````

**Expected Output:**
```
MATH       120.0
ENGLISH     30.0
dtype: float64
```
