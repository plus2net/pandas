## 1. Multi-Metric Aggregation
Compute multiple statistical functions across columns simultaneously.

````python
import pandas as pd

df_scores = pd.DataFrame({
    'MATH': [30, 40, 50, 80],
    'ENGLISH': [20, 30, 40, 90]
})

metrics = df_scores.agg(['sum', 'min', 'max', 'mean'])
print('Aggregated Metrics:')
print(metrics)

assert metrics.loc['sum', 'MATH'] == 200
assert metrics.loc['mean', 'ENGLISH'] == 45.0
````

**Expected Output:**
```
Aggregated Metrics:
       MATH  ENGLISH
sum   200.0    180.0
min    30.0     20.0
max    80.0     90.0
mean   50.0     45.0
```

## 2. Column-Specific Mapping & GroupBy Aggregations
Apply targeted aggregation dictionaries to GroupBy objects.

````python
sales_df = pd.DataFrame({
    'PRODUCT': ['Pen', 'Pen', 'Book', 'Book'],
    'UNITS_SOLD': [5, 15, 10, 20],
    'UNIT_PRICE': [2.5, 2.5, 15.0, 15.0]
})

custom_agg = sales_df.groupby('PRODUCT').agg({
    'UNITS_SOLD': 'sum',
    'UNIT_PRICE': 'mean'
})

print('GroupBy Custom Aggregations:')
print(custom_agg)

assert custom_agg.loc['Book', 'UNITS_SOLD'] == 30
assert custom_agg.loc['Pen', 'UNIT_PRICE'] == 2.5
````

**Expected Output:**
```
GroupBy Custom Aggregations:
         UNITS_SOLD  UNIT_PRICE
PRODUCT                        
Book             30        15.0
Pen              20         2.5
```
