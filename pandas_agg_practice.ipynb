## 1. Quick Multi-Function Aggregation

Calculate total `sum` and `mean` simultaneously.

````python
import pandas as pd

df = pd.DataFrame({'MATH': [30, 40, 50], 'ENGLISH': [20, 30, 40]})
result = df.agg(['sum', 'mean'])
print(result)

# Self-validating assertion
assert result.loc['sum', 'MATH'] == 120
assert result.loc['mean', 'ENGLISH'] == 30.0
````

**Expected output**

```text
       MATH  ENGLISH
sum   120.0     90.0
mean   40.0     30.0
```

## 2. Multi-Metric Function Lists

Pass multiple statistical aggregation strings.

````python
df_scores = pd.DataFrame({
    'MATH': [45, 60, 75, 90],
    'ENGLISH': [50, 55, 65, 80]
})

summary = df_scores.agg(['min', 'max', 'mean', 'std'])
print(summary)

# Self-validating assertion
assert summary.loc['min', 'MATH'] == 45
assert summary.loc['max', 'ENGLISH'] == 80
````

**Expected output**

```text
           MATH    ENGLISH
min   45.000000  50.000000
max   90.000000  80.000000
mean  67.500000  62.500000
std   19.364917  13.228757
```

## 3. Column-Specific Mapping

Use a dictionary mapping to apply custom aggregation rules to specific columns.

````python
sales_df = pd.DataFrame({
    'UNITS': [10, 20, 15, 30],
    'PRICE': [5.0, 5.0, 12.0, 12.0]
})

custom_summary = sales_df.agg({
    'UNITS': 'sum',
    'PRICE': 'mean'
})
print(custom_summary)

# Self-validating assertion
assert custom_summary['UNITS'] == 75
assert custom_summary['PRICE'] == 8.5
````

**Expected output**

```text
UNITS    75.0
PRICE     8.5
dtype: float64
```
