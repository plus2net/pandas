## 1. Basic Boolean Column Reduction

Test whether all elements in each column evaluate to `True`.

````python
import pandas as pd

df = pd.DataFrame({'A': [True, True], 'B': [True, False]})
result = df.all()
print(result)

# Self-validating assertion
assert result['A'] == True
assert result['B'] == False
````

**Expected output**

```text
A     True
B    False
dtype: bool
```

## 2. Row-Wise Evaluation (`axis=1`)

Verify whether each student passed all subjects.

````python
df_scores = pd.DataFrame({
    'PASS_MATH': [True, True, True],
    'PASS_ENG': [True, False, True]
}, index=['Student1', 'Student2', 'Student3'])

passed_all = df_scores.all(axis=1)
print(passed_all)

# Self-validating assertion
assert passed_all['Student1'] == True
assert passed_all['Student2'] == False
assert passed_all['Student3'] == True
````

**Expected output**

```text
Student1     True
Student2    False
Student3     True
dtype: bool
```

## 3. Data Validation & Non-Zero Stock Checks

Check if inventory counts across items remain strictly greater than zero.

````python
df_inventory = pd.DataFrame({
    'STOCK_A': [15, 20, 5],
    'STOCK_B': [10, 0, 8]
})

in_stock = (df_inventory > 0).all()
print(in_stock)

# Self-validating assertion
assert in_stock['STOCK_A'] == True
assert in_stock['STOCK_B'] == False
````

**Expected output**

```text
STOCK_A     True
STOCK_B    False
dtype: bool
```

## 4. Handling Missing Data (NaN)

Test `all()` behavior when `NaN` entries are present.

````python
import numpy as np

df_nan = pd.DataFrame({'A': [True, np.nan], 'B': [False, np.nan]})
result_nan = df_nan.all(skipna=True)
print(result_nan)

# Self-validating assertion
assert result_nan['A'] == True
assert result_nan['B'] == False
````

**Expected output**

```text
A     True
B    False
dtype: bool
```
