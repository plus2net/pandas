## 1. Basic Boolean Column Reduction

Test whether at least one element in each column is `True`.

````python
import pandas as pd

df = pd.DataFrame({'A': [False, False], 'B': [True, False]})
result = df.any()
print(result)

# Self-validating assertion
assert result['A'] == False
assert result['B'] == True
````

**Expected output**

```text
A    False
B     True
dtype: bool
```

## 2. Row-Wise Anomaly & Overtime Checks (`axis=1`)

Determine if employees logged overtime hours in any quarter.

````python
df_overtime = pd.DataFrame({
    'Q1': [5, 0, 12],
    'Q2': [0, 0, 15],
    'Q3': [8, 0, 0]
}, index=['Emp1', 'Emp2', 'Emp3'])

worked_ot = (df_overtime > 0).any(axis=1)
print(worked_ot)

# Self-validating assertion
assert worked_ot['Emp1'] == True
assert worked_ot['Emp2'] == False
assert worked_ot['Emp3'] == True
````

**Expected output**

```text
Emp1     True
Emp2    False
Emp3     True
dtype: bool
```

## 3. Missing Data Detection (`isna().any()`)

Identify columns containing missing data.

````python
import numpy as np

df_data = pd.DataFrame({
    'ID': [101, 102, 103],
    'SCORE': [88.5, np.nan, 92.0]
})

has_nan = df_data.isna().any()
print(has_nan)

# Self-validating assertion
assert has_nan['ID'] == False
assert has_nan['SCORE'] == True
````

**Expected output**

```text
ID       False
SCORE     True
dtype: bool
```
