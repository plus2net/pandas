## 1. API to DataFrame

To fetch live JSON data from a public Representational State Transfer Application Programming Interface (REST API) and convert it directly into a Pandas DataFrame, use the Python `requests` library combined with `pd.json_normalize()` or list conversion.

````python
import requests
import pandas as pd

# Fetch live public currency exchange data endpoint
url = "https://api.exchangerate-api.com/v4/latest/USD"
response = requests.get(url)

# Check for a successful HTTP response
if response.status_code == 200:
    data = response.json()
    
    # Extract rates dictionary and convert to DataFrame
    df = pd.DataFrame(list(data['rates'].items()), columns=['Currency', 'Rate'])
    
    # Display the first few rows
    print(df.head())
    assert not df.empty
    assert 'Currency' in df.columns and 'Rate' in df.columns
else:
    print(f"Error fetching data: {response.status_code}")
````

**Expected output**

```text
  Currency     Rate
0      AED   3.6725
1      AFN  71.2150
2      ALL  96.5000
3      AMD  387.200
4      ANG   1.7900
```

## 2. Handling Nested JSON Structures

Real-world financial and economic APIs often return multi-level nested dictionaries or lists. Standard `pd.DataFrame(data)` will fail or produce messy columns when faced with nested structures. Use `pd.json_normalize()` to flatten them automatically.

````python
# Example parsing a nested JSON payload
nested_data = [
    {"market": {"region": "US", "code": "NYSE"}, "asset": "AAPL", "price": 185.5},
    {"market": {"region": "US", "code": "NASDAQ"}, "asset": "MSFT", "price": 420.2}
]

df_flat = pd.json_normalize(nested_data)
print(df_flat)
assert "market.region" in df_flat.columns
assert df_flat.loc[0, "asset"] == "AAPL"
````

**Expected output**

```text
  market.region market.code asset  price
0            US        NYSE  AAPL  185.5
1            US      NASDAQ  MSFT  420.2
```

## 3. HTTP Error Handling & Robustness

Production scripts must handle network failures, timeouts, and malformed API responses gracefully using try-except blocks.

````python
url = "https://api.exchangerate-api.com/v4/latest/USD"

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()
    data = response.json()
    df = pd.DataFrame(list(data['rates'].items()), columns=['Currency', 'Rate'])
    print("Successfully fetched and validated records:", len(df))
    assert len(df) > 0
except requests.exceptions.HTTPError as err:
    print(f"HTTP error occurred: {err}")
except requests.exceptions.Timeout:
    print("The request timed out.")
except Exception as err:
    print(f"An error occurred: {err}")
````

## 4. Exercise: Fetch and Filter Currency Rates

Create a DataFrame from the USD exchange rates API and filter out all currencies where the exchange rate is greater than 100. Validate that the resulting DataFrame only contains rates > 100.

````python
# Exercise Solution
response = requests.get("https://api.exchangerate-api.com/v4/latest/USD")
data = response.json()
df = pd.DataFrame(list(data['rates'].items()), columns=['Currency', 'Rate'])

# Filter for high exchange rate currencies
high_rates = df[df['Rate'] > 100]
print(high_rates.head())
assert (high_rates['Rate'] > 100).all()
````
