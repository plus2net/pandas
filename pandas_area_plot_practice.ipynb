## 1. Basic Stacked Area Plot

By default, `df.plot.area()` stacks numeric columns on top of each other (`stacked=True`). Each layer represents the cumulative sum.

````python
import pandas as pd
import matplotlib.pyplot as plt

my_dict = {
    "Sale": [80, 90, 75, 80, 85, 80, 75],
    "Cost": [50, 40, 55, 45, 65, 60, 55],
    "Profit": [8, 12, 13, 10, 9, 12, 13]
}
df = pd.DataFrame(data=my_dict)
print(df)

df.plot.area(title="Area Plot: Sales, Cost and Profit", figsize=(7, 4))
plt.show()
assert len(df) == 7
assert df["Sale"].sum() == 565
````

**Expected output**

```text
   Sale  Cost  Profit
0    80    50       8
1    90    40      12
2    75    55      13
3    80    45      10
4    85    65       9
5    80    60      12
6    75    55      13
```

## 2. Unstacked Area Plot with Transparency

Pass `stacked=False` with `alpha=0.4` so layers share a baseline without occluding each other.

````python
df.plot.area(stacked=False, alpha=0.4, title="Unstacked Area Comparison", figsize=(7, 4))
plt.show()
assert df["Profit"].sum() == 77
````

## 3. Custom Color Palettes and Styling

Define custom color tuples to control series fills explicitly.

````python
my_colors = [(0.9, 0.4, 0.3), (0.1, 0.8, 0.1), (0.1, 0.5, 0.9)]
df.plot.area(title="Custom Palette Area Plot", color=my_colors, figsize=(6, 3))
plt.show()
assert len(my_colors) == 3
````

## 4. Practical Workflow: Department Expense Accumulation

Track quarterly expenditures across Engineering, Marketing, and Operations with explicit totals and assertions.

````python
dept_exp = pd.DataFrame({
    "Quarter": ["Q1", "Q2", "Q3", "Q4"],
    "Engineering": [45000, 52000, 58000, 64000],
    "Marketing": [20000, 28000, 35000, 42000],
    "Operations": [15000, 16000, 17500, 19000]
})

assert len(dept_exp) == 4
total_expenses = dept_exp[["Engineering", "Marketing", "Operations"]].sum().sum()
assert total_expenses == 411500
print("Total Operating Expenses:", total_expenses)

ax = dept_exp.plot.area(
    x="Quarter",
    y=["Engineering", "Marketing", "Operations"],
    title="Annual Cumulative Department Expenses",
    figsize=(7, 4),
    grid=True
)
ax.set_ylabel("Expenditure ($)")
plt.show()
````

**Expected output**

```text
Total Operating Expenses: 411500
```

## Exercise: Multi-Channel Revenue Accumulation

Create a DataFrame with 4 months (`Jan`, `Feb`, `Mar`, `Apr`) and three revenue streams: `Hardware` (`[12000, 15000, 18000, 22000]`), `Software` (`[8000, 9500, 11000, 14000]`), and `Services` (`[5000, 6000, 7500, 9000]`). Plot a stacked area chart with Month on the X-axis, enable gridlines, title the plot *"Q1-Q2 Revenue Streams"*, and calculate total revenue across all streams.

## Exercise Solution

Verify your implementation with assertions below:

````python
streams = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Hardware": [12000, 15000, 18000, 22000],
    "Software": [8000, 9500, 11000, 14000],
    "Services": [5000, 6000, 7500, 9000]
})

total_rev = streams[["Hardware", "Software", "Services"]].sum().sum()
assert total_rev == 137000
print("Total Revenue:", total_rev)

streams.plot.area(
    x="Month",
    y=["Hardware", "Software", "Services"],
    title="Q1-Q2 Revenue Streams",
    grid=True,
    figsize=(6, 3)
)
plt.show()
````

**Expected output**

```text
Total Revenue: 137000
```
