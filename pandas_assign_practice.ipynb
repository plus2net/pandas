# Pandas: Assign New Columns with assign()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-assign.php](https://www.plus2net.com/python/pandas-assign.php)

Run cells in order in Google Colab.

## Create numerical sample dataset

```python
import pandas as pd

df = pd.DataFrame({
    "PRODUCT": ["Pen", "Pencil", "Notebook", "Eraser"],
    "QTY": [10, 25, 5, 40],
    "PRICE": [15.0, 5.0, 80.0, 3.0]
})
print(df)
assert df.shape == (4, 3)
```

## Basic column assignment

```python
df_total = df.assign(TOTAL_VALUE=df["QTY"] * df["PRICE"])
print(df_total)
assert df_total.loc[0, "TOTAL_VALUE"] == 150.0 and "TOTAL_VALUE" not in df.columns
```

## Method chaining with lambda functions

```python
df_chained = (
    df.query("QTY > 8")
      .assign(TOTAL=lambda x: x["QTY"] * x["PRICE"])
      .assign(TAX=lambda x: x["TOTAL"] * 0.1)
)
print(df_chained)
assert df_chained.loc[0, "TAX"] == 15.0
```

## Assigning multiple columns sequentially

```python
df_multi = df.assign(
    DISCOUNT_PRICE=lambda x: x["PRICE"] * 0.9,
    FINAL_TOTAL=lambda x: x["QTY"] * x["DISCOUNT_PRICE"]
)
print(df_multi)
assert df_multi.loc[0, "FINAL_TOTAL"] == 135.0
```

## Exercise solution

```python
ex_df = pd.DataFrame({"MARKS": [80, 90, 70]})
answer = ex_df.assign(PERCENTAGE=lambda x: (x["MARKS"] / 100.0) * 100)
print(answer)
assert answer["PERCENTAGE"].max() == 90.0
```
