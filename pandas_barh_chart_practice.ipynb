## 1. Basic Horizontal Bar Chart

`df.plot.barh()` places categories along the vertical Y-axis and values along the horizontal X-axis.

````python
import pandas as pd
import matplotlib.pyplot as plt

my_dict = {
    "NAME": ["Ravi", "Raju", "Alex", "Ron", "Geek", "Kim"],
    "MARK": [20, 30, 40, 30, 30, 30]
}
df = pd.DataFrame(data=my_dict)
print(df)

df.plot.barh(x="NAME", y="MARK", title="Student Marks", figsize=(6, 3.5))
plt.tight_layout()
plt.show()
assert len(df) == 6
assert df["MARK"].sum() == 180
````

**Expected output**

```text
   NAME  MARK
0  Ravi    20
1  Raju    30
2  Alex    40
3   Ron    30
4  Geek    30
5   Kim    30
```

## 2. Sorting Leaderboards Ascending (Top Performer on Top)

In Matplotlib, row 0 starts at the bottom. Sort ascending with `sort_values(ascending=True)` to place the highest score at the top.

````python
sorted_df = df.sort_values(by="MARK", ascending=True)
ax = sorted_df.plot.barh(x="NAME", y="MARK", title="Ranked Student Marks", figsize=(6, 3.5), color="cornflowerblue", legend=False)
ax.bar_label(ax.containers[0], padding=3)
ax.set_xlim(0, 50)
plt.tight_layout()
plt.show()
assert sorted_df.iloc[-1]["NAME"] == "Alex"
assert sorted_df.iloc[-1]["MARK"] == 40
````

## 3. Practical Workflow: Sales Performance Leaderboard

Calculate total revenue, sort the team, and create a stacked horizontal bar chart with assertions.

````python
sales_team = pd.DataFrame({
    "Representative": ["Alice", "Bob", "Charlie", "Diana", "Evan"],
    "Q1_Sales": [25000, 18000, 32000, 29000, 15000],
    "Q2_Sales": [27000, 21000, 34000, 31000, 19000]
})

assert len(sales_team) == 5
assert sales_team["Q1_Sales"].sum() == 119000
assert sales_team["Q2_Sales"].sum() == 132000

sales_team["Total_Sales"] = sales_team["Q1_Sales"] + sales_team["Q2_Sales"]
ranked = sales_team.sort_values(by="Total_Sales", ascending=True)

print("Top Sales Rep:", ranked.iloc[-1]["Representative"])
print("Top Total Sales:", ranked.iloc[-1]["Total_Sales"])
assert ranked.iloc[-1]["Representative"] == "Charlie"
assert ranked.iloc[-1]["Total_Sales"] == 66000

ax = ranked.plot.barh(x="Representative", y=["Q1_Sales", "Q2_Sales"], stacked=True, figsize=(7, 4), title="H1 Sales Leaderboard")
ax.set_xlabel("Revenue ($)")
plt.tight_layout()
plt.show()
````

**Expected output**

```text
Top Sales Rep: Charlie
Top Total Sales: 66000
```

## Exercise: Warehouse Inventory Stock Levels

Create a DataFrame with 5 products and available inventory counts: `Wireless Mouse` (120), `Mechanical Keyboard` (45), `USB-C Dock` (80), `Webcam 4K` (35), and `Noise-Cancelling Headphones` (60). Sort the items in ascending order so the most stocked item appears at the top of the horizontal bar chart, title the chart *"Warehouse Stock Distribution"*, and calculate the total inventory unit count.

## Exercise Solution

Verify your implementation with assertions below:

````python
inventory = pd.DataFrame({
    "Product": ["Wireless Mouse", "Mechanical Keyboard", "USB-C Dock", "Webcam 4K", "Noise-Cancelling Headphones"],
    "Stock": [120, 45, 80, 35, 60]
})

total_units = inventory["Stock"].sum()
assert total_units == 340
print("Total Stock Units:", total_units)

sorted_inv = inventory.sort_values(by="Stock", ascending=True)
ax = sorted_inv.plot.barh(x="Product", y="Stock", title="Warehouse Stock Distribution", figsize=(7, 3.5), legend=False, color="steelblue")
ax.bar_label(ax.containers[0], padding=3)
ax.set_xlim(0, 140)
plt.tight_layout()
plt.show()
````

**Expected output**

```text
Total Stock Units: 340
```
