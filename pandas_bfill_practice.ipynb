## 1. Basic Column Backward Fill

Propagate next valid observations backward up columns.

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({'A': [np.nan, 2, 3], 'B': [4, np.nan, 6]})
result = df.bfill()
print(result)

# Self-validating assertions
assert result.loc[0, 'A'] == 2.0
assert result.loc[1, 'B'] == 6.0
````

**Expected output**

```text
     A    B
0  2.0  4.0
1  2.0  6.0
2  3.0  6.0
```

## 2. Backward Fill with `limit=2`

Restrict backward filling to maximum 2 consecutive missing rows.

````python
df_lab = pd.DataFrame({
    'READING': [np.nan, np.nan, np.nan, 50.0, 55.0]
})

df_capped = df_lab.bfill(limit=2)
print(df_capped)

# Self-validating assertions
assert pd.isna(df_capped.loc[0, 'READING'])
assert df_capped.loc[1, 'READING'] == 50.0
````

**Expected output**

```text
   READING
0      NaN
1     50.0
2     50.0
3     50.0
4     55.0
```
