## 1. Basic Backward Fill

Propagate valid observations backward up columns.

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({'A': [np.nan, 2, 3], 'B': [4, np.nan, 6]})
result = df.bfill()
print(result)

# Self-validating assertion
assert result.loc[0, 'A'] == 2.0
assert result.loc[1, 'B'] == 6.0
````

**Expected output**

```text
     A    B
0  2.0  4.0
1  2.0  6.0
2  3.0  6.0
```
