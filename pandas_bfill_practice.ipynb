# Pandas DataFrame `bfill()` Method Comprehensive Practice

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-bfill.php](https://www.plus2net.com/python/pandas-bfill.php)

## 1. Above-the-Fold Basic Backward Fill

Propagate next valid observations backward up columns.

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'A': [np.nan, 2, 3], 'B': [4, np.nan, 6]})
print(df.bfill())

# Self-validating assertions
result = df.bfill()
assert result.loc[0, 'A'] == 2.0
assert result.loc[1, 'B'] == 6.0
```

## 2. Backward Fill with Consecutive `limit` Caps

Restrict backward filling to maximum consecutive missing rows.

```python
df_lab = pd.DataFrame({
    'READING': [np.nan, np.nan, np.nan, 50.0, 55.0]
})

df_capped = df_lab.bfill(limit=2)
print(df_capped)

# Self-validating assertions
assert pd.isna(df_capped.loc[0, 'READING'])
assert df_capped.loc[1, 'READING'] == 50.0
```

## 3. Row-Wise Backward Fill Across Columns (`axis=1`)

Propagate valid values right-to-left across columns.

```python
df_targets = pd.DataFrame({
    'PRELIM': [np.nan, 10],
    'REVIEW': [np.nan, 12],
    'FINAL': [25, 15]
})

df_back_row = df_targets.bfill(axis=1)
print(df_back_row)

# Self-validating assertions
assert df_back_row.loc[0, 'PRELIM'] == 25.0
assert df_back_row.loc[1, 'PRELIM'] == 10.0
```

## 4. GroupBy-Isolated Backward Fill

Isolate backward filling within categorical groups.

```python
df_items = pd.DataFrame({
    'CAT': ['A', 'A', 'B', 'B'],
    'VAL': [np.nan, 100.0, np.nan, 200.0]
})

df_items['VAL'] = df_items.groupby('CAT')['VAL'].bfill()
print(df_items)

# Self-validating assertions
assert df_items.loc[0, 'VAL'] == 100.0
assert df_items.loc[2, 'VAL'] == 200.0
```
