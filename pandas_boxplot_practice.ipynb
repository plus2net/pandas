### 1. Basic Box Plot
Create a basic box plot summarizing Math and English exam marks.


```python
import pandas as pd
import matplotlib.pyplot as plt

my_dict = {
    'MATH': [30, 33, 32, 32, 31, 34, 32],
    'ENG': [32, 31, 35, 34, 35, 34, 32]
}
df = pd.DataFrame(data=my_dict)
print(df)

ax = df.boxplot(figsize=(5, 3.5), grid=True)
plt.title("Student Marks Distribution")
plt.show()
```

### 2. Grouped Box Plots with `by`
Compare distributions partitioned by categorical classes.


```python
data = {
    'CLASS': ['A', 'A', 'A', 'B', 'B', 'B', 'B'],
    'SCORE': [75, 82, 88, 64, 70, 72, 95]
}
df_class = pd.DataFrame(data)

df_class.boxplot(column='SCORE', by='CLASS', figsize=(6, 4))
plt.suptitle("")
plt.title("Test Score Dispersion by Classroom")
plt.show()
```

### 3. Horizontal Box Plot with `vert=False`
Display boxes horizontally with custom labels.


```python
df.plot.box(vert=False, figsize=(6, 3), title="Horizontal Score Distribution")
plt.xlabel("Scores")
plt.show()
```

### 4. Practical Workflow: Departmental Salary Distribution
Inspect salary dispersion across departments and identify outliers.


```python
import numpy as np

np.random.seed(42)
departments = ["Engineering"] * 25 + ["Marketing"] * 25 + ["Finance"] * 25
salaries = (
    list(np.random.normal(95000, 12000, 25)) +
    list(np.random.normal(72000, 8000, 25)) +
    list(np.random.normal(84000, 10000, 25))
)
salaries[0] = 145000  # Outlier

df_salaries = pd.DataFrame({
    "Department": departments,
    "Salary": np.round(salaries, -2)
})

summary = df_salaries.groupby("Department")["Salary"].describe()
print(summary[["25%", "50%", "75%", "max"]])

assert len(df_salaries) == 75
assert df_salaries["Salary"].max() == 145000.0

df_salaries.boxplot(column="Salary", by="Department", grid=True, rot=15)
plt.suptitle("")
plt.title("Salary Distribution by Department")
plt.ylabel("Annual Salary (USD)")
plt.show()
```

### 5. Hands-on Exercise: Exam Score IQR & Whiskers
Calculate quartiles and plot horizontal box plots for 3 subjects.


```python
exam_data = {
    "Physics": [58, 64, 72, 75, 78, 82, 85, 98],
    "Chemistry": [62, 68, 70, 74, 76, 80, 84, 88],
    "Biology": [70, 75, 79, 81, 84, 86, 89, 92]
}
df_exams = pd.DataFrame(exam_data)

q1_phys = df_exams["Physics"].quantile(0.25)
q3_phys = df_exams["Physics"].quantile(0.75)
iqr_phys = q3_phys - q1_phys

print(f"Physics Q1: {q1_phys:.1f}")
print(f"Physics Q3: {q3_phys:.1f}")
print(f"Physics IQR: {iqr_phys:.1f}")

assert q1_phys == 70.0
assert q3_phys == 82.75
assert iqr_phys == 12.75

df_exams.plot.box(vert=False, grid=True, figsize=(6.5, 3.5), title="Science Exam Spread")
plt.xlabel("Score out of 100")
plt.show()
```
