# Pandas: Trim Outliers with clip()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-clip.php](https://www.plus2net.com/python/pandas-clip.php)

## Create numerical sample dataset with outliers

```python
import pandas as pd

prices = pd.Series([10, 45, 120, 50, 500], name="PRICE")
print(prices)
assert prices.max() == 500
```

## Clipping with lower and upper bounds

```python
clipped = prices.clip(lower=20, upper=100)
print(clipped)
assert clipped.max() == 100 and clipped.min() == 20
```

## Clipping with a single boundary

```python
capped_max = prices.clip(upper=80)
print(capped_max)
assert capped_max.max() == 80
```

## Exercise solution

```python
scores = pd.Series([25, 45, 85, 95, 110], name="SCORE")
answer = scores.clip(lower=30, upper=100)
print(answer)
assert answer.max() == 100 and answer.min() == 30
```
