# Pandas: Merge & Fill Nulls with combine_first()

## Create primary and secondary datasets

```python
import pandas as pd
import numpy as np

df1 = pd.DataFrame({
    "MATH": [85, np.nan, 78],
    "SCIENCE": [np.nan, 92, 80]
}, index=["Ravi", "Raju", "Alex"])

df2 = pd.DataFrame({
    "MATH": [70, 90, 75],
    "SCIENCE": [88, 95, 82]
}, index=["Ravi", "Raju", "Alex"])

print(df1)
assert pd.isna(df1.loc["Raju", "MATH"])
```

## Basic combine_first operation

```python
result = df1.combine_first(df2)
print(result)
assert result.loc["Raju", "MATH"] == 90.0 and result.loc["Ravi", "SCIENCE"] == 88.0
```

## Index alignment and new rows

```python
df3 = pd.DataFrame({"MATH": [95], "SCIENCE": [89]}, index=["John"])
combined = df1.combine_first(df3)
print(combined)
assert "John" in combined.index
```

## Combining Series objects

```python
s1 = pd.Series([10, np.nan, 30], index=["a", "b", "c"])
s2 = pd.Series([15, 25, 35], index=["a", "b", "c"])
s_combined = s1.combine_first(s2)
print(s_combined)
assert s_combined["b"] == 25.0
```

## Exercise solution

```python
s_main = pd.Series([100, np.nan, 300], index=["item1", "item2", "item3"])
s_backup = pd.Series([150, 200, 350], index=["item1", "item2", "item3"])
answer = s_main.combine_first(s_backup)
print(answer)
assert answer["item2"] == 200.0
```
