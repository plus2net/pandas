# Pandas: Calculate Correlation with corr()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-corr.php](https://www.plus2net.com/python/pandas-corr.php)

Run cells in order in Google Colab.

## Create numerical sample DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "HOURS": [2, 4, 6, 8, 10],
    "ATTENDANCE": [50, 60, 75, 90, 95],
    "SCORE": [55, 65, 78, 92, 98]
})
print(df)
assert df.shape == (5, 3)
```

## Basic Pearson correlation matrix

```python
corr_matrix = df.corr()
print(corr_matrix)
assert corr_matrix.loc["HOURS", "SCORE"] > 0.99
```

## Alternative correlation methods (Spearman & Kendall)

```python
spearman_corr = df.corr(method="spearman")
print(spearman_corr)
assert spearman_corr.loc["HOURS", "SCORE"] == 1.0
```

## Exercise solution

```python
ad_df = pd.DataFrame({"AD_SPEND": [10, 20, 30, 40], "SALES": [15, 28, 42, 55]})
answer = ad_df.corr()
print(answer)
assert answer.loc["AD_SPEND", "SALES"] > 0.99
```
