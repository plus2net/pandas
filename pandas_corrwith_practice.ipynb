## 1. Basic Pairwise Correlation

Compute correlation between DataFrame columns and a target baseline Series.

````python
import pandas as pd

df = pd.DataFrame({
    'A': [1, 2, 3, 4],
    'B': [4, 3, 2, 1]
})
target = pd.Series([10, 20, 30, 40])

result = df.corrwith(target)
print(result)

# Self-validating assertion
assert result['A'] == 1.0
assert result['B'] == -1.0
````

**Expected output**

```text
A    1.0
B   -1.0
dtype: float64
```
