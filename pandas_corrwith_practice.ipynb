## 1. Above-the-Fold Basic Pairwise Correlation

Compute correlation between DataFrame columns and a target baseline Series.

```python
import pandas as pd

df = pd.DataFrame({
    'A': [1, 2, 3, 4],
    'B': [4, 3, 2, 1]
})
target = pd.Series([10, 20, 30, 40])

result = df.corrwith(target)
print(result)

# Self-validating assertion
assert result['A'] == 1.0
assert result['B'] == -1.0
```

## 2. Spearman Rank Correlation against Target Series

Use non-parametric rank correlation (`method='spearman'`) to evaluate non-linear monotonic relationships.

```python
df_features = pd.DataFrame({
    'Spend': [100, 200, 500, 1000],
    'Clicks': [10, 25, 40, 45]
})
y_sales = pd.Series([15, 30, 75, 150])

corr_spearman = df_features.corrwith(y_sales, method='spearman')
print(corr_spearman)

# Self-validating assertion
assert corr_spearman['Spend'] == 1.0
```
