# Pandas: Calculate Covariance with cov()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-cov.php](https://www.plus2net.com/python/pandas-cov.php)

Run cells in order in Google Colab.

## Create numerical sample DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "ASSET_A": [10, 20, 30, 40, 50],
    "ASSET_B": [12, 24, 33, 45, 57],
    "VOLUME": [100, 150, 130, 180, 200]
})
print(df)
assert df.shape == (5, 3)
```

## Basic covariance matrix

```python
cov_matrix = df.cov()
print(cov_matrix)
assert cov_matrix.loc["ASSET_A", "ASSET_B"] == 277.5
```

## Covariance between two Series

```python
s_cov = df["ASSET_A"].cov(df["ASSET_B"])
print(f"Covariance: {s_cov}")
assert s_cov == 277.5
```

## Exercise solution

```python
ret_x = pd.Series([1, 2, 3, 4, 5])
ret_y = pd.Series([5, 4, 3, 2, 1])
answer = ret_x.cov(ret_y)
print(answer)
assert answer == -2.5
```
