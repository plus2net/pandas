# Pandas: Compute Cross Tabulations with pd.crosstab()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-crosstab.php](https://www.plus2net.com/python/pandas-crosstab.php)

Run cells in order in Google Colab.

## Create categorical sample dataset

```python
import pandas as pd
df = pd.DataFrame({
    "SECTION": ["A", "A", "A", "B", "B", "B", "B"],
    "RESULT": ["Pass", "Pass", "Fail", "Pass", "Fail", "Fail", "Pass"],
    "SCORE": [85, 90, 40, 78, 45, 35, 88]
})
print(df)
assert df.shape == (7, 3)
```

## Basic frequency cross-tabulation

```python
ct = pd.crosstab(df["SECTION"], df["RESULT"])
print(ct)
assert ct.loc["A", "Pass"] == 2
```

## Adding row and column totals (margins)

```python
ct_margins = pd.crosstab(df["SECTION"], df["RESULT"], margins=True, margins_name="Total")
print(ct_margins)
assert ct_margins.loc["Total", "Total"] == 7
```

## Calculating proportions and percentages

```python
ct_norm = pd.crosstab(df["SECTION"], df["RESULT"], normalize="index") * 100
print(ct_norm.round(1))
assert round(ct_norm.loc["B", "Pass"], 1) == 50.0
```

## Applying aggregation functions with values

```python
ct_mean = pd.crosstab(df["SECTION"], df["RESULT"], values=df["SCORE"], aggfunc="mean")
print(ct_mean)
assert ct_mean.loc["A", "Pass"] == 87.5
```

## Exercise solution

```python
gender = pd.Series(["M", "F", "M", "F", "F"], name="GENDER")
grade = pd.Series(["A", "B", "A", "A", "B"], name="GRADE")
answer = pd.crosstab(gender, grade, margins=True)
print(answer)
assert answer.loc["All", "All"] == 5
```
