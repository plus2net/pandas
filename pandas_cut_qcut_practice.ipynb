# Pandas: Bin Continuous Numerical Data with cut() & qcut()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-cut-qcut.php](https://www.plus2net.com/python/pandas-cut-qcut.php)

Run cells in order in Google Colab.

## Create numerical sample dataset

```python
import pandas as pd
df = pd.DataFrame({
    "NAME": ["Ravi", "Raju", "Alex", "John", "Sara", "Priya"],
    "SCORE": [35, 52, 68, 74, 88, 95]
})
print(df)
assert df.shape == (6, 2)
```

## Equal-width binning with pd.cut()

```python
df["WIDTH_BIN"] = pd.cut(df["SCORE"], bins=3)
print(df[["NAME", "SCORE", "WIDTH_BIN"]])
assert df["WIDTH_BIN"].nunique() == 3
```

## Custom bin boundaries and labels

```python
bins = [0, 50, 75, 100]
labels = ["Fail", "Average", "Excellent"]
df["GRADE"] = pd.cut(df["SCORE"], bins=bins, labels=labels)
print(df[["NAME", "SCORE", "GRADE"]])
assert df.loc[0, "GRADE"] == "Fail" and df.loc[5, "GRADE"] == "Excellent"
```

## Equal-frequency binning with pd.qcut()

```python
df["TERTILE"] = pd.qcut(df["SCORE"], q=3, labels=["Low", "Medium", "High"])
print(df[["NAME", "SCORE", "TERTILE"]])
assert df["TERTILE"].value_counts().tolist() == [2, 2, 2]
```

## Exercise solution

```python
ages = pd.Series([18, 25, 38, 45, 62, 70], name="AGE")
answer = pd.qcut(ages, q=3, labels=["Young", "Middle", "Senior"])
print(answer)
assert answer.value_counts().tolist() == [2, 2, 2]
```
