## Workflow: inspect a synthetic order extract

Each order has a date, product, status and amount. The sample includes a test order, an unparseable amount and a missing amount. These are practice records, not measured business performance.

````python
import pandas as pd
raw = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "date": ["2026-09-01", "2026-09-01", "2026-09-02", "2026-09-02", "2026-09-03", "2026-09-03", "2026-09-04", "2026-09-04"],
    "product": ["Pen", "Book", "Pen", "Notebook", "Book", "Pen", "Notebook", "Book"],
    "status": ["paid", "paid", "paid", "test", "paid", "paid", "paid", "paid"],
    "amount": ["10", "30", "15", "0", "bad", None, "50", "20"]
})
print(raw.head())
print("Rows and columns:", raw.shape)
raw.info()
assert raw["order_id"].is_unique
````

## Define the population and keep excluded rows

Parse dates using the source format and amounts as numbers. Report paid orders with known non-negative amounts. A zero amount is allowed by this sample rule. Do not treat an unknown amount as zero.

````python
cleaned = raw.copy()
cleaned["date"] = pd.to_datetime(raw["date"], format="%Y-%m-%d", errors="coerce")
cleaned["amount"] = pd.to_numeric(raw["amount"], errors="coerce").astype("Float64")
valid = cleaned["status"].eq("paid") & cleaned["date"].notna() & cleaned["amount"].ge(0).fillna(False)
analysis = cleaned.loc[valid].copy()
review = raw.loc[~valid].copy()
print("Analysis orders:")
print(analysis)
print("Excluded original records:")
print(review)
assert analysis["order_id"].tolist() == [101, 102, 103, 107, 108]
assert len(analysis) + len(review) == len(raw)
````

## Calculate the overall summary

Select the amount measure rather than averaging every numeric field. The accepted sample has five orders and known revenue 125. Describe and a small named set of measures provide complementary views.

````python
print(analysis["amount"].describe())
metrics = analysis["amount"].agg(["count", "sum", "mean", "median", "min", "max"])
print(metrics)
assert metrics["count"] == 5
assert metrics["sum"] == 125
assert metrics["mean"] == 25
assert metrics["median"] == 20
````

## Summarise products and reconcile the totals

Count orders with size and available amounts with count. They match here because every accepted order has a known amount. Check that grouped revenue sums to the same total as the analysis table.

````python
product_summary = analysis.groupby("product", dropna=False).agg(orders=("order_id", "size"), known_amounts=("amount", "count"), revenue=("amount", "sum"), mean_order=("amount", "mean"))
product_summary = product_summary.sort_values("revenue", ascending=False)
print(product_summary)
print("Order frequency:")
print(analysis["product"].value_counts())
assert product_summary["orders"].sum() == 5
assert product_summary["revenue"].sum() == analysis["amount"].sum()
````

## Create a fixed band report with boundary checks

Use Low: [0, 20], Medium: (20, 40] and High: (40, 100]. The amount 20 belongs to Low. Retain empty defined categories with observed=False if the report requires a consistent set of bands.

````python
analysis["amount_band"] = pd.cut(analysis["amount"].astype(float), [0, 20, 40, 100], labels=["Low", "Medium", "High"], include_lowest=True)
band_summary = analysis.groupby("amount_band", observed=False).agg(orders=("order_id", "size"), revenue=("amount", "sum"))
print(band_summary)
assert band_summary["orders"].tolist() == [3, 1, 1]
assert analysis.loc[analysis["order_id"].eq(108), "amount_band"].iloc[0] == "Low"
assert analysis["amount_band"].notna().all()
````

**Expected output**

```text

Medium            1     30.0
High              1     50.0
```

## Compare daily totals and moving averages

The accepted sample has observations on September 1, 2 and 4. The full sample confirms no accepted orders on September 3, so the daily accepted-order report uses zero for that day. In a real extract, confirm coverage before interpreting an empty day as zero. A two-day rolling mean then includes that zero.

````python
daily = analysis.set_index("date")["amount"].sort_index().resample("D").sum(min_count=1)
daily_report = daily.fillna(0).rename("revenue").to_frame()
daily_report["two_day_mean"] = daily_report["revenue"].rolling(2, min_periods=2).mean()
print(daily_report)
assert daily_report["revenue"].tolist() == [40, 15, 0, 70]
assert daily_report["revenue"].sum() == 125
assert daily_report["two_day_mean"].iloc[-1] == 35
````

## Reshape the same accepted population

A pivot table aggregates product revenue by date. Because accepted values are known, an absent product-date combination is displayed as zero in this report. Verify that the reshaped total agrees with the source.

````python
matrix = analysis.pivot_table(index="date", columns="product", values="amount", aggfunc="sum", fill_value=0, observed=True)
print(matrix)
assert matrix.to_numpy().sum() == 125
````

## Exercise: add an order and predict the report

Add a paid Pen order dated September 4 with amount 5. Predict the accepted order count, revenue, mean amount and Pen revenue before running the solution. Do not edit the existing raw table in place.

## Exercise solution

The extra order changes the population, so recompute every dependent summary from the extended data.

````python
extra = pd.DataFrame({"order_id": [109], "date": ["2026-09-04"], "product": ["Pen"], "status": ["paid"], "amount": ["5"]})
extended = pd.concat([raw, extra], ignore_index=True)
extended["date"] = pd.to_datetime(extended["date"], format="%Y-%m-%d", errors="coerce")
extended["amount"] = pd.to_numeric(extended["amount"], errors="coerce").astype("Float64")
keep = extended["status"].eq("paid") & extended["date"].notna() & extended["amount"].ge(0).fillna(False)
answer = extended.loc[keep].copy()
print("Accepted orders:", len(answer))
print("Revenue:", answer["amount"].sum())
print("Mean amount:", round(float(answer["amount"].mean()), 2))
print("Pen revenue:", answer.loc[answer["product"].eq("Pen"), "amount"].sum())
assert len(answer) == 6
assert answer["amount"].sum() == 130
assert answer.loc[answer["product"].eq("Pen"), "amount"].sum() == 30
assert len(raw) == 8
````

**Expected output**

```text
Accepted orders: 6
Revenue: 130.0
Mean amount: 21.67
Pen revenue: 30.0
```

## Check the interpretation before sharing results

<strong>Are counts and revenue interchangeable?</strong> No; report units explicitly. <strong>Should missing amounts become zero?</strong> Only with evidence that they mean zero; keep uncertain amounts in review. <strong>Why did totals change after a merge?</strong> Inspect repeated join keys and validate the expected cardinality. <strong>Why does a mean exclude some records?</strong> Many reductions skip missing values; report their available count. <strong>Can an outlier be deleted automatically?</strong> Investigate the source and define a rule before excluding it. <strong>Does a chart prove a relationship?</strong> It describes the selected data; causal conclusions require more evidence.
