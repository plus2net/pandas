## Workflow: inspect a synthetic order extract

This sample has text variants, blank products, an invalid amount and a repeated order. The rules are: a non-blank product, a non-negative integer amount and one record per order ID. Keep the raw data for review.

````python
import pandas as pd
raw = pd.DataFrame({
    "order_id": [101, 102, 102, 103, 104, 105],
    "product": [" Notebook ", "PEN", "PEN", "  ", None, "Café mug"],
    "amount": ["1,200", "50", "50", "bad", "70", "90"]
})
df = raw.copy()
print(df)
print("Present values:")
print(df.count())
print("Types:")
print(df.dtypes)
````

**Expected output**

```text
   order_id     product amount
0       101   Notebook   1,200
1       102         PEN     50
2       102         PEN     50
3       103                bad
4       104         NaN     70
5       105    Café mug     90
Present values:
order_id    6
product     5
amount      6
dtype: int64
Types:
order_id    int64
product       str
amount        str
dtype: object
```

## Normalise text without losing the source

Trim surrounding spaces, treat blanks as missing and create a separate case-normalised comparison key. Keep the readable product spelling.

````python
product = df["product"].astype("string").str.strip()
df["product"] = product.mask(product.eq(""), pd.NA)
df["product_key"] = df["product"].str.casefold()
print(df[["product", "product_key"]])
assert df["product"].isna().sum() == 2
assert df.loc[5, "product"] == "Café mug"
````

**Expected output**

```text
    product product_key
0  Notebook    notebook
1       PEN         pen
2       PEN         pen
3      <NA>        <NA>
4      <NA>        <NA>
5  Café mug    café mug
```

## Remove grouping commas and validate integer amounts

The original comma-removal idea is suitable only for a known format. Here commas are thousands separators and only whole, non-negative amounts are accepted; decimal-comma formats need another rule. Vectorised replacement handles missing text without a Python lambda.

````python
amount_text = df["amount"].astype("string").str.replace(",", "", regex=False)
numbers = pd.to_numeric(amount_text, errors="coerce")
amount_ok = (numbers.notna() & numbers.ge(0) & numbers.mod(1).eq(0)).fillna(False)
df["amount"] = numbers.where(amount_ok).astype("Int64")
print(df[["order_id", "amount"]])
assert df["amount"].isna().sum() == 1
assert df.loc[0, "amount"] == 1200
````

**Expected output**

```text
   order_id  amount
0       101    1200
1       102      50
2       102      50
3       103    <NA>
4       104      70
5       105      90
```

## Review missing fields before selecting valid rows

Reason counts can overlap. Do not sum them to calculate rejected rows. This example keeps rejected records separately instead of inventing amounts or product names.

````python
bad_product = df["product"].isna()
bad_amount = df["amount"].isna()
valid = ~(bad_product | bad_amount)
review = df.loc[~valid].copy()
accepted = df.loc[valid].copy()
print("Missing product:", int(bad_product.sum()))
print("Invalid or missing amount:", int(bad_amount.sum()))
print("Rejected rows:", len(review))
print(review)
assert len(review) == 2
````

**Expected output**

```text
Missing product: 2
Invalid or missing amount: 1
Rejected rows: 2
   order_id product  amount product_key
3       103    <NA>    <NA>        <NA>
4       104    <NA>      70        <NA>
```

## Check duplicate keys and retain one consistent record

For each repeated key, inspect whether other values conflict. This sample has identical repeated orders; keeping the first is justified by that check. Conflicting records should be resolved using a business rule.

````python
conflicts = accepted.groupby("order_id")[["product_key", "amount"]].nunique(dropna=False).gt(1).any(axis=1)
assert not conflicts.any(), "Resolve conflicting orders before deduplicating"
print("Repeated keys:")
print(accepted.loc[accepted.duplicated("order_id", keep=False)])
clean = accepted.drop_duplicates(subset=["order_id"], keep="first").copy()
print("Clean orders:")
print(clean)
assert clean["order_id"].tolist() == [101, 102, 105]
````

**Expected output**

```text
Repeated keys:
   order_id product  amount product_key
1       102     PEN      50         pen
2       102     PEN      50         pen
Clean orders:
   order_id   product  amount product_key
0       101  Notebook    1200    notebook
1       102       PEN      50         pen
5       105  Café mug      90    café mug
```

## Validate and export the result

Check the agreed rules after transformation. Compare counts with the raw extract and retain rejected rows for audit. See <a href="https://www.plus2net.com/python/pandas-to_excel.php">Excel export</a> and <a href="https://www.plus2net.com/python/pandas-read_csv.php">CSV input</a>.

````python
assert clean["order_id"].is_unique
assert clean[["product", "amount"]].notna().all().all()
assert clean["amount"].ge(0).all()
print("Raw rows:", len(raw), "Review rows:", len(review), "Clean rows:", len(clean))
print(clean.to_csv(index=False).rstrip())
# Optional file export in Colab:
# clean.to_csv("clean_orders.csv", index=False)
````

## Preserve valid Unicode text

The pattern [^\x00-\x7F] selects non-ASCII characters, not invalid UTF-8. Removing it globally can damage names and international text. If an external system explicitly requires ASCII, isolate that conversion, keep the source and inspect the loss.

````python
labels = pd.Series(["Café", "Notebook", "東京"], dtype="string")
ascii_only = labels.str.replace(r"[^\x00-\x7F]+", "", regex=True)
print(pd.DataFrame({"original": labels, "ASCII-only demonstration": ascii_only}))
assert labels.iloc[0] == "Café"
````

**Expected output**

```text
   original ASCII-only demonstration
0      Café                      Caf
1  Notebook                 Notebook
2        東京
```
