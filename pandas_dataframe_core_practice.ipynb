# Pandas DataFrame: Architecture, Construction & Core Operations

Interactive companion notebook for the Plus2Net [Pandas DataFrame Core Guide](https://www.plus2net.com/python/pandas-dataframe.php). Run cells in order in Google Colab. All sample data is embedded directly in this notebook.

## 1. Creating DataFrames from Dictionaries & NumPy 2D Arrays

```python
import pandas as pd
import numpy as np

# Construct from dictionary of equal-length lists
student_dict = {
    "NAME": ["Ravi", "Raju", "Alex"],
    "ID": [1, 2, 3],
    "MATH": [30, 40, 50],
    "ENGLISH": [20, 30, 40]
}
df_students = pd.DataFrame(data=student_dict)
print("Students DataFrame:")
print(df_students)
assert df_students.shape == (3, 4)

# Construct from NumPy 2D array
matrix = np.array([[10, 20], [30, 40], [50, 60]])
df_from_np = pd.DataFrame(matrix, columns=["Metric_A", "Metric_B"])
print("From NumPy Array:")
print(df_from_np)
assert df_from_np.shape == (3, 2)
```

## 2. Ingesting Data from String Text & io.StringIO

```python
from io import StringIO

raw_csv = StringIO("""col1;col2;col3
1;5.40;Geek
2;43.25;Ravi
3;41.70;Ron
4;34.20;Alex
""")

df_string = pd.read_csv(raw_csv, sep=";")
print("DataFrame from StringIO:")
print(df_string)
assert df_string.shape == (4, 3)
assert df_string["col2"].sum() == 124.55
```

## 3. Managing Row Indices: Numeric, String & set_index()

```python
# Assign custom numeric index
df_students.index = [101, 102, 103]
print("Custom Numeric Index:")
print(df_students)
assert list(df_students.index) == [101, 102, 103]

# Set existing column as index
df_indexed = df_students.set_index("ID")
print("ID Column Set as Index:")
print(df_indexed)
assert df_indexed.index.name == "ID"
```

## 4. Column Manipulation & Subset Selection

```python
# Select specific columns
subset = df_students[["NAME", "MATH"]]
print("Selected Subset (NAME & MATH):")
print(subset)
assert subset.shape == (3, 2)

# Dynamic exclusion (all columns except ID)
filtered_cols = [c for c in df_students.columns if c != "ID"]
print("Excluded ID Column:")
print(df_students[filtered_cols])
assert "ID" not in filtered_cols
```

## 5. Essential DataFrame Attributes

```python
print("Shape:", df_students.shape)
print("Data Types:")
print(df_students.dtypes)
print("Total Elements:", df_students.size)
print("Dimensions:", df_students.ndim)
assert df_students.size == 12
assert df_students.ndim == 2
```

## 6. Hands-on Exercise: Product Inventory Tracker

```python
inventory_data = {
    "sku": ["SKU-101", "SKU-102", "SKU-103", "SKU-104"],
    "product": ["Wireless Mouse", "Mechanical Keyboard", "USB Hub", "Monitor Stand"],
    "category": ["Accessories", "Peripherals", "Accessories", "Furniture"],
    "price": [24.99, 89.50, 15.00, 65.00],
    "stock": [50, 20, 100, 15]
}

df_inv = pd.DataFrame(inventory_data).set_index("sku")
df_inv["total_val"] = df_inv["price"] * df_inv["stock"]

premium_items = df_inv[df_inv["price"] > 50]
print("Premium Products (Price > $50):")
print(premium_items[["product", "price", "total_val"]])

assert len(premium_items) == 2
assert "Mechanical Keyboard" in premium_items["product"].values
assert round(df_inv["total_val"].sum(), 2) == 5514.50
print(f"Total Catalog Value: ${df_inv['total_val'].sum():.2f}")
```
