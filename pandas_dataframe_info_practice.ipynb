## Print a DataFrame summary

Call <code>df.info()</code> directly. The summary describes the index, columns, non-missing counts, dtypes and estimated memory use. It does not calculate means or percentiles; use <a href="https://www.plus2net.com/python/pandas-dataframe-describe.php">describe()</a> for those statistics.

````python
import pandas as pd
import io

def student_data():
    return pd.DataFrame({
        "NAME": ["Ravi", "Raju", "Alex"],
        "ID": [1, 2, 3],
        "MATH": [30, 40, 50],
        "ENGLISH": [20, 30, 40]
    })

df = student_data()
df.info()
````


## Why does print(df.info()) display None?

The method prints a report and returns None. Wrapping it in <code>print()</code> adds that return value after the report. Store the report through <code>buf</code>, rather than assigning the method result as if it were a DataFrame.

````python
df = student_data()
result = df.info()
print("Return value:", result)
assert result is None
# print(df.info()) also prints an additional None.
````

**Expected output**

## Read non-null counts alongside row counts

Non-null means non-missing, rather than nonzero. Zero scores are valid observations. Here MATH has two known values in three rows. An empty string is also non-missing; clean placeholder text explicitly when appropriate.

````python
df = student_data()
df.loc[1, "MATH"] = float("nan")
df.loc[2, "ENGLISH"] = 0
df.info(show_counts=True)
print("Rows:", len(df))
print("Missing per column:")
print(df.isna().sum())
assert df["MATH"].count() == 2
assert df["ENGLISH"].count() == 3
````

## Choose full or compact output with verbose

## Control truncation with max_cols

When verbose is left at its default, more columns than <code>max_cols</code> causes truncation. Four columns with max_cols=3 produces compact output; max_cols=4 allows full output. Explicit verbose=True can request full details regardless of that automatic threshold.

````python
df = student_data()
print("Four columns, limit three:")
df.info(max_cols=3)
print("Four columns, limit four:")
df.info(max_cols=4)
print("Explicit full output:")
df.info(max_cols=3, verbose=True)
````

## Inspect display options without changing global settings

Default verbosity and count display depend on the max_info_columns and max_info_rows options. Use <code>option_context()</code> for a temporary change; the previous settings are restored after the block. See <a href="https://www.plus2net.com/python/pandas-display-options.php">Pandas display options</a>.

````python
df = student_data()
print("Column threshold:", pd.options.display.max_info_columns)
print("Row threshold:", pd.options.display.max_info_rows)
before = pd.options.display.max_info_columns
with pd.option_context("display.max_info_columns", 2):
    df.info()
assert pd.options.display.max_info_columns == before
````

## Get memory figures as data rather than printed text

Use <code>memory_usage()</code> to obtain a Series of bytes by column and index. For object strings, deep inspection can report more than the shallow estimate. Do not parse the text produced by info() when a structured API already provides the measurements.

````python
texts = pd.DataFrame({"description": pd.Series(["Notebook for classroom use", "Blue pen for office use", "Notebook for classroom use"], dtype="object")})
shallow = texts.memory_usage(index=True, deep=False)
deep = texts.memory_usage(index=True, deep=True)
print("Shallow bytes:")
print(shallow)
print("Deep bytes:")
print(deep)
print("Total deep bytes:", deep.sum())
assert deep.sum() >= shallow.sum()
````

**Expected output**

```text
Shallow bytes:
Index          132
description     24
dtype: int64
Deep bytes:
Index          132
description    222
dtype: int64
Total deep bytes: 354
```

## Capture the report with buf and StringIO

Pass a writable buffer to collect the text. Use its getvalue() method after info() has written the report. You can display it, save it or include it in a diagnostic log.

````python
df = student_data()
buffer = io.StringIO()
df.info(buf=buffer, verbose=True, show_counts=True, memory_usage="deep")
report_text = buffer.getvalue()
print(report_text.rstrip())
assert "MATH" in report_text
# Optional file save in Colab:
# with open("student_info.txt", "w", encoding="utf-8") as file:
#     file.write(report_text)
````

## Practical inspection: a messy sales extract

A non-null count cannot detect invalid numeric text or a missing-value placeholder such as an empty string. Inspect actual values and convert deliberately. The sample is synthetic; it includes blank customers, invalid dates and one invalid amount.

````python
def sales_data():
    return pd.DataFrame({
        "order_id": [101, 102, 103, 104],
        "customer": ["Alex", "", None, "Ravi"],
        "order_date": ["2026-01-01", "2026-01-02", "bad-date", None],
        "amount": ["12.50", "20", "unknown", "0"]
    })

sales = sales_data()
sales.info(show_counts=True)
print(sales)
assert sales["amount"].count() == 4
````

## Compare the summary before and after conversion

Normalize blank customer strings to missing, parse dates with an explicit format and convert amounts to numbers. Coercion changes invalid values to missing; preserve the source table to identify what failed. Explore <a href="https://www.plus2net.com/python/pandas-dt-to_datetime.php">to_datetime()</a> and <a href="https://www.plus2net.com/python/pandas-dataframe-astype.php">astype()</a> for further conversion examples.

````python
raw = sales_data()
clean = raw.copy()
clean["customer"] = clean["customer"].astype("string").str.strip().replace("", pd.NA)
clean["order_date"] = pd.to_datetime(clean["order_date"], format="%Y-%m-%d", errors="coerce")
clean["amount"] = pd.to_numeric(clean["amount"], errors="coerce")
clean.info(verbose=True, show_counts=True)
print("Invalid non-missing amount inputs:")
print(raw.loc[raw["amount"].notna() & clean["amount"].isna(), ["order_id", "amount"]])
print("Missing after cleaning:")
print(clean.isna().sum())
assert clean["amount"].count() == 3
assert clean["customer"].isna().sum() == 2
````

## Build a structured missing-value report

Use count(), isna() and dtypes to create a reusable audit table. This can be exported without parsing info() output. A high missing percentage is a prompt to investigate; it is not an automatic reason to delete a field or fill it with zero. See <a href="https://www.plus2net.com/python/pandas-dataframe-fillna.php">fillna()</a> and <a href="https://www.plus2net.com/python/pandas-dataframe-dropna.php">dropna()</a>.

````python
audit = pd.DataFrame({
    "dtype": clean.dtypes.astype(str),
    "non_missing": clean.count(),
    "missing": clean.isna().sum()
})
audit["missing_percent"] = clean.isna().mean().mul(100).round(1)
print(audit)
assert audit.loc["amount", "missing"] == 1
assert audit.loc["customer", "missing_percent"] == 50
````

**Expected output**

```text

order_id             int64            4        0              0.0
customer            string            2        2             50.0
order_date  datetime64[us]            2        2             50.0
amount             float64            3        1             25.0
```

## Inspect an empty table with an explicit schema

An empty DataFrame can still have columns and declared dtypes. A table with rows containing only missing values is not empty. Check shape and missing counts rather than interpreting empty as meaning all values are missing.

````python
empty = pd.DataFrame({"order_id": pd.Series(dtype="int64"), "amount": pd.Series(dtype="float64")})
empty.info(show_counts=True)
all_missing = pd.DataFrame({"amount": [float("nan")]})
print("Zero-row table is empty:", empty.empty)
print("One missing row is empty:", all_missing.empty)
assert empty.empty and not all_missing.empty
````

**Expected output**

```text
One missing row is empty: False
```

## Common questions and pitfalls

<strong>Why are some counts absent?</strong> Request verbose=True and show_counts=True. <strong>Why did conversion create more missing values?</strong> Inspect invalid inputs when using errors="coerce". <strong>Does info() fix the data?</strong> No; it reports structure. <strong>Does non-null mean valid?</strong> No; blank strings and invalid text may still count. <strong>Why do memory figures differ?</strong> Dtypes, string backends, index storage and platform details affect the values. Use the same environment for meaningful comparisons.

## Exercise: inspect quantities before calculating

Create quantity values "2", "bad", None and "0", paired with order IDs 1 to 4. Convert quantity to numeric with invalid inputs coerced to missing. Print a full info report, calculate missing quantities and identify the order with invalid non-missing input. Expect two missing values; order 2 has the invalid text.

## Exercise solution

Missing input and invalid input both become missing after conversion, but only the invalid non-missing input is a parsing failure. Zero remains a valid quantity.

````python
raw_orders = pd.DataFrame({"order_id": [1, 2, 3, 4], "quantity": ["2", "bad", None, "0"]})
orders = raw_orders.copy()
orders["quantity"] = pd.to_numeric(orders["quantity"], errors="coerce")
orders.info(verbose=True, show_counts=True)
failed = raw_orders["quantity"].notna() & orders["quantity"].isna()
print("Missing quantities:", orders["quantity"].isna().sum())
print("Invalid input order IDs:", raw_orders.loc[failed, "order_id"].tolist())
assert orders["quantity"].isna().sum() == 2
assert raw_orders.loc[failed, "order_id"].tolist() == [2]
assert orders.loc[3, "quantity"] == 0
````

**Expected output**

```text
<class 'pandas.DataFrame'>
RangeIndex: 4 entries, 0 to 3
Data columns (total 2 columns):
 #   Column    Non-Null Count  Dtype  
---  ------    --------------  -----  
 0   order_id  4 non-null      int64  
 1   quantity  2 non-null      float64
dtypes: float64(1), int64(1)

Invalid input order IDs: [2]
```
