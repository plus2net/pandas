<p>When <code>start</code> and <code>end</code> are supplied without <code>periods</code>, the usual default frequency is one calendar day.</p>

<p>At most three of start, end, periods and freq may be supplied. Specify at least two of start, end and periods. With start/end/periods and no freq, points are spaced evenly.</p>

## Basic Date Range Using start and end





<p>The default frequency is daily, represented by <code>D</code>. Both the starting and ending dates are included by default.</p>

<div class="alert alert-light border" role="alert">The displayed datetime resolution such as <code>datetime64[us]</code> or <code>datetime64[ns]</code> can differ between Pandas versions and depending on the supplied values. The generated dates remain the important part of the result.</div>



````python
import pandas as pd

dates=pd.date_range(
    start='2020-04-20',
    end='2020-04-27'
)

print(dates)
````

**Expected output**

```text
DatetimeIndex(['2020-04-20', '2020-04-21', '2020-04-22', '2020-04-23',
               '2020-04-24', '2020-04-25', '2020-04-26', '2020-04-27'],
              dtype='datetime64[us]', freq='D')
```

## Generate a Fixed Number of Dates with periods



<p>Use <code>periods</code> to specify how many timestamps should be generated.</p>

<p><strong>start + periods</strong></p>



<p><strong>end + periods</strong></p>



<p>This generates five daily dates ending on 5 September 2026.</p>

<p><strong>start + end + periods</strong></p>

<p>If <code>start</code>, <code>end</code> and <code>periods</code> are supplied without a frequency, Pandas creates evenly spaced timestamps between the two boundaries.</p>



<p>The timestamps are evenly distributed between the start and end values. They are not automatically restricted to complete days.</p>



````python
import pandas as pd

dates=pd.date_range(
    start='2026-09-01',
    periods=5
)

print(dates)

dates=pd.date_range(
    end='2026-09-05',
    periods=5
)
print(dates)

dates=pd.date_range(
    start='2020-04-20',
    end='2020-04-27',
    periods=3
)

print(dates)
````

## Generate Dates Using Today's Date



<p><code>Timestamp.today()</code> can supply the current local date and time.</p>



<p>This creates exactly ten dates ending with today. Using <code>normalize()</code> changes the current timestamp to midnight before the range is generated.</p>

<p><strong>Next Five Hours from the Current Time</strong></p>



<p>The actual output depends on the time at which the program is run.</p>

<p>These two examples execute the actual local clock calls; dates change on each run. Only counts are shown for repeatable validation.</p>

````python
import pandas as pd

today=pd.Timestamp.today().normalize()

dates=pd.date_range(
    end=today,
    periods=10,
    freq='D'
)

print("Generated timestamps:", len(dates))
assert len(dates) in (5, 10)

import pandas as pd

now=pd.Timestamp.now()

dates=pd.date_range(
    start=now,
    periods=5,
    freq='h'
)

print("Generated timestamps:", len(dates))
assert len(dates) in (5, 10)
````

**Expected output**

```text
Generated timestamps: 10
Generated timestamps: 5
```

## Convert date_range() to a Python List



<p><code>date_range()</code> returns a <code>DatetimeIndex</code>. Use <a href="pandas-dataframe-tolist.php"><code>tolist()</code></a> to convert it to a Python <a href="list.php">list</a>.</p>



<p>The resulting list contains Pandas <code>Timestamp</code> objects.</p>

<p><strong>Convert to Date Strings</strong></p>





````python
import pandas as pd

dates=pd.date_range(
    start='2026-09-01',
    periods=5
)

date_list=dates.tolist()

print(date_list)

date_strings=dates.strftime(
    '%Y-%m-%d'
).tolist()

print(date_strings)
````

**Expected output**

```text
[Timestamp('2026-09-01 00:00:00'), Timestamp('2026-09-02 00:00:00'), Timestamp('2026-09-03 00:00:00'), Timestamp('2026-09-04 00:00:00'), Timestamp('2026-09-05 00:00:00')]

```

## Normalize Time to Midnight



<p>With <code>normalize=True</code>, the start and end values are normalized to midnight before the date range is generated.</p>



<p>Without normalization, the time <code>14:35:20</code> would be retained in each generated timestamp.</p>



````python
import pandas as pd

dates=pd.date_range(
    start='2026-09-05 14:35:20',
    periods=3,
    freq='D',
    normalize=True
)

print(dates)
````

**Expected output**

```text
DatetimeIndex(['2026-09-05', '2026-09-06', '2026-09-07'], dtype='datetime64[us]', freq='D')
```

## Create a Timezone-Aware Date Range with tz



<p>The <code>tz</code> parameter creates a timezone-aware <code>DatetimeIndex</code>.</p>





````python
import pandas as pd

dates=pd.date_range(
    start='2026-09-05 09:00',
    periods=3,
    freq='4h',
    tz='Asia/Kolkata'
)

print(dates)
````

## How Frequency Boundaries Affect the Result



<p>For anchored frequencies such as month start or month end, the starting value itself may not be a valid timestamp for that frequency.</p>

<p>For example:</p>



<p>15 January is not a month-start boundary, so the first valid <code>MS</code> timestamp is 1 February.</p>

<p>Similarly, with <code>ME</code>, Pandas moves to the next valid month-end timestamp when necessary.</p>



````python
dates=pd.date_range(
    start='2026-01-15',
    periods=3,
    freq='MS'
)

print(dates)
````

**Expected output**

```text
DatetimeIndex(['2026-02-01', '2026-03-01', '2026-04-01'], dtype='datetime64[us]', freq='MS')
```

## Common Mistakes with pandas.date_range()



<p><strong>1. Calling DataFrame.date_range()</strong></p>

<p>Incorrect:</p>

<div class="p-3 bg-light border rounded"><code>df.<span class="p2n-function">date_range</span><span class="p2n-bracket">(</span>...<span class="p2n-bracket">)</span><span class="p2n-bracket"></span></code></div>

<p>Correct:</p>

<div class="p-3 bg-light border rounded"><code>pd.<span class="p2n-function">date_range</span><span class="p2n-bracket">(</span>...<span class="p2n-bracket">)</span><span class="p2n-bracket"></span></code></div>

<p><strong>2. Using Old Frequency Aliases</strong></p>

<p>For current Pandas, avoid old code such as:</p>

<div class="p-3 bg-light border rounded"><code>freq=<span class="text-success">&#x27;H&#x27;</span>
freq=<span class="text-success">&#x27;T&#x27;</span>
freq=<span class="text-success">&#x27;M&#x27;</span>
freq=<span class="text-success">&#x27;A-FEB&#x27;</span><span class="p2n-bracket"></span></code></div>

<p>Use:</p>

<div class="p-3 bg-light border rounded"><code>freq=<span class="text-success">&#x27;h&#x27;</span>
freq=<span class="text-success">&#x27;min&#x27;</span>
freq=<span class="text-success">&#x27;ME&#x27;</span>
freq=<span class="text-success">&#x27;YE-FEB&#x27;</span><span class="p2n-bracket"></span></code></div>

<p><strong>3. Using closed Instead of inclusive</strong></p>

<p>Old:</p>

<div class="p-3 bg-light border rounded"><code>closed=<span class="text-success">&#x27;right&#x27;</span><span class="p2n-bracket"></span></code></div>

<p>Current:</p>

<div class="p-3 bg-light border rounded"><code>inclusive=<span class="text-success">&#x27;right&#x27;</span><span class="p2n-bracket"></span></code></div>

<p><strong>4. Expecting periods to Mean Days</strong></p>

<p><code>periods</code> means the number of generated timestamps. Their distance depends on <code>freq</code>.</p>

<div class="p-3 bg-light border rounded"><code>pd.<span class="p2n-function">date_range</span><span class="p2n-bracket">(</span>
    start=<span class="text-success">&#x27;2026-09-01&#x27;</span>,
    periods=<span class="text-danger">3</span>,
    freq=<span class="text-success">&#x27;ME&#x27;</span>
<span class="p2n-bracket">)</span><span class="p2n-bracket"></span></code></div>

<p>This generates three month-end timestamps, not three days.</p>

<p><strong>5. Supplying Too Many Conflicting Parameters</strong></p>

<p>The function determines the range from combinations of <code>start</code>, <code>end</code>, <code>periods</code> and <code>freq</code>. Do not try to independently fix all four values when they conflict.</p>

<p><strong>6. Assuming an Anchored Frequency Always Starts on start</strong></p>

<p>For frequencies such as <code>MS</code> or <code>ME</code>, the first result must be a valid boundary for that frequency.</p>



## Summary of pandas.date_range()



<ul>
<li><code>pd.date_range()</code> creates a <code>DatetimeIndex</code>.</li>
<li>It is a Pandas function, not a DataFrame method.</li>
<li>Use <code>start</code> and <code>end</code> to define date boundaries.</li>
<li>Use <code>periods</code> to control how many timestamps are generated.</li>
<li>Use <code>freq</code> to control the distance between timestamps.</li>
<li><code>D</code> generates calendar days and <code>B</code> standard business days.</li>
<li>Use <code>MS</code> for month start and <code>ME</code> for month end.</li>
<li>Use <code>YS</code> for year start and <code>YE</code> for year end.</li>
<li>Use lowercase <code>h</code> for hours and <code>min</code> for minutes.</li>
<li>Multiple frequency units can be combined, such as <code>5h30min</code>.</li>
<li><code>inclusive</code> controls whether start and end boundaries are included.</li>
<li><code>normalize=True</code> resets the start and end times to midnight before generating the range.</li>
<li><code>tz</code> creates a timezone-aware DatetimeIndex.</li>
<li><code>name</code> assigns a name to the DatetimeIndex.</li>
<li><code>unit</code> controls datetime resolution where supported.</li>
<li><code>tolist()</code> converts the DatetimeIndex to a Python list.</li>
<li>Anchored frequencies such as <code>MS</code> and <code>ME</code> start from the next valid boundary when necessary.</li>
<li>Older aliases such as <code>H</code>, <code>T</code>, <code>M</code>, <code>Y</code> and <code>A-FEB</code> should not be used in current Pandas date_range() examples.</li>
</ul>
