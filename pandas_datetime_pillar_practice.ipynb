# Pandas Datetime Operations: Timestamps, Ranges, and Resampling

Interactive companion notebook for the Plus2Net [Pandas Datetime Pillar Hub](https://www.plus2net.com/python/pandas-dt.php). Run cells in order in Google Colab. All sample data is embedded.

## 1. Parsing Timestamps with pd.to_datetime()

```python
import pandas as pd
import numpy as np

# Parse current timestamp
now_ts = pd.to_datetime("now")
print("Current Time:", now_ts)

# Parse custom formatted string
custom_ts = pd.to_datetime("12252026:233445", format="%m%d%Y:%H%M%S")
print("Parsed Custom Date:", custom_ts)
assert custom_ts.year == 2026
assert custom_ts.month == 12
assert custom_ts.day == 25

df = pd.DataFrame({
    "NAME": ["Ravi", "Raju", "Alex", "Ron"],
    "dt_start": ["2026-01-15 08:30:00", "2026-02-20 14:15:00", "2026-03-10 18:45:00", "2026-04-05 09:00:00"]
})
df["dt_start"] = pd.to_datetime(df["dt_start"])
print(df.dtypes)
assert pd.api.types.is_datetime64_any_dtype(df["dt_start"])
```

## 2. Extracting Calendar Parts with the .dt Accessor

```python
df["year"] = df["dt_start"].dt.year
df["month"] = df["dt_start"].dt.month
df["day_name"] = df["dt_start"].dt.day_name()
df["is_weekend"] = df["dt_start"].dt.dayofweek >= 5
print(df)
assert list(df["year"]) == [2026, 2026, 2026, 2026]
assert list(df["month"]) == [1, 2, 3, 4]
```

## 3. Date Arithmetic with DateOffset & Timedelta

```python
df["dt_plus_7d"] = df["dt_start"] + pd.Timedelta(days=7)
df["next_month"] = df["dt_start"] + pd.DateOffset(months=1)
print(df[["NAME", "dt_start", "dt_plus_7d", "next_month"]])
assert df.loc[0, "next_month"].month == 2
```

## 4. Retail Daily Revenue Resampling Workflow

```python
np.random.seed(42)
date_idx = pd.date_range(start="2026-01-01", end="2026-03-31", freq="D")
retail_df = pd.DataFrame({
    "daily_sales": np.random.randint(500, 3000, size=len(date_idx))
}, index=date_idx)

weekly_sales = retail_df.resample("W").sum()
print("First 4 weeks of sales:")
print(weekly_sales.head(4))
assert len(weekly_sales) >= 13
```

## 5. Exercise: Employee Tenure & Service Anniversary

```python
emp_df = pd.DataFrame({
    "emp_name": ["Priya", "Karan", "Sneha", "Vikram"],
    "hire_date": pd.to_datetime(["2020-03-15", "2018-07-01", "2022-11-20", "2015-01-10"])
})
ref_date = pd.to_datetime("2026-06-01")
emp_df["tenure_days"] = (ref_date - emp_df["hire_date"]).dt.days
emp_df["next_5yr_anniversary"] = emp_df["hire_date"] + pd.DateOffset(years=5)
print(emp_df)
assert emp_df.loc[0, "next_5yr_anniversary"].year == 2025
assert (emp_df["tenure_days"] > 0).all()
```
