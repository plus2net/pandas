### 1. Parsing Timestamps with `to_datetime`
Parse strings, dates, and extract components with `.dt` accessor.


```python
import pandas as pd

now_ts = pd.to_datetime("now")
print("Current Timestamp:", now_ts)

custom_ts = pd.to_datetime("12252026:233445", format="%m%d%Y:%H%M%S")
print("Custom Parsed:", custom_ts)
assert custom_ts.year == 2026
assert custom_ts.month == 12
```

### 2. Time-Series Resampling & Rolling Statistics
Resample daily revenue into weekly aggregates and verify sum.


```python
import numpy as np

np.random.seed(42)
timestamps = pd.date_range(start="2026-01-01", periods=30, freq="D")
sales = np.random.normal(2500, 400, 30)

df_sales = pd.DataFrame({
    "date": timestamps,
    "daily_sales": np.round(sales, 2)
}).set_index("date")

weekly = df_sales.resample("W").sum()
print("Weekly Resampled:\n", weekly.head(3))

assert len(df_sales) == 30
assert np.isclose(weekly["daily_sales"].sum(), df_sales["daily_sales"].sum())

```

### 3. Hands-on Exercise: Employee Tenure & 5-Year Anniversary
Calculate elapsed tenure days and future service anniversary dates.


```python
staff = pd.DataFrame({
    "employee": ["Alice", "Bob", "Charlie", "Diana"],
    "hire_date": ["2021-03-15", "2022-07-01", "2023-11-20", "2025-01-10"]
})
staff["hire_date"] = pd.to_datetime(staff["hire_date"])

ref_date = pd.to_datetime("2026-10-11")
staff["tenure_days"] = (ref_date - staff["hire_date"]).dt.days
staff["anniversary_5yr"] = staff["hire_date"] + pd.DateOffset(years=5)

print(staff)

assert staff["tenure_days"].max() > 1500
assert staff["anniversary_5yr"].iloc[0] == pd.Timestamp("2026-03-15")
print("All assertions passed cleanly!")
```
