### 1. Basic Kernel Density Estimation
Generate a smooth KDE probability curve from subject scores.


```python
import pandas as pd
import matplotlib.pyplot as plt

my_dict = {
    'MATH': [30, 33, 32, 32, 31, 34, 32]
}
df = pd.DataFrame(data=my_dict)
print(df)

df.plot.density(figsize=(6, 3.5), title="Math Marks Density Estimation")
plt.show()
```

### 2. Bandwidth Tuning with `bw_method`
Adjust `bw_method` to control curve smoothness and peak resolution.


```python
df.plot.kde(bw_method=0.3, figsize=(6, 3.5), title="KDE with bw_method=0.3")
plt.show()
```

### 3. Comparing Multiple Distributions with Line Styles
Plot Math and English distributions with custom dashed styles.


```python
multi_df = pd.DataFrame({
    'MATH': [30, 33, 32, 32, 31, 34, 32],
    'ENG': [28, 30, 29, 31, 27, 32, 29]
})

multi_df.plot.density(
    style=['-', '--'],
    color=['#1f77b4', '#d62728'],
    figsize=(6.5, 3.5),
    title="Math vs English Score Densities"
)
plt.show()
```

### 4. Practical Workflow: Shipping Duration Densities
Simulate and compare delivery times across Standard and Express shipping.


```python
import numpy as np

np.random.seed(42)
n_orders = 1000

std_days = np.random.normal(5.2, 1.4, n_orders)
exp_days = np.random.normal(2.1, 0.6, n_orders)

df_shipping = pd.DataFrame({
    "Standard_Shipping_Days": np.clip(std_days, 1.0, 12.0),
    "Express_Shipping_Days": np.clip(exp_days, 0.5, 5.0)
})

print(f"Standard Median Days: {df_shipping['Standard_Shipping_Days'].median():.2f}")
print(f"Express Median Days: {df_shipping['Express_Shipping_Days'].median():.2f}")

assert len(df_shipping) == 1000
assert df_shipping["Express_Shipping_Days"].mean() < df_shipping["Standard_Shipping_Days"].mean()

ax = df_shipping.plot.density(
    style=["-", "--"],
    color=["#2b5c8f", "#d95f02"],
    bw_method=0.35,
    figsize=(7.5, 4.5),
    title="Delivery Lead Time Probability Densities by Tier"
)
plt.xlabel("Delivery Time (Days)")
plt.ylabel("Probability Density")
plt.grid(True, linestyle="--", alpha=0.6)
plt.show()
```

### 5. Hands-on Exercise: Retail Transaction Value Densities
Compare basket value distributions across two retail branches.


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(202)
n = 1000
branch_a = np.clip(np.random.normal(65, 15, n), 10, 150)
branch_b = np.clip(np.random.normal(90, 25, n), 15, 200)

df_retail = pd.DataFrame({
    "Branch_A": branch_a,
    "Branch_B": branch_b
})

assert len(df_retail) == 1000
assert df_retail["Branch_B"].mean() > df_retail["Branch_A"].mean()

print(f"Branch A Mean: ${df_retail['Branch_A'].mean():.2f}")
print(f"Branch B Mean: ${df_retail['Branch_B'].mean():.2f}")

df_retail.plot.kde(
    bw_method=0.3,
    style=["-", "--"],
    figsize=(7, 4),
    title="Retail Basket Value Density by Branch",
    grid=True
)
plt.xlabel("Basket Value ($)")
plt.ylabel("Density")
plt.show()
```
