## Syntax

`DataFrame.describe(percentiles=None, include=None, exclude=None)`

Use fractions between 0 and 1 for percentiles. Include and exclude select DataFrame dtypes; they are ignored for a Series.

## Describe One Numeric Column

<p><code>describe()</code> summarises non-missing values. Run this setup example first; later student examples call <code>student_data()</code> to start from fresh data.</p>

````python
import pandas as pd

def student_data():
    df = pd.DataFrame({
        "NAME": ["Ravi", "Raju", "Alex", "Ron", "King", "Jack"],
        "ID": [1, 2, 3, 4, 5, 6],
        "MATH": [80, 40, 70, 70, 60, 30],
        "ENGLISH": [80, 70, 40, 50, 60, 30],
    })
    # Explicit dtype makes text selection consistent across Pandas versions.
    df["NAME"] = df["NAME"].astype("string")
    return df

students = student_data()
print(students["MATH"].describe().to_string())
````

<p><strong>Expected output</strong></p>

```text
count     6.000000
mean     58.333333
std      19.407902
min      30.000000
25%      45.000000
50%      65.000000
75%      70.000000
max      80.000000
```

<p>The mean mark is about 58.33, while the median is 65. These are different summaries of the same six marks.</p>

## Describe a DataFrame

<p>For mixed numeric and text data, the default describes numeric columns. A numeric identifier is included even though its average usually has no useful meaning.</p>

````python
students = student_data()
print(students.describe().to_string())
````

<p>For a meaningful marks report, select <code>students[["MATH", "ENGLISH"]]</code> before describing the data. Missing values are excluded independently in each column.</p>

## Understand the Statistics

<div class="table-responsive"><table class="table table-bordered"><thead><tr><th scope="col">Statistic</th><th scope="col">Interpretation</th></tr></thead><tbody>
<tr><td>count</td><td>Number of non-missing observations in that column, not necessarily the number of rows.</td></tr>
<tr><td>mean</td><td>Arithmetic average. Large or small unusual values can affect it strongly.</td></tr>
<tr><td>std</td><td>Sample standard deviation, using a denominator of count minus one. With only one observation it is NaN.</td></tr>
<tr><td>min / max</td><td>Smallest and largest observed values.</td></tr>
<tr><td>25% / 50% / 75%</td><td>Quartiles; 50% is the median. Interpolation can produce a value that does not appear in the input.</td></tr>
<tr><td>unique</td><td>Number of distinct non-missing text or category values.</td></tr>
<tr><td>top / freq</td><td>A most frequent value and its frequency. In a tie, do not rely on which value is selected as top.</td></tr>
</tbody></table></div>

## Choose Custom Percentiles

<p>Percentiles use fractions from 0 to 1. The 50% median is included even when it is not in the supplied list.</p>

````python
students = student_data()
print(students["MATH"].describe(percentiles=[0.45, 0.68, 0.89]).to_string())
````

<p><strong>Expected output</strong></p>

```text

```

## Include All Column Types

<p><code>include="all"</code> combines statistics appropriate for each dtype. NaN in the combined report can mean a statistic does not apply, rather than a missing source record.</p>

````python
students = student_data()
print(students.describe(include="all").to_string())
````

<p><strong>Expected output</strong></p>

```text
        NAME        ID       MATH    ENGLISH
count      6  6.000000   6.000000   6.000000
unique     6       NaN        NaN        NaN
top     Ravi       NaN        NaN        NaN
freq       1       NaN        NaN        NaN
mean     NaN  3.500000  58.333333  55.000000
std      NaN  1.870829  19.407902  18.708287
min      NaN  1.000000  30.000000  30.000000
25%      NaN  2.250000  45.000000  42.500000
50%      NaN  3.500000  65.000000  55.000000
75%      NaN  4.750000  70.000000  67.500000
max      NaN  6.000000  80.000000  80.000000
```

<p>All student names occur once. Any of them may appear as <code>top</code>; the important result is <code>freq = 1</code>.</p>

## Select Text Columns

<p>Use an explicit string dtype and <code>include=["string"]</code>. For columns genuinely stored as object dtype, use <code>include=["object"]</code>. Avoid the removed NumPy alias <code>np.object</code>.</p>

## Missing Values and Small Samples

<p>Count shows only available observations. An all-missing numeric column has a count of zero and undefined summary statistics.</p>

````python
values = pd.DataFrame({"observed": [10.0, None, 30.0], "empty": [float("nan")] * 3})
print(values.describe().to_string())
print("Missing values:")
print(values.isna().sum().to_string())
````

<p><strong>Expected output</strong></p>

```text
        observed  empty
count   2.000000    0.0
mean   20.000000    NaN
std    14.142136    NaN
min    10.000000    NaN
25%    15.000000    NaN
50%    20.000000    NaN
75%    25.000000    NaN
max    30.000000    NaN
Missing values:
observed    1
empty       3
```

<p>Use <a href="pandas-dataframe-isnull.php">isnull()</a> or <code>isna()</code> for missingness. Describe does not tell you which rows are missing.</p>

## Compare Statistics Before and After Filling

<p>Zero filling changes the average and count. Median filling estimates a value and changes count too, even when the average stays the same.</p>

````python
scores = pd.Series([10.0, 20.0, None, 30.0])
report = pd.DataFrame({"observed": scores.describe(), "zero_filled": scores.fillna(0).describe(), "median_filled": scores.fillna(scores.median()).describe()})
print(report.loc[["count", "mean", "50%", "std"]].to_string())
````

<p><strong>Expected output</strong></p>

```text
       observed  zero_filled  median_filled
count       3.0     4.000000       4.000000
mean       20.0    15.000000      20.000000
50%        20.0    15.000000      20.000000
std        10.0    12.909944       8.164966
```

<p>A higher count after filling is not evidence that you collected more observations. Read <a href="pandas-dataframe-fillna.php">fillna() and practical missing-value handling</a> before choosing a replacement.</p>

## Practical Example: Inspect a Sales CSV

<p>Use the same synthetic five-order CSV as the fillna lesson. The code loads it from Plus2Net; an internet connection is required. Exclude order IDs from the numeric report.</p>

````python
import pandas as pd

DATA_URL = "https://www.plus2net.com/python/download/pandas_fillna_sample_sales.csv"
sales = pd.read_csv(DATA_URL, dtype={"category": "string"})
print(sales[["quantity", "unit_price", "discount"]].describe().to_string())
print("\nMissing values:")
print(sales[["quantity", "unit_price", "discount"]].isna().sum().to_string())
print("\nRows:", len(sales))
````

<p><strong>Expected output</strong></p>

```text
       quantity  unit_price  discount
count  4.000000    4.000000  3.000000
mean   2.000000   18.750000  1.666667
std    0.816497    8.539126  2.886751
min    1.000000   10.000000  0.000000
25%    1.750000   13.750000  0.000000
50%    2.000000   17.500000  0.000000
75%    2.250000   22.500000  2.500000
max    3.000000   30.000000  5.000000

Missing values:
quantity      1
unit_price    1
discount      2

Rows: 5
```

<p>There are five rows but only four known quantities and prices, and three recorded discounts. That is a useful cleaning signal. The price median is not permission to invent the price of an incomplete order.</p>

## Verify the Agreed Cleaning Rules

<p>Continue from the preceding sales example. A blank discount means zero under this exercise rule; quantities and prices remain unknown. Compare the reports.</p>

````python
cleaned = sales.fillna({"category": "Unknown", "discount": 0})
print(cleaned[["quantity", "unit_price", "discount"]].describe().to_string())
assert len(cleaned) == len(sales)
assert cleaned["quantity"].isna().sum() == sales["quantity"].isna().sum()
assert cleaned["unit_price"].isna().sum() == sales["unit_price"].isna().sum()
print("Row count and unknown quantities/prices preserved.")
````

<p><strong>Expected output</strong></p>

```text
       quantity  unit_price  discount
count  4.000000    4.000000  5.000000
mean   2.000000   18.750000  1.000000
std    0.816497    8.539126  2.236068
min    1.000000   10.000000  0.000000
25%    1.750000   13.750000  0.000000
50%    2.000000   17.500000  0.000000
75%    2.250000   22.500000  0.000000
max    3.000000   30.000000  5.000000
Row count and unknown quantities/prices preserved.
```

<p>The discount count becomes five and its mean changes. Quantity and price counts remain four. Always explain the rule behind a change in statistics.</p>

## Common Questions and Mistakes

<ul><li><strong>Why is a column missing?</strong> The default mixed-data report selects numeric columns. Check dtypes and use include.</li><li><strong>Why is count smaller than the row count?</strong> That column has missing values.</li><li><strong>Why does top change?</strong> Tied most-frequent labels do not have a guaranteed winner.</li><li><strong>Why is std NaN?</strong> There may be fewer than two non-missing observations.</li><li><strong>Why does selecting text fail?</strong> Check whether its dtype is string, object or category. A selection with no matching columns raises an error.</li><li><strong>Does describe detect outliers?</strong> It provides clues; inspect records and use plots before drawing conclusions.</li></ul>

## Key Takeaways

- Count excludes missing values independently for each column.
- Numeric identifiers should usually be excluded from meaningful reports.
- Text and category summaries differ from numeric summaries.
- A tied top value is not guaranteed to be the same in every environment.
