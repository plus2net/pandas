## Describe a specific option

describe_option() prints an option description with its registered default and current value. The original max_rows example is retained using the full name. Text and defaults may differ by installed Pandas version.

````python
import pandas as pd
pd.describe_option("display.max_rows")
assert isinstance(pd.get_option("display.max_rows"), (int, type(None)))
````

**Expected output**

```text
display.max_rows : int
    If max_rows is exceeded, switch to truncate view. Depending on
    `large_repr`, objects are either centrally truncated or printed as
    a summary view.

    'None' value means unlimited. Beware that printing a large number of rows
    could cause your rendering environment (the browser, etc.) to crash.

    In case python/IPython is running in a terminal and `large_repr`
    equals 'truncate' this can be set to 0 and pandas will auto-detect
    the height of the terminal and print a truncated object which fits
    the screen height. The IPython notebook, IPython qtconsole, or
    IDLE do not run in a terminal and hence it is not possible to do
    correct auto-detection.
    [default: 60] [currently: 60]
```

## Inspect a temporary custom value

The original value of 10 is retained. Default and current value are different concepts. Use a context so this demonstration leaves your earlier settings intact.

````python
before = pd.get_option("display.max_rows")
with pd.option_context("display.max_rows", 10):
    pd.describe_option("display.max_rows")
    assert pd.get_option("display.max_rows") == 10
print("Previous value restored:", pd.get_option("display.max_rows") == before)
assert pd.get_option("display.max_rows") == before
````

## Get the installed option catalogue

The original long catalogue includes settings from an older Pandas release, including removed names. Generate the list from the installed version instead of treating that snapshot as current. Call pd.describe_option() directly to print the full list. This example captures that same output and prints selected catalogue counts so the notebook stays readable.

````python
import io
from contextlib import redirect_stdout
buffer = io.StringIO()
with redirect_stdout(buffer):
    pd.describe_option()
catalogue = buffer.getvalue()
headers = [line.split(" :", 1)[0] for line in catalogue.splitlines()
           if line and not line[0].isspace() and " :" in line]
for prefix in ["display.", "compute.", "io.", "mode.", "plotting."]:
    print(prefix, sum(name.startswith(prefix) for name in headers))
assert "display.max_rows" in catalogue and "display.max_columns" in catalogue
````

**Expected output**

```text
display. 29
compute. 3
io. 12
mode. 5
plotting. 2
```

## Inspect a setting before changing it

Use descriptions to understand a setting, get_option to read it and option_context for a temporary rendering decision. No table values change.

````python
pd.describe_option("display.precision")
data = pd.DataFrame({"amount": [1.23456, 9.87654]})
with pd.option_context("display.precision", 2):
    print(data)
assert data["amount"].tolist() == [1.23456, 9.87654]
````

**Expected output**

```text
display.precision : int
    Floating point output precision in terms of number of places after the
    decimal, for regular formatting as well as scientific notation. Similar
    to ``precision`` in :meth:`numpy.set_printoptions`.
    [default: 6] [currently: 6]
   amount
0    1.23
1    9.88
```

## Common mistakes

describe_option() prints text and normally returns None; it does not return an option-value mapping. Use <a href="pandas-display-options-get-set.php">get_option()</a> for values. Default values are not necessarily your previous values. Prefer <a href="pandas-display-option_context.php">option_context()</a> for temporary changes and <a href="pandas-display-reset_option.php">reset_option()</a> for defaults. Option names and documentation can change between versions, so regenerate the catalogue after upgrading.

## Exercise solution

Read the existing value first and keep the change scoped.

````python
saved = pd.get_option("display.precision")
pd.describe_option("display.precision")
with pd.option_context("display.precision", 3):
    print("Inside:", pd.get_option("display.precision"))
    assert pd.get_option("display.precision") == 3
print("Restored:", pd.get_option("display.precision") == saved)
assert pd.get_option("display.precision") == saved
````

**Expected output**
