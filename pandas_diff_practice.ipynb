# Pandas: Calculate Discrete Differences with diff()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-diff.php](https://www.plus2net.com/python/pandas-diff.php)

Run cells in order in Google Colab.

## Create numerical sample series

```python
import pandas as pd

s = pd.Series([10, 15, 12, 18], name="VAL")
print(s)
assert s.shape == (4,)
```

## Basic discrete difference calculation

```python
res = s.diff()
print(res)
assert res.iloc[1] == 5.0 and res.iloc[2] == -3.0
```

## Configuring lag periods

```python
res_lag = s.diff(periods=2)
print(res_lag)
assert res_lag.iloc[2] == 2.0
```

## Differencing DataFrames across rows or columns

```python
df = pd.DataFrame({
    "A": [10, 20, 30],
    "B": [15, 25, 40]
})
df_diff = df.diff(axis=1)
print(df_diff)
assert df_diff.loc[0, "B"] == 5.0
```

## Exercise solution

```python
temps = pd.Series([30, 32, 31, 35], name="TEMP")
answer = temps.diff()
print(answer)
assert answer.iloc[3] == 4.0
```
