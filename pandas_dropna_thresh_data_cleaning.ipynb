## Create the student table with missing values

The table uses id as its index, so id is not counted among the four data columns. Each example starts from a fresh copy. The <a href="https://www.plus2net.com/python/download/student-dropna_1.xlsx">student workbook link</a> is available for the Excel route; see <a href="https://www.plus2net.com/python/pandas-read_excel.php">read_excel()</a>. The embedded examples do not depend on that download.

````python
import pandas as pd
import math
def student_data():
    rows = [
        [1, "John Deo", "Four", 75, "female"],
        [2, "Max Ruin", "Three", 85, "male"],
        [None, "Arnold", "Three", 55, "male"],
        [4, "Krish Star", "Four", 60, "female"],
        [None, "John Mik", "Four", 60, "female"],
        [6, "Alex Jo", "Four", 55, None],
        [7, "My John", "Five", 78, "male"],
        [None, None, None, None, None],
        [9, "Tes Qry", "Six", 78, "male"],
        [10, None, "Four", 55, "female"],
        [11, "Ronald", "Six", None, "female"],
        [12, "Recky", "Six", 94, "female"],
        [13, "Ron", None, 55, None],
        [14, "King", None, None, None]
    ]
    return pd.DataFrame(rows, columns=["id", "name", "class1", "mark", "gender"]).set_index("id")
df = student_data()
print(df)
print("Non-missing counts per column:")
print(df.count())
assert df.shape == (14, 4)
````

## Keep rows with at least two non-missing fields

axis=0 removes rows. thresh=2 counts present values in the data columns; it is not the maximum number of missing cells. Rows with zero or one present field are removed.

````python
df = student_data()
result = df.dropna(axis=0, thresh=2)
print(result)
assert len(result) == 12
assert result.notna().sum(axis=1).ge(2).all()
````

## Keep columns with at least eleven non-missing values

## Require at least 70 percent of fields in each row

Convert the fraction to an integer requirement with math.ceil(). With four fields, 70 percent means at least three present values. Define which fields count before choosing the percentage.

````python
df = student_data()
fraction = 0.70
required = math.ceil(df.shape[1] * fraction)
result = df.dropna(axis=0, thresh=required)
print("Required fields:", required)
print(result)
assert required == 3 and len(result) == 11
````

## Require at least 80 percent of rows in each column

With fourteen rows, 80 percent requires twelve non-missing values. Calculate the denominator from the intended population. Filtering rows first changes this requirement and potentially the result.

````python
df = student_data()
required = math.ceil(len(df) * 0.80)
result = df.dropna(axis=1, thresh=required)
print("Required rows:", required)
print(result)
assert required == 12 and result.columns.tolist() == ["name"]
````

## Count only relevant fields with subset

For row filtering, subset specifies columns used in the count while retaining the other columns in the output. An always-present helper column should not inflate a completeness score.

````python
df = student_data()
fields = ["name", "mark"]
result = df.dropna(subset=fields, thresh=2)
print(result)
assert len(result) == 10
assert result[fields].notna().all().all()
````

## Choose a separate output or inplace modification

inplace=True modifies the DataFrame and returns None. Do not assign that return value to df. Separate output tables make it easier to compare retained and rejected records.

````python
df = student_data()
returned = df.dropna(thresh=2, inplace=True)
print("Retained rows:", len(df))
print("Return value:", returned)
assert returned is None and len(df) == 12
````

**Expected output**

```text
Retained rows: 12
Return value: None
```

## Do not combine how and thresh

how specifies any/all missing-value logic, while thresh specifies a minimum present-value count. Passing both raises TypeError. Omit how when using thresh. This caught error is included to show the invalid combination safely.

````python
df = student_data()
try:
    df.dropna(how="any", thresh=2)
except TypeError as error:
    print(type(error).__name__ + ": " + str(error))
print("Valid threshold result rows:", len(df.dropna(thresh=2)))
````

**Expected output**

```text
TypeError: You cannot set both the how and thresh arguments at the same time.
Valid threshold result rows: 12
```

## Practical workflow: retain sufficiently complete customer records

This synthetic rule requires a customer ID plus at least two of email, phone and city. A count across the whole table would wrongly let ID or audit columns satisfy the contact requirement. Normalise blank strings before counting.

````python
raw = pd.DataFrame({"customer_id": [101, 102, 103, 104, None], "email": ["a@example.com", "", None, "d@example.com", "e@example.com"], "phone": ["111", "222", None, "  ", "555"], "city": ["Delhi", "Pune", "Mumbai", None, "Kochi"]})
df = raw.copy()
fields = ["email", "phone", "city"]
for field in fields:
    text = df[field].astype("string").str.strip()
    df[field] = text.mask(text.eq(""), pd.NA)
present = df[fields].notna().sum(axis=1)
keep = df["customer_id"].notna() & present.ge(2)
clean = df.loc[keep].copy()
review = raw.loc[~keep].copy()
print(clean)
print("Review rows:")
print(review)
assert clean["customer_id"].tolist() == [101, 102]
````

**Expected output**

```text
   customer_id          email phone   city
0        101.0  a@example.com   111  Delhi
1        102.0           <NA>   222   Pune
Review rows:
   customer_id          email phone    city
2        103.0            NaN   NaN  Mumbai
3        104.0  d@example.com           NaN
4          NaN  e@example.com   555   Kochi
```

## Check completeness without confusing it with validity

Non-missing does not mean valid: email syntax, phone format and customer identity require separate checks. Retain rejected records for review and justify the threshold based on intended use. See <a href="https://www.plus2net.com/python/pandas-validate_email.php">email validation</a>, <a href="https://www.plus2net.com/python/pandas-dataframe-count.php">non-missing counts</a> and <a href="https://www.plus2net.com/python/pandas-dataframe-fillna.php">fillna()</a>.

````python
candidate = df.dropna(subset=fields, thresh=2)
candidate = candidate.loc[candidate["customer_id"].notna()]
assert candidate.equals(clean)
assert clean[fields].notna().sum(axis=1).ge(2).all()
assert len(raw) == len(clean) + len(review)
print("Input:", len(raw), "Retained:", len(clean), "Review:", len(review))
print(clean.to_csv(index=False).rstrip())
# Optional Colab exports:
# clean.to_csv("complete_customer_records.csv", index=False)
# review.to_csv("incomplete_customer_records.csv", index=False)
````

```text
101.0,a@example.com,111,Delhi
102.0,,222,Pune
```

## Common questions

<strong>Does thresh count missing cells?</strong> No; it counts non-missing cells. <strong>Are blanks and spaces missing?</strong> Not automatically; define and normalise them. <strong>Does the index count?</strong> No. <strong>Can a threshold require a specific field?</strong> Not by itself; combine a completeness rule with an explicit required-field check. <strong>What if thresh exceeds the number of considered fields?</strong> No rows can meet that requirement. <strong>Which percentage should I use?</strong> Choose it from the required use of the data, and inspect which records it excludes. <strong>Is it safe to remove sparse columns automatically?</strong> Review their meaning first; rare fields may still be important.

## Exercise: require two of three measurements

Create rows [1, 2, missing], [1, missing, missing] and [missing, 2, 3] in columns a, b and c. Retain rows with at least two present measurements. Expect the first and third rows.

## Exercise solution

The requirement applies to the three measurement columns only.

````python
practice = pd.DataFrame({"a": [1, 1, None], "b": [2, None, 2], "c": [None, None, 3]})
answer = practice.dropna(subset=["a", "b", "c"], thresh=2)
print(answer)
assert answer.index.tolist() == [0, 2]
````

**Expected output**

```text
     a    b    c
0  1.0  2.0  NaN
2  NaN  2.0  3.0
```
