# Pandas: High-Performance Evaluation with eval()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-eval.php](https://www.plus2net.com/python/pandas-eval.php)

Run cells in order in Google Colab.

## Create numerical sample dataset

```python
import pandas as pd

df = pd.DataFrame({
    "ITEM": ["Apple", "Banana", "Cherry", "Date"],
    "QTY": [10, 25, 15, 30],
    "PRICE": [20.0, 5.0, 50.0, 12.0]
})
print(df)
assert df.shape == (4, 3)
```

## Basic column evaluation

```python
res = df.eval("QTY * PRICE")
print(res)
assert res.iloc[0] == 200.0
```

## Modifying DataFrames with inplace=True

```python
df.eval("TOTAL = QTY * PRICE", inplace=True)
print(df)
assert "TOTAL" in df.columns and df.loc[0, "TOTAL"] == 200.0
```

## Referencing local variables using @

```python
tax_rate = 0.1
df.eval("TAX = TOTAL * @tax_rate", inplace=True)
print(df)
assert df.loc[0, "TAX"] == 20.0
```

## Exercise solution

```python
ex_df = pd.DataFrame({"PRICE": [100, 200, 300], "DISCOUNT": [10, 20, 30]})
ex_df.eval("NET = PRICE - DISCOUNT", inplace=True)
print(ex_df)
assert ex_df.loc[0, "NET"] == 90
```
