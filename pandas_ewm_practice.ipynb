## 1. Above-the-Fold Basic EWM Mean

Compute exponentially weighted moving average with span=2.

```python
import pandas as pd

df = pd.DataFrame({'Price': [10, 11, 12, 13]})
result = df.ewm(span=2).mean()
print(result)

# Self-validating assertion
assert round(result.loc[3, 'Price'], 4) == 12.5417
```

## 2. Exponential Moving Average with `span`

Compute an EMA equivalent to a 3-period span.

```python
df_prices = pd.DataFrame({'Close': [100, 102, 101, 105, 107]})
df_prices['EMA_3'] = df_prices['Close'].ewm(span=3).mean()
print(df_prices)

# Self-validating assertion
assert df_prices.loc[0, 'EMA_3'] == 100.0
```
