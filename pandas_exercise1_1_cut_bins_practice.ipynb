# Pandas DataFrame Exercise 1-1: Binning with cut(), Groupby & Plots

Interactive companion notebook for [Pandas Exercise 1-1](https://www.plus2net.com/python/pandas-dataframe-exercise1-1.php). Covers synthetic score generation, binning into grade intervals, and plotting scatter, pie, and bar charts.

## Score Generation & Discretization with pd.cut()

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
n = 25
df = pd.DataFrame({
    "ID": np.arange(1, n + 1),
    "MATH": np.random.randint(0, 100, size=n),
    "ENG": np.random.randint(0, 100, size=n)
})
grade_labels = ["Fail", "Third", "Second", "First"]
df["my_result"] = pd.cut(
    x=df["MATH"],
    bins=[0, 40, 50, 75, 101],
    labels=grade_labels,
    right=False
)
print(df.head())
assert len(df) == 25
result_counts = df.groupby("my_result", observed=False)[["ID"]].count()
result_counts.columns = ["Count"]
print("Grade Counts:", result_counts)
assert result_counts["Count"].sum() == 25
```
