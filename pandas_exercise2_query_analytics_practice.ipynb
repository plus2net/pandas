## Ingest kw1_short.xlsx & Execute Practice Queries

```python
import pandas as pd
import os

excel_path = "download/kw1_short.xlsx"
df = pd.read_excel(excel_path)
print("Total Records:", len(df))
assert len(df) == 49

# Queries containing HTML
html_df = df[df["Query"].str.contains("HTML", case=False, na=False)].copy()
print(f"HTML Queries: {len(html_df)}")
assert len(html_df) == 15

# Total Clicks
total_clicks = html_df["Clicks"].sum()
print("Total Clicks:", total_clicks)
assert total_clicks == 3528

# Top Click Query
top_query = html_df[html_df["Clicks"] == html_df["Clicks"].max()]
print("Top Query:", top_query.iloc[0]["Query"])
assert top_query.iloc[0]["Clicks"] == 406
assert top_query.iloc[0]["Query"] == "marquee speed html"

# Query Lengths
html_df["LENGTH"] = html_df["Query"].str.len()
min_len = html_df["LENGTH"].min()
shortest = html_df[html_df["LENGTH"] == min_len]
print(f"Shortest Query ({min_len} chars):", shortest.iloc[0]["Query"])
assert min_len == 13
assert "html calendar" in shortest["Query"].values
```
