# Pandas Datetime Exercise 3-2: Birthday Arithmetic, Ages, and strftime

Interactive companion notebook for [Pandas Exercise 3-2](https://www.plus2net.com/python/pandas-dataframe-exercise3-2.php). Covers timestamp parsing, component extraction, timedelta math, and strftime formatting.

## Birthday Timestamp Parsing & Component Extraction

```python
import pandas as pd
import numpy as np

dt_bd = pd.to_datetime("2000-12-23 16:20:05")
print("Birthday:", dt_bd)
print("Year:", dt_bd.year)
print("Month:", dt_bd.month)
print("Day:", dt_bd.day)
print("Hour:", dt_bd.hour)
assert dt_bd.year == 2000
assert dt_bd.month == 12
assert dt_bd.day == 23
```

## Target Age and Elapsed Duration Calculations

```python
dt_target = pd.to_datetime("2020-01-01 13:22:34")
dt_diff = dt_target - dt_bd
print("Difference:", dt_diff)
print(f"Age in Days: {dt_diff.days} days")
age_weeks = dt_diff / pd.Timedelta(weeks=1)
print(f"Age in Weeks: {age_weeks:.2f} weeks")
assert dt_diff.days == 6947
```

## DateOffset Calendar Shifts & strftime Formatting

```python
dt_future = dt_bd + pd.DateOffset(years=5, months=2, days=25)
print("Birthday + 5y 2m 25d:", dt_future)
assert dt_future.year == 2006
assert dt_future.month == 3
assert dt_future.day == 20

dt_sample = pd.to_datetime("2026-04-18 20:59:42.123456")
print(dt_sample.strftime("%d-%b-%Y"))
print(dt_sample.strftime("%d-%B-%Y"))
print(dt_sample.strftime("%H:%M:%S.%f"))
assert dt_sample.strftime("%d-%b-%Y") == "18-Apr-2026"
```

## Business Day Sequences

```python
b_days = pd.date_range(end="2026-10-09", periods=10, freq="B")
print("Last 10 Business Days:")
for d in b_days:
    print(d.strftime("%Y-%m-%d (%A)"))
assert len(b_days) == 10
assert (b_days.dayofweek < 5).all()
```
