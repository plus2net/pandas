# Pandas Datetime Exercise 3-3: Backlink Crawl Date Analysis

Interactive companion notebook for [Pandas Exercise 3-3](https://www.plus2net.com/python/pandas-dataframe-exercise3-3.php). Analyzes 10,707 external crawled backlinks from Google Search Console export .

## Load & Inspect Backlinks Dataset

```python
import pandas as pd
import numpy as np
import os

excel_path = "download/links_external.xlsx"
df_links = pd.read_excel(excel_path)
print("Shape:", df_links.shape)
print("Columns:", df_links.columns.tolist())
assert df_links.shape[0] == 10707
assert df_links.shape[1] == 2
```

## Temporal Groupings by Year and Month

```python
df_links["Last crawled"] = pd.to_datetime(df_links["Last crawled"])
yearly = df_links.groupby(df_links["Last crawled"].dt.year)["Linking page"].count()
print("Links Crawled by Year:")
print(yearly)

monthly = df_links.groupby(df_links["Last crawled"].dt.month)["Linking page"].count()
print("Links Crawled by Month (All Years):")
print(monthly)
assert yearly.sum() == 10707
```

## Domain Filtering with str.contains()

```python
domain_links = df_links[df_links["Linking page"].str.contains("example2", case=False, na=False)]
print(f"Total example2.com Backlinks: {len(domain_links)}")
print("Annual Distribution of example2.com Links:")
print(domain_links.groupby(domain_links["Last crawled"].dt.year)["Linking page"].count())
assert len(domain_links) == 2677
```
