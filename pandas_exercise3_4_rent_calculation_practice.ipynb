# Pandas Datetime Exercise 3-4: Accommodation Stay & Rent Calculation

Interactive companion notebook for [Pandas Exercise 3-4](https://www.plus2net.com/python/pandas-dataframe-exercise3-4.php). Covers check-in date durations, missing ID auditing, nested  logic, and billable overstay identification.

## Load Excel Roster and Audit Missing IDs

```python
import pandas as pd
import numpy as np
import os

excel_path = "download/rent-calculation.xlsx"
df_rent = pd.read_excel(excel_path)
print("Columns:", df_rent.columns.tolist())
print("Students by Dept:")
print(df_rent["dept"].value_counts())

missing_ids = df_rent[df_rent["id"].isnull()]
print(f"Missing ID Records: {len(missing_ids)}")
assert len(missing_ids) == 2
```

## Duration Calculation Relative to Base Billing Date

```python
df_rent["ck_date"] = pd.to_datetime(df_rent["ck_date"])
base_date = pd.to_datetime("2020-06-01")
df_rent["diff_days"] = (base_date - df_rent["ck_date"]).dt.days

over_200 = df_rent[df_rent["diff_days"] > 200]
print(f"Students stayed > 200 days: {len(over_200)}")
assert len(over_200) > 0
```

## Departmental Free Days Allocation with Nested np.where()

```python
df_rent["allowed"] = np.where(
    df_rent["dept"] == "mktg", 50,
    np.where(df_rent["dept"] == "production", 65,
    np.where(df_rent["dept"] == "planning", 45, np.nan))
)

billable = df_rent[df_rent["diff_days"] > df_rent["allowed"]]
print(f"Billable Overstay Students: {len(billable)}")
print(billable[["name", "dept", "diff_days", "allowed"]])
assert len(billable) > 0
```
