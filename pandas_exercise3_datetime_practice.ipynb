# Pandas Datetime Exercise 3-1: 365-Day Maintenance Log Queries

Interactive companion notebook for [Pandas Exercise 3-1](https://www.plus2net.com/python/pandas-dataframe-exercise3.php). Contains synthetic 365-day dataset generator and all 8 query solutions.

## Generate Synthetic Maintenance Log Dataset

```python
import datetime
import pandas as pd
import numpy as np
import random

random.seed(42)
dt_range = 365
base_end = pd.to_datetime("2026-10-10")
dt_list = pd.date_range(end=base_end, periods=dt_range).tolist()

l_activity = ["Checking", "Cleaning", "Inspection", "Repair", "Painting"]
l_dept = ["Dept1", "Dept2", "Dept3", "Dept4"]
l_eqpt = ["AC", "Fan", "Pump", "Generator", "Server", "Battery", "UPS"]

my_dict = {
    "LOG_ID": list(range(1001, 1001 + dt_range)),
    "LOG_DATE": dt_list,
    "ACTIVITY": [random.choice(l_activity) for _ in range(dt_range)],
    "DEPT": [random.choice(l_dept) for _ in range(dt_range)],
    "EQUIPMENT": [random.choice(l_eqpt) for _ in range(dt_range)]
}
df = pd.DataFrame(my_dict)
print(df.head())
assert len(df) == 365
assert df["LOG_ID"].iloc[0] == 1001
```

## Execute Maintenance Log Queries (Q1 to Q8)

```python
# Q1: Records from 2026
q1 = df[df["LOG_DATE"].dt.year == 2026]
print(f"Q1: 2026 Logs: {len(q1)}")

# Q2: Inspection activity in 2026
q2 = df[(df["LOG_DATE"].dt.year == 2026) & (df["ACTIVITY"] == "Inspection")]
print(f"Q2: 2026 Inspections: {len(q2)}")

# Q3: Inspection OR Repair in Dept1 during 2026
q3 = df[(df["LOG_DATE"].dt.year == 2026) & (df["ACTIVITY"].isin(["Inspection", "Repair"])) & (df["DEPT"] == "Dept1")]
print(f"Q3: Dept1 Inspections/Repairs: {len(q3)}")

# Q4: Weekend activities
q4 = df[df["LOG_DATE"].dt.dayofweek >= 5]
print(f"Q4: Weekend maintenance records: {len(q4)}")
assert len(q4) > 0

# Q5: Monthly Activity Breakdown
q5 = df.groupby([df["LOG_DATE"].dt.month, "ACTIVITY"])["LOG_ID"].count().unstack().fillna(0)
print("Q5: Monthly Activity Breakdown Table:")
print(q5.head())
```
