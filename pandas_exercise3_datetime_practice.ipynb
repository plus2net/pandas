### 1. Generate 365-Day Synthetic Maintenance Log
Create the sample DataFrame using `pd.date_range()` and random choice.


```python
import datetime
import pandas as pd
import random

random.seed(42)
dt_range = 365
dt_list = pd.date_range(end=datetime.date.today(), periods=dt_range).tolist()

l_activity = ["Checking", "Cleaning", "Inspection", "Repair", "Painting"]
l_dept = ["Dept1", "Dept2", "Dept3", "Dept4"]
l_eqpt = ["AC", "Fan", "Pump", "Generator", "Server", "Battery", "UPS"]

df = pd.DataFrame({
    "Log_ID": [1000 + i for i in range(dt_range)],
    "dt": pd.to_datetime(dt_list),
    "dept": [random.choice(l_dept) for _ in range(dt_range)],
    "activity": [random.choice(l_activity) for _ in range(dt_range)],
    "eqpt": [random.choice(l_eqpt) for _ in range(dt_range)]
})

print(df.head())
assert len(df) == 365
assert df["Log_ID"].iloc[0] == 1000
```

### 2. Time-Series Queries: Latest Equipment Service & Wednesdays
Find latest inspection for Pump and filter tasks performed on Wednesdays.


```python
pump_logs = df[df["eqpt"] == "Pump"]
latest_pump = pump_logs["dt"].max()
days_elapsed = (pd.to_datetime("now") - latest_pump).days

print(f"Latest Pump Service: {latest_pump.strftime('%Y-%m-%d')}")
print(f"Days Elapsed: {days_elapsed} days")

wednesdays = df[df["dt"].dt.weekday == 2]
print(f"Wednesday Task Count: {len(wednesdays)}")
assert len(wednesdays) in [52, 53]
```

### 3. Temporal Grouping by Year and Month


```python
by_month = df.groupby(df["dt"].dt.month)["Log_ID"].count()
print("Tasks by Month:\n", by_month)
assert by_month.sum() == 365
print("All time-series assertions passed cleanly!")
```
