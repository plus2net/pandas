# Pandas: Explode List-Like Columns into Rows

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-explode.php](https://www.plus2net.com/python/pandas-explode.php)

Run cells in order in Google Colab.

## Create nested list sample data

```python
import pandas as pd
df = pd.DataFrame({
    "NAME": ["Ravi", "Raju", "Alex"],
    "ID": [1, 2, 3],
    "SUBJECTS": [["Math", "Physics"], ["English"], ["Chemistry", "Biology", "Math"]]
})
print(df)
assert df.shape == (3, 3)
```

## Basic explode on a single column

```python
exploded = df.explode("SUBJECTS")
print(exploded)
assert len(exploded) == 6
```

## Resetting index with ignore_index

```python
reset = df.explode("SUBJECTS", ignore_index=True)
print(reset)
assert reset.index.tolist() == list(range(6))
```

## Exploding multiple matching columns

```python
multi_df = pd.DataFrame({
    "NAME": ["Ravi", "Raju"],
    "SUBJECTS": [["Math", "Physics"], ["English", "History"]],
    "MARKS": [[85, 90], [78, 88]]
})
multi_exploded = multi_df.explode(["SUBJECTS", "MARKS"], ignore_index=True)
print(multi_exploded)
assert len(multi_exploded) == 4
```

## Handling empty lists and missing values

```python
empty_df = pd.DataFrame({
    "NAME": ["Ravi", "Raju"],
    "SUBJECTS": [["Math"], []]
})
empty_exploded = empty_df.explode("SUBJECTS", ignore_index=True)
print(empty_exploded)
assert pd.isna(empty_exploded.iloc[1]["SUBJECTS"])
```

## Exercise solution

```python
ex_df = pd.DataFrame({
    "NAME": ["Ravi", "Raju"],
    "SCORES": [[90, 85, 95], [70, 75]]
})
answer = ex_df.explode("SCORES", ignore_index=True)
print(answer)
assert len(answer) == 5 and answer.index.tolist() == list(range(5))
```
