## 1. Basic Categorical Factorization

Encode fruit categories into integer labels and unique indexes.

````python
import pandas as pd

s = pd.Series(['apple', 'banana', 'apple'])
labels, uniques = pd.factorize(s)
print(labels)
print(uniques)

# Self-validating assertion
assert list(labels) == [0, 1, 0]
assert len(uniques) == 2
````

**Expected output**

```text
[0 1 0]
Index(['apple', 'banana'], dtype='object')
```
