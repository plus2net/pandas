## 1. Above-the-Fold Basic Categorical Factorization

Encode fruit categories into integer codes and unique indexes.

```python
import pandas as pd

s = pd.Series(['apple', 'banana', 'apple'])
labels, uniques = pd.factorize(s)
print('Codes:', labels)
print('Uniques:', uniques)

# Self-validating assertions
assert list(labels) == [0, 1, 0]
assert len(uniques) == 2
```

## 2. Handling Missing Values (`NaN` Sentinel Code `-1`)

Verify that missing entries receive sentinel codes of `-1`.

```python
import numpy as np

s_nan = pd.Series(['High', np.nan, 'Low', 'High'])
codes, uniques = pd.factorize(s_nan)
print('Codes:', codes)
print('Uniques:', uniques)

# Self-validating assertions
assert codes[1] == -1
assert len(uniques) == 2
```

## 3. Alphabetical Sorting of Unique Categories

Guarantee consistent integer codes by sorting categories alphabetically.

```python
s_dept = pd.Series(['Sales', 'Admin', 'Sales', 'IT'])
codes_sorted, uniques_sorted = pd.factorize(s_dept, sort=True)
print('Sorted Codes:', codes_sorted)
print('Sorted Uniques:', uniques_sorted)

# Self-validating assertions
assert uniques_sorted[0] == 'Admin'
```
