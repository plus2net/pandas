# Pandas DataFrame `ffill()` Method Deep-Dive Practice

## 1. Basic Column Forward Fill

Propagate valid observations forward down columns.

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({'A': [1, np.nan, 3], 'B': [np.nan, 5, 6]})
result = df.ffill()
print(result)

# Self-validating assertions
assert result.loc[1, 'A'] == 1.0
assert pd.isna(result.loc[0, 'B'])
assert result.loc[1, 'B'] == 5.0
````

**Expected output**

```text
     A    B
0  1.0  NaN
1  1.0  5.0
2  3.0  6.0
```

## 2. Limiting Consecutive Forward Fills (`limit=2`)

Prevent infinite forward propagation across sensor outages using limit caps.

````python
df_sensor = pd.DataFrame({
    'TEMP_C': [21.5, np.nan, np.nan, np.nan, 24.0]
}, index=pd.date_range('2026-01-01 08:00', periods=5, freq='h'))

df_capped = df_sensor.ffill(limit=2)
print(df_capped)

# Self-validating assertions
assert df_capped.iloc[1, 0] == 21.5
assert df_capped.iloc[2, 0] == 21.5
assert pd.isna(df_capped.iloc[3, 0])
````

**Expected output**

```text
                     TEMP_C
2026-01-01 08:00:00    21.5
2026-01-01 09:00:00    21.5
2026-01-01 10:00:00    21.5
2026-01-01 11:00:00     NaN
2026-01-01 12:00:00    24.0
```

## 3. Row-Wise Forward Fill (`axis=1`)

Propagate missing survey response data horizontally across columns.

````python
df_survey = pd.DataFrame({
    'STAGE_1': ['Pass', 'Pending'],
    'STAGE_2': [np.nan, 'Pass'],
    'STAGE_3': [np.nan, np.nan]
})

df_row_filled = df_survey.ffill(axis=1)
print(df_row_filled)

# Self-validating assertions
assert df_row_filled.loc[0, 'STAGE_3'] == 'Pass'
assert df_row_filled.loc[1, 'STAGE_3'] == 'Pass'
````

**Expected output**

```text
  STAGE_1  STAGE_2 STAGE_3
0    Pass     Pass    Pass
1 Pending     Pass    Pass
```

## 4. GroupBy-Isolated Forward Fill

Prevent cross-category data leakage by forward filling inside group boundaries.

````python
df_multi = pd.DataFrame({
    'TICKER': ['AAPL', 'AAPL', 'GOOG', 'GOOG'],
    'PRICE': [180.0, np.nan, np.nan, 140.0]
})

df_multi['PRICE'] = df_multi.groupby('TICKER')['PRICE'].ffill()
print(df_multi)

# Self-validating assertions
assert df_multi.loc[1, 'PRICE'] == 180.0
assert pd.isna(df_multi.loc[2, 'PRICE'])
````

**Expected output**

```text
  TICKER  PRICE
0   AAPL  180.0
1   AAPL  180.0
2   GOOG    NaN
3   GOOG  140.0
```
