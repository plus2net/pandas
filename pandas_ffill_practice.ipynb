## 1. Basic Forward Fill

Propagate valid observations forward down columns.

````python
import pandas as pd
import numpy as np

df = pd.DataFrame({'A': [1, np.nan, 3], 'B': [np.nan, 5, 6]})
result = df.ffill()
print(result)

# Self-validating assertion
assert result.loc[1, 'A'] == 1.0
assert pd.isna(result.loc[0, 'B'])
````

**Expected output**

```text
     A    B
0  1.0  NaN
1  1.0  5.0
2  3.0  6.0
```
