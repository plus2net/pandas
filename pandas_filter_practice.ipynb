## 1. Substring Label Matching (`like`)

Filter columns containing the substring `'SCORE'`.

````python
import pandas as pd

df = pd.DataFrame({'MATH_SCORE': [85, 90], 'ENG_SCORE': [78, 88], 'AGE': [20, 21]})
result = df.filter(like='SCORE', axis=1)
print(result)

# Self-validating assertion
assert list(result.columns) == ['MATH_SCORE', 'ENG_SCORE']
assert 'AGE' not in result.columns
````

**Expected output**

```text
   MATH_SCORE  ENG_SCORE
0          85         78
1          90         88
```

## 2. Partial Year Substring Filter

Isolate quarterly sales columns containing `'2025'`.

````python
df_sales = pd.DataFrame({
    'Q1_2025': [100, 150],
    'Q2_2025': [200, 250],
    'ANNUAL_TARGET': [500, 600]
})

df_quarterly = df_sales.filter(like='2025')
print(df_quarterly)

# Self-validating assertion
assert list(df_quarterly.columns) == ['Q1_2025', 'Q2_2025']
````

**Expected output**

```text
   Q1_2025  Q2_2025
0      100      200
1      150      250
```

## 3. Regular Expression Column Matching (`regex`)

Filter column labels starting with `'cat'` using regular expressions.

````python
df_data = pd.DataFrame({
    'cat_a': [1, 2],
    'cat_b': [3, 4],
    'dog_a': [5, 6]
})

df_cats = df_data.filter(regex='^cat')
print(df_cats)

# Self-validating assertion
assert list(df_cats.columns) == ['cat_a', 'cat_b']
````

**Expected output**

```text
   cat_a  cat_b
0      1      3
1      2      4
```
