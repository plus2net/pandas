# Extracting GitHub Repository Statistics into a Pandas DataFrame

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-github-api.php](https://www.plus2net.com/python/pandas-github-api.php)

Run cells in order in Google Colab. Uses public GitHub REST API endpoints.

## 1. GitHub API to DataFrame

Fetch public repository metrics and load them into a Pandas DataFrame.

```python
import requests
import pandas as pd

url = "https://api.github.com/repos/pandas-dev/pandas"
headers = {"Accept": "application/vnd.github.v3+json", "User-Agent": "Plus2Net-Colab"}
response = requests.get(url, headers=headers)

if response.status_code == 200:
    repo_data = response.json()
    df = pd.DataFrame([{
        'Repository': repo_data['full_name'],
        'Stars': repo_data['stargazers_count'],
        'Forks': repo_data['forks_count'],
        'Open_Issues': repo_data['open_issues_count'],
        'Language': repo_data['language']
    }])
    print(df)
    assert not df.empty
    assert 'Stars' in df.columns
else:
    print(f"Error: {response.status_code}")
```

## 2. Flattening Contributors Data

Use `pd.json_normalize()` to flatten lists of contributors.

```python
url_contrib = "https://api.github.com/repos/pandas-dev/pandas/contributors?per_page=5"
response_contrib = requests.get(url_contrib, headers=headers)
if response_contrib.status_code == 200:
    df_contrib = pd.json_normalize(response_contrib.json())
    print(df_contrib[['login', 'contributions']])
    assert len(df_contrib) > 0
```

## 3. Saving Repository Metrics
Export the DataFrame to CSV and SQLite database.

```python
import sqlite3

df.to_csv("github_repo_stats.csv", index=False)
conn = sqlite3.connect("github_data.db")
df.to_sql("repo_metrics", conn, if_exists="replace", index=False)
conn.close()
print("GitHub metrics successfully saved and verified!")
assert not df.empty
```
