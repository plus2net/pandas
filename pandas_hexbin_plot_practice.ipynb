### 1. Basic Hexbin Plot
Generate 10,000 bivariate points and plot a 2D hexbin density chart.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
n = 10000
df = pd.DataFrame({
    'x': np.random.randn(n),
    'y': np.random.randn(n)
})

ax = df.plot.hexbin(x='x', y='y', gridsize=20, figsize=(6.5, 4.5), title="2D Gaussian Hexbin Density")
plt.show()
```

### 2. Logarithmic Scaling and Colormaps
Use `bins='log'` and `cmap='plasma'` to enhance density contrast.


```python
df.plot.hexbin(
    x='x',
    y='y',
    gridsize=25,
    bins='log',
    cmap='plasma',
    mincnt=1,
    figsize=(7, 4.5),
    title="Logarithmic Hexbin Density (bins='log')"
)
plt.show()
```

### 3. Practical Workflow: Web Analytics Engagement Density
Analyze session duration vs pages viewed across 10,000 visitors.


```python
np.random.seed(42)
n_sessions = 10000

dur_casual = np.random.normal(3.5, 1.2, int(n_sessions * 0.8))
pages_casual = np.random.normal(2.5, 0.8, int(n_sessions * 0.8))

dur_power = np.random.normal(18.0, 4.0, int(n_sessions * 0.2))
pages_power = np.random.normal(12.0, 2.5, int(n_sessions * 0.2))

df_web = pd.DataFrame({
    "Duration_Minutes": np.clip(np.concatenate([dur_casual, dur_power]), 0.5, 45.0),
    "Pages_Viewed": np.clip(np.concatenate([pages_casual, pages_power]), 1.0, 25.0)
})

print(f"Total Sessions: {len(df_web)}")
print(f"Median Duration: {df_web['Duration_Minutes'].median():.2f} mins")
print(f"Median Pages Viewed: {df_web['Pages_Viewed'].median():.2f}")

assert len(df_web) == 10000

ax = df_web.plot.hexbin(
    x="Duration_Minutes",
    y="Pages_Viewed",
    gridsize=30,
    cmap="YlGnBu",
    bins="log",
    figsize=(7.5, 5),
    title="User Engagement Density (10,000 Web Sessions)"
)
plt.xlabel("Session Duration (Minutes)")
plt.ylabel("Pages Viewed")
plt.show()
```

### 4. Hands-on Exercise: Logistics Fleet Route Density
Visualize 5,000 shipment routes by distance and delivery duration.


```python
np.random.seed(101)
n = 5000
distance = np.random.normal(15, 4, n)
duration = np.random.normal(35, 10, n)

df_logistics = pd.DataFrame({
    "Distance_KM": np.clip(distance, 1, 35),
    "Duration_Mins": np.clip(duration, 5, 75)
})

assert len(df_logistics) == 5000
assert df_logistics["Distance_KM"].min() >= 1.0

print(f"Mean Distance: {df_logistics['Distance_KM'].mean():.2f} km")
print(f"Mean Duration: {df_logistics['Duration_Mins'].mean():.2f} mins")

df_logistics.plot.hexbin(
    x="Distance_KM",
    y="Duration_Mins",
    gridsize=25,
    cmap="cividis",
    mincnt=1,
    figsize=(7, 4.5),
    title="Delivery Route Density"
)
plt.xlabel("Route Distance (km)")
plt.ylabel("Transit Time (minutes)")
plt.show()
```
