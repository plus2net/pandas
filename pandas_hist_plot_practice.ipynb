## 1. Basic Histogram with Custom Bins

A histogram groups values into equal-width intervals (bins) and displays the frequency count.

````python
import pandas as pd
import matplotlib.pyplot as plt

my_dict = {
    "NAME": ["Ravi", "Raju", "Alex", "Ron", "Tor"],
    "MATH": [30, 40, 50, 30, 50]
}
df = pd.DataFrame(data=my_dict)
print(df)

df["MATH"].plot.hist(bins=3, title="Math Score Distribution", figsize=(6, 3.5))
plt.tight_layout()
plt.show()
assert len(df) == 5
assert df["MATH"].sum() == 200
````

**Expected output**

```text
   NAME  MATH
0  Ravi    30
1  Raju    40
2  Alex    50
3   Ron    30
4   Tor    50
```

## 2. Multi-Series Histogram Overlay with Transparency

Pass `alpha=0.5` so overlapping bars are visible across multiple columns.

````python
scores = pd.DataFrame({
    "Math": [30, 40, 50, 30, 50, 60, 70, 75],
    "Science": [55, 60, 65, 70, 75, 80, 85, 90]
})
scores.plot.hist(bins=5, alpha=0.5, figsize=(7, 4), title="Math vs Science Score Distribution")
plt.tight_layout()
plt.show()
assert scores["Math"].mean() == 50.625
assert scores["Science"].mean() == 72.5
````

## 3. Practical Workflow: Employee Salary Distribution Analysis

Examine salary brackets and payroll statistics with assertions.

````python
salary_data = pd.DataFrame({
    "Employee_ID": range(1, 11),
    "Salary": [45000, 52000, 48000, 75000, 62000, 58000, 89000, 64000, 71000, 95000]
})

assert len(salary_data) == 10
total_payroll = salary_data["Salary"].sum()
assert total_payroll == 659000
print("Total Payroll:", total_payroll)
print("Median Salary:", salary_data["Salary"].median())
assert salary_data["Salary"].median() == 63000.0

ax = salary_data["Salary"].plot.hist(bins=5, figsize=(7, 4), color="darkcyan", edgecolor="white", grid=True, title="Salary Distribution")
ax.set_xlabel("Salary Bracket ($)")
ax.set_ylabel("Employee Count")
plt.tight_layout()
plt.show()
````

## Exercise: Customer Age Group Distribution

Create a DataFrame representing 8 registered customers with ages: `[22, 25, 29, 34, 38, 42, 46, 52]`. Plot a histogram with 4 bins and an explicit edge color (`edgecolor="black"`), enable gridlines, title the plot *"Customer Age Distribution"*, and calculate the mean customer age.

## Exercise Solution

Verify your implementation with assertions below:

````python
customers = pd.DataFrame({
    "Customer_ID": range(101, 109),
    "Age": [22, 25, 29, 34, 38, 42, 46, 52]
})

mean_age = customers["Age"].mean()
assert mean_age == 36.0
print("Average Customer Age:", mean_age)

ax = customers["Age"].plot.hist(bins=4, color="coral", edgecolor="black", grid=True, figsize=(6, 3.5), title="Customer Age Distribution")
ax.set_xlabel("Age")
ax.set_ylabel("Customer Count")
plt.tight_layout()
plt.show()
````

**Expected output**

```text
Average Customer Age: 36.0
```
