## 1. Above-the-Fold Basic Column Maximum Index Lookup

Find index labels corresponding to column maximum values.

```python
import pandas as pd

df = pd.DataFrame({
    'Sales': [250, 450, 300],
    'Profit': [40, 90, 60]
}, index=['Store_A', 'Store_B', 'Store_C'])

result = df.idxmax()
print(result)

# Self-validating assertions
assert result['Sales'] == 'Store_B'
assert result['Profit'] == 'Store_B'
```

## 2. Locating Highest Value Across Columns (`axis=1`)

Evaluate each row individually and return the column header containing the maximum value.

```python
df_scores = pd.DataFrame({
    'Math': [88, 92, 79],
    'Science': [95, 85, 91],
    'English': [70, 89, 95]
}, index=['Student_1', 'Student_2', 'Student_3'])

df_scores['Top_Subject'] = df_scores.idxmax(axis=1)
print(df_scores)

# Self-validating assertions
assert df_scores.loc['Student_1', 'Top_Subject'] == 'Science'
```

## 3. GroupBy-Isolated Peak Index Lookup

Pull entire row records corresponding to group peak values.

```python
df_sales = pd.DataFrame({
    'Region': ['East', 'East', 'West', 'West'],
    'Rep': ['Alice', 'Bob', 'Charlie', 'David'],
    'Revenue': [5000, 7500, 6200, 4800]
})

idx_max = df_sales.groupby('Region')['Revenue'].idxmax()
print(df_sales.loc[idx_max])

# Self-validating assertions
assert df_sales.loc[idx_max[0], 'Rep'] == 'Bob'
```
