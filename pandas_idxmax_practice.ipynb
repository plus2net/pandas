## 1. Basic Column Maximum Index Lookup

Find index labels corresponding to column maximum values.

````python
import pandas as pd

df = pd.DataFrame({
    'Sales': [250, 450, 300],
    'Profit': [40, 90, 60]
}, index=['Store_A', 'Store_B', 'Store_C'])

result = df.idxmax()
print(result)

# Self-validating assertion
assert result['Sales'] == 'Store_B'
assert result['Profit'] == 'Store_B'
````

**Expected output**

```text
Sales     Store_B
Profit    Store_B
dtype: object
```
