## 1. Basic Column Minimum Index Lookup

Find index labels corresponding to column minimum values.

````python
import pandas as pd

df = pd.DataFrame({
    'Latency_ms': [120, 45, 88],
    'Cost_USD': [15, 25, 10]
}, index=['Server_A', 'Server_B', 'Server_C'])

result = df.idxmin()
print(result)

# Self-validating assertion
assert result['Latency_ms'] == 'Server_B'
assert result['Cost_USD'] == 'Server_C'
````

**Expected output**

```text
Latency_ms    Server_B
Cost_USD      Server_C
dtype: object
```
