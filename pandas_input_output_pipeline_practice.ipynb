### 1. Ingesting and Exporting CSV Data
Read delimited text records and export with custom compression.


```python
import pandas as pd
import io

csv_data = """id,customer,region,amount
1,Acme Corp,North,1200
2,Beta LLC,South,850
3,Acme Corp,North,450
4,Gamma Inc,East,2100
5,Beta LLC,South,620
"""
df = pd.read_csv(io.StringIO(csv_data))
print(df)
assert len(df) == 5
```

### 2. Relational Database Round-Trip (SQLite)
Persist DataFrame records into a relational database and extract with SQL queries.


```python
import sqlite3

conn = sqlite3.connect(":memory:")
df.to_sql("sales", conn, index=False)

q = """
SELECT region,
COUNT(*) as order_count,
SUM(amount) as total_sales,
AVG(amount) as avg_order
FROM sales
GROUP BY region
ORDER BY total_sales DESC
"""
df_summary = pd.read_sql(q, conn)
print(df_summary)

assert len(df_summary) == 3
assert df_summary["total_sales"].sum() == 5220
```

### 3. Exporting to JSON and HTML
Convert aggregated tables to JSON records and styled HTML table markups.


```python
json_out = df_summary.to_json(orient="records", indent=2)
print("JSON Output:\n", json_out)

html_out = df_summary.to_html(index=False)
print("\nHTML Table length:", len(html_out))
```

### 4. Hands-on Exercise: Warehouse Inventory Pipeline
Filter warehouse stocks via SQL HAVING and export formatted outputs.


```python
stock_data = {
    "product": ["Widget", "Gadget", "Widget", "Tool", "Gadget"],
    "warehouse": ["Alpha", "Beta", "Alpha", "Gamma", "Beta"],
    "quantity": [70, 60, 85, 40, 55]
}
df_stock = pd.DataFrame(stock_data)
df_stock.to_sql("stock", conn, index=False)

q_warehouse = """
SELECT warehouse, SUM(quantity) as total_qty
FROM stock
GROUP BY warehouse
HAVING total_qty > 100
ORDER BY total_qty DESC
"""
df_result = pd.read_sql(q_warehouse, conn)
print(df_result)

assert len(df_result) == 2
assert df_result["total_qty"].tolist() == [155, 115]
print("All assertions passed cleanly!")
```
