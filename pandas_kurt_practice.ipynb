# Pandas: Calculate Kurtosis with kurt()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-kurt.php](https://www.plus2net.com/python/pandas-kurt.php)

Run cells in order in Google Colab.

## Create numerical sample DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "A": [1, 2, 3, 4, 5, 6, 7, 8, 9],
    "B": [2, 2, 2, 5, 5, 5, 8, 8, 8]
})
print(df)
assert df.shape == (9, 2)
```

## Basic kurtosis calculation

```python
res = df.kurt()
print(res)
assert res["A"] == -1.2
```

## Kurtosis on a single Series

```python
k_val = df["A"].kurtosis()
print(f"Excess Kurtosis: {k_val}")
assert k_val == -1.2
```

## Exercise solution

```python
vals = pd.Series([10, 10, 10, 10, 50], name="VAL")
answer = vals.kurt()
print(answer)
assert answer == 5.0
```
