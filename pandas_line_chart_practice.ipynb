## 1. Basic Single-Series Line Chart

Calling `df.plot.line()` plots numeric columns against the row index by default. Here we plot student mark trends.

````python
import pandas as pd
import matplotlib.pyplot as plt

my_dict = {
    "NAME": ["Ravi", "Raju", "Alex", "Ron", "Geek", "Kim"],
    "MARK": [20, 30, 40, 30, 40, 50]
}
df = pd.DataFrame(data=my_dict)
print(df)

ax = df.plot.line(title="Student Mark")
plt.show()
assert len(df) == 6
assert df["MARK"].sum() == 210
````

**Expected output**

```text
   NAME  MARK
0  Ravi    20
1  Raju    30
2  Alex    40
3   Ron    30
4  Geek    40
5   Kim    50
```

## 2. Multi-Series Line Plot

When multiple columns are specified, Pandas draws a distinct line for each measure and generates an automatic legend.

````python
df_subjects = pd.DataFrame({
    "NAME": ["Ravi", "Raju", "Alex", "Ron", "Geek", "Kim"],
    "MATH": [20, 30, 40, 30, 40, 50],
    "ENGLISH": [80, 70, 60, 70, 90, 80]
})
print(df_subjects)
df_subjects.plot.line(x="NAME", y=["MATH", "ENGLISH"], title="Subject Comparison", figsize=(6, 3), grid=True)
plt.show()
assert df_subjects["MATH"].mean() == 35.0
assert df_subjects["ENGLISH"].mean() == 75.0
````

**Expected output**

```text
   NAME  MATH  ENGLISH
0  Ravi    20       80
1  Raju    30       70
2  Alex    40       60
3   Ron    30       70
4  Geek    40       90
5   Kim    50       80
```

## 3. Line Styling: Patterns, Grids, and Legends

You can customize line styles using Matplotlib syntax (e.g. `'-.'`, `':'`, `'--'`), enable gridlines, and format the figure size.

````python
my_style = ["-.", ":"]
ax = df_subjects.plot.line(x="NAME", y=["MATH", "ENGLISH"], style=my_style, figsize=(6, 3), grid=True, title="Custom Line Styles")
plt.show()
assert len(my_style) == 2
````

## 4. Dual-Scale Comparison with Secondary Y-Axis

When comparing volume and dollar values, plotting them on a single scale compresses the smaller metric. Use `secondary_y=True` to create a dedicated right-hand Y-axis.

````python
sales_data = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Units_Sold": [120, 150, 190, 240, 290, 340],
    "Revenue": [12000, 15500, 18200, 26000, 31000, 38500]
})
print(sales_data)

ax = sales_data.plot.line(
    x="Month",
    secondary_y=["Revenue"],
    mark_right=True,
    figsize=(7, 4),
    title="Units vs Revenue Trend",
    grid=True
)
ax.set_ylabel("Units Sold (Left Axis)")
plt.show()
assert sales_data["Revenue"].sum() == 141200
assert sales_data["Units_Sold"].sum() == 1330
````

**Expected output**

```text
  Month  Units_Sold  Revenue
0   Jan         120    12000
1   Feb         150    15500
2   Mar         190    18200
3   Apr         240    26000
4   May         290    31000
5   Jun         340    38500
```

## Exercise: Weekly Website Visitor Trends

Create a DataFrame with 5 weekdays (`Mon` to `Fri`) and daily visitor counts: `[100, 150, 130, 180, 220]`. Plot a line chart with Day on the X-axis, enable gridlines, title the chart *"Weekly Traffic"*, and calculate the total weekly visitor count.

## Exercise Solution

Verify your implementation with assertions below:

````python
traffic = pd.DataFrame({
    "Day": ["Mon", "Tue", "Wed", "Thu", "Fri"],
    "Visitors": [100, 150, 130, 180, 220]
})

assert traffic["Visitors"].sum() == 780
print("Total Visitors:", traffic["Visitors"].sum())

traffic.plot.line(x="Day", y="Visitors", title="Weekly Traffic", grid=True, figsize=(6, 3))
plt.show()
````

**Expected output**

```text
Total Visitors: 780
```
