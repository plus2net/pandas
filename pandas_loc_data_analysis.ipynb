## Use square brackets and labels

loc is an indexer, not a function. Use <code>df.loc[row_selector, column_selector]</code>. Integers identify index labels; <a href="https://www.plus2net.com/python/pandas-dataframe-iloc.php">iloc</a> uses integer positions. A colon selects all labels on an axis. Boolean Series align by index labels.

## Select one or several columns

A single column label returns a Series; a list of column labels returns a DataFrame.

````python
df = student_data()
print(df.loc[:, "class"].head())
print(df.loc[:, ["class", "name"]].head())
assert df.loc[:, ["class", "name"]].shape == (35, 2)
````

## Select rows and cells by labels

With the default index, label 2 is the third row. Use a list to retain a DataFrame. Use <a href="https://www.plus2net.com/python/pandas-dataframe-at.php">at</a> for a single cell on uniquely labelled axes.

````python
df = student_data()
print(df.loc[0, ["class", "name"]])
print(df.loc[[2, 3], ["class", "name"]])
print(df.loc[[2, 3], :])
print("Arnold mark:", df.loc[2, "mark"])
assert df.loc[2, "name"] == "Arnold"
````

## Select inclusive label ranges

Label slices include both endpoints. An unsorted unique index with existing bounds follows its current order, not alphabetical order. Nonunique unsorted bounds can be ambiguous; for value ranges use a mask or sort the index under an explicit rule.

````python
by_id = student_data().set_index("id").sort_index()
print(by_id.loc[4:6])
named = student_data().set_index("name")
print(named.loc["Ronald":"Bigy"])
assert by_id.loc[4:6].index.tolist() == [4, 5, 6]
assert named.loc["Ronald":"Bigy"].index.tolist() == ["Ronald", "Recky", "Kty", "Bigy"]
````

## Select with callable expressions

A callable receives the current table. Reference a name column before moving it to the index; afterwards use the index explicitly. Boolean results must have compatible labels or the correct array length.

````python
df = student_data()
print(df.loc[lambda x: x["name"].eq("Arnold")])
print(df.loc[lambda x: x["mark"].gt(90)])
print(df.loc[lambda x: x["gender"].eq("male")].head())
print(df.loc[lambda x: x["name"].str.len().gt(10)].head())
named = df.set_index("name")
print(named.loc[lambda x: x.index.str.len() > 10].head())
````

## Match a complete name or a substring

Use eq() for exact values. <a href="https://www.plus2net.com/python/pandas-str-contains.php">str.contains()</a> can match partial text. Literal matching with na=False makes missing names nonmatches.

````python
df = student_data()
print(df.loc[df["name"].eq("Alex John")])
print(df.loc[df["name"].str.contains("De", case=False, regex=False, na=False)])
assert df.loc[df["name"].eq("Alex John"), "id"].tolist() == [6]
````

## Match or exclude an ending

Specify missing-value treatment before inversion. With na=False, inversion includes missing names; add notna() when only known names should be kept.

````python
df = student_data()
ends = df["name"].str.endswith("John", na=False)
print(df.loc[ends])
print("Not ending with John IDs:", df.loc[~ends, "id"].tolist())
assert df.loc[ends, "id"].tolist() == [6, 10, 21]
````

## Match or exclude a beginning

startswith() is a literal prefix check. Case sensitivity applies to the text as stored.

````python
df = student_data()
starts = df["name"].str.startswith("Jo", na=False)
print(df.loc[starts])
print("Other IDs:", df.loc[~starts, "id"].tolist())
assert df.loc[starts, "id"].tolist() == [1, 5]
````

## Match any of several strings

Regex alternation matches either term. Escape terms when constructing a regex from literal strings, or combine separate literal masks with |.

````python
df = student_data()
terms = ["Jo", "Al"]
pattern = "|".join(re.escape(term) for term in terms)
any_term = df["name"].str.contains(pattern, na=False)
print(df.loc[any_term])
print("Neither term IDs:", df.loc[~any_term, "id"].tolist())
print("No Jo IDs:", df.loc[~df["name"].str.contains("Jo", regex=False, na=False), "id"].tolist())
````

## Require both strings

Use &amp; for AND and | for OR; Python and/or do not combine whole Series. Both masks must describe the same table.

````python
df = student_data()
jo = df["name"].str.contains("Jo", regex=False, na=False)
al = df["name"].str.contains("Al", regex=False, na=False)
print("Both:")
print(df.loc[jo & al])
print("Either IDs:", df.loc[jo | al, "id"].tolist())
assert df.loc[jo & al, "id"].tolist() == [6]
````

**Expected output**

```text

```

## Combine text and numeric conditions

Parentheses separate comparisons. Build named masks when the rule becomes hard to read. See <a href="https://www.plus2net.com/python/pandas-dataframe-query.php">query()</a> for fixed expression-based filters.

````python
df = student_data()
print(df.loc[df["name"].str.contains("Mike", regex=False, na=False) | df["id"].eq(6)])
by_id = df["id"].eq(5)
by_name = df["name"].str.contains("Al", case=False, regex=False, na=False)
low_mark = df["mark"].lt(50)
print(df.loc[by_id | by_name | low_mark])
````

## Understand Boolean Series alignment

loc aligns a Boolean Series to the row index. A Boolean list or array is positional and must match the row count. Build masks from the current table to avoid unintended alignment.

````python
small = pd.DataFrame({"amount": [10, 20, 30]}, index=[101, 102, 103])
mask = pd.Series([True, False, True], index=[103, 101, 102])
print(small.loc[mask])
print("Array selection:")
print(small.loc[mask.to_numpy()])
assert small.loc[mask].index.tolist() == [102, 103]
````

**Expected output**

```text
     amount
102      20
103      30
Array selection:
     amount
101      10
103      30
```

## Add a row with an unused label

df.loc[len(df)] is appropriate only when that label is unused, as with this fresh default index. On a filtered or differently indexed table it can overwrite an existing label. Validate the label and match values to the column order.

````python
df = student_data()
new_label = len(df)
if new_label in df.index:
    raise ValueError("New label already exists")
new_list = [36, "new name", "Four", 77, "male"]
df.loc[new_label] = new_list
print(df.tail(2))
assert len(df) == 36 and df.loc[new_label, "id"] == 36
````

**Expected output**

```text
    id        name class  mark  gender
34  35  Rows Noump   Six    88  female
35  36    new name  Four    77    male
```

## Update a column under a condition

Assign with one loc operation. Each example below uses a fresh table. In this demonstration the rule sets high marks to 100; a real grade change requires an approved rule.

````python
df = student_data()
df.loc[df["mark"].ge(90), "mark"] = 100
print(df.loc[df["mark"].eq(100)])
df = student_data()
condition = df["mark"].ge(90) & df["class"].eq("Six")
df.loc[condition, "mark"] = 100
print("Class Six updates:")
print(df.loc[condition])
````

## Update a category or name directly

Changing the in-memory DataFrame does not write back to the original CSV. Keep source values or a change record when corrections must be traceable.

````python
df = student_data()
condition = df["mark"].ge(90) & df["class"].eq("Six")
df.loc[condition, "class"] = "Seven"
print(df.loc[condition])
df = student_data()
df.loc[df["name"].eq("Rows Noump"), "name"] = "New Name"
print(df.tail(1))
assert df.loc[34, "name"] == "New Name"
````

## Practical example: select paid orders and review unknown amounts

This synthetic table preserves invalid amounts for review. Clean numeric values before combining selection rules. The <a href="https://www.plus2net.com/python/pandas-data-cleaning.php">Data Cleaning hub</a> covers related validation steps.

````python
raw = pd.DataFrame({
    "order_id": [101, 102, 103, 104, 105],
    "status": ["paid", "paid", "test", "paid", "paid"],
    "amount": ["20", "35", "99", "bad", "10"]
})
amount = pd.to_numeric(raw["amount"], errors="coerce")
review = raw.loc[amount.isna()].copy()
eligible = raw["status"].eq("paid") & amount.ge(20)
selected = raw.loc[eligible, ["order_id", "amount"]].copy()
selected["amount"] = amount.loc[selected.index]
print(selected)
print("Review IDs:", review["order_id"].tolist())
assert selected["order_id"].tolist() == [101, 102]
assert selected["amount"].sum() == 55
````

**Expected output**

```text
   order_id  amount
0       101    20.0
1       102    35.0
Review IDs: [104]
```

## Assign a review flag and export the intended records

Set flags on the working table with one loc assignment. Export selected orders separately from the review data; unknown amounts should not become zero merely to pass a numeric rule. See <a href="https://www.plus2net.com/python/pandas-input-output.php">Input and Output</a>.

````python
working = raw.copy()
working["needs_review"] = False
working.loc[amount.isna(), "needs_review"] = True
print(working)
print(selected.to_csv(index=False).rstrip())
assert working["needs_review"].sum() == 1
# Optional Colab exports:
# selected.to_csv("selected_orders.csv", index=False)
# review.to_csv("review_orders.csv", index=False)
````

## Common questions

<strong>Why did loc[2] select a different row after indexing?</strong> 2 is a label. <strong>Are slice endpoints included?</strong> Yes, label slices include both bounds when resolvable. <strong>Why did selection return multiple rows?</strong> The row label may be duplicated. <strong>Why does a mask fail?</strong> Its index may be unalignable, or its array length may differ. <strong>Why did assignment replace a row?</strong> The requested label already existed. <strong>Should I edit a chained selection?</strong> Assign directly through the source DataFrame loc indexer.
