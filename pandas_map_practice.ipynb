## 1. Quick Dictionary Mapping

Map animal keys to species descriptions; unmapped keys yield `NaN`.

````python
import pandas as pd

s = pd.Series(['cat', 'dog', 'cow'])
mapping = {'cat': 'feline', 'dog': 'canine'}
result = s.map(mapping)
print(result)

# Self-validating assertion
assert result[0] == 'feline'
assert result[1] == 'canine'
assert pd.isna(result[2])
````

**Expected output**

```text
0    feline
1    canine
2       NaN
dtype: object
```

## 2. Categorical Code to Grade Name Mapping

Map integer student grade codes to descriptive strings.

````python
df_students = pd.DataFrame({
    'NAME': ['Ravi', 'Ananya', 'Alex'],
    'GRADE_CODE': [1, 2, 3]
})

grade_map = {1: 'Grade A', 2: 'Grade B', 3: 'Grade C'}
df_students['GRADE_NAME'] = df_students['GRADE_CODE'].map(grade_map)
print(df_students)

# Self-validating assertion
assert df_students.loc[0, 'GRADE_NAME'] == 'Grade A'
assert df_students.loc[1, 'GRADE_NAME'] == 'Grade B'
````

**Expected output**

```text
     NAME  GRADE_CODE GRADE_NAME
0    Ravi           1    Grade A
1  Ananya           2    Grade B
2    Alex           3    Grade C
```

## 3. String Formatting with Lambdas

Format numerical price figures as currency strings.

````python
prices = pd.Series([19.99, 5.5, 100.0])
formatted_prices = prices.map(lambda x: f"${x:.2f}")
print(formatted_prices)

# Self-validating assertion
assert formatted_prices[0] == '$19.99'
assert formatted_prices[1] == '$5.50'
````

**Expected output**

```text
0     $19.99
1      $5.50
2    $100.00
dtype: object
```
