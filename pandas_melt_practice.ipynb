# Pandas: Unpivot DataFrames with melt()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-melt.php](https://www.plus2net.com/python/pandas-melt.php)

Run cells in order in Google Colab.

## Create wide-format sample data

```python
import pandas as pd
df = pd.DataFrame({
    "NAME": ["Ravi", "Raju", "Alex"],
    "ID": [1, 2, 3],
    "MATH": [30, 40, 50],
    "ENGLISH": [20, 30, 40]
})
print(df)
assert df.shape == (3, 4)
```

## Basic melting with id_vars

```python
melted = pd.melt(df, id_vars=["NAME", "ID"])
print(melted)
assert len(melted) == 6
```

## Selecting specific value columns

```python
specific = pd.melt(df, id_vars=["NAME"], value_vars=["MATH"])
print(specific)
assert len(specific) == 3
```

## Customizing variable and value names

```python
custom = pd.melt(df, id_vars=["NAME"], value_vars=["MATH", "ENGLISH"], 
                  var_name="SUBJECT", value_name="MARKS")
print(custom)
assert "SUBJECT" in custom.columns and "MARKS" in custom.columns
```

## Resetting index during melt

```python
reset = pd.melt(df, id_vars=["NAME"], value_vars=["MATH"], ignore_index=True)
print(reset)
assert reset.index.tolist() == [0, 1, 2]
```

## Exercise solution

```python
answer = pd.melt(df, id_vars=["NAME", "ID"], value_vars=["MATH", "ENGLISH"],
                  var_name="COURSE", value_name="SCORE", ignore_index=True)
print(answer)
assert len(answer) == 6 and "COURSE" in answer.columns
```
