# Pandas: Calculate Statistical Mode with mode()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-mode.php](https://www.plus2net.com/python/pandas-mode.php)

Run cells in order in Google Colab.

## Create sample Series & DataFrame

```python
import pandas as pd

s = pd.Series([10, 20, 20, 30, 40], name="SCORES")
print(s)
assert s.shape == (5,)
```

## Basic mode calculation

```python
m_val = s.mode()
print(m_val)
assert m_val.iloc[0] == 20
```

## Handling multimodal data (multiple modes)

```python
multi_s = pd.Series([10, 10, 20, 20, 30], name="VALS")
multi_mode = multi_s.mode()
print(multi_mode)
assert list(multi_mode) == [10, 20]
```

## Exercise solution

```python
fruits = pd.Series(["apple", "banana", "apple", "cherry", "banana", "apple"], name="FRUIT")
answer = fruits.mode()
print(answer)
assert answer.iloc[0] == "apple"
```
