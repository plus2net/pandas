## True means present, False means missing

notnull() returns a Boolean mask with the same shape as the source. None and NaN produce False. notna() is an equivalent name; isna()/isnull() return the inverse mask.

````python
import pandas as pd
import numpy as np
df = pd.DataFrame({"NAME": ["Ravi", "Raju", "Alex", None, "King", None], "ID": [1, 2, np.nan, 4, 5, 6], "MATH": [80, 40, 70, 70, 82, 30], "ENGLISH": [81, 70, 40, 50, np.nan, 30]})
print(df.notnull())
assert df.notnull().equals(df.notna())
assert df.notnull().equals(~df.isna())
assert not df.notnull().loc[3, "NAME"]
````

**Expected output**

```text
    NAME     ID  MATH  ENGLISH
0   True   True  True     True
1   True   True  True     True
2   True  False  True     True
3  False   True  True     True
4   True   True  True    False
5  False   True  True     True
```

## Check whether any value is present

A global any() over the Boolean array asks whether at least one present cell exists. It does not mean every row is complete or that a value is valid. A Series check inspects one column only.

````python
df = student_data()
print("Any present cell:", df.notnull().to_numpy().any())
print("Any name:", df["name"].notnull().any())
print("Any name or mark cell:", df[["name", "mark"]].notnull().to_numpy().any())
assert df.notnull().to_numpy().any()
````

**Expected output**

```text
Any present cell: True
Any name: True
Any name or mark cell: True
```

## Distinguish entirely missing, partly present and complete rows

any(axis=1) checks for at least one present field per row. all(axis=1) checks that every considered field is present. Negating any() selects entirely missing rows.

````python
df = student_data()
has_any = df.notnull().any(axis=1)
complete = df.notnull().all(axis=1)
print("Entirely missing rows:")
print(df.loc[~has_any])
print("Rows with at least one present field:")
print(df.loc[has_any])
print("Complete rows:", int(complete.sum()))
assert df.index[~has_any].tolist() == [7]
assert complete.sum() == 6
````

## Require several fields together

Combine masks with & for AND, or use all(axis=1) over required columns. Use | or any(axis=1) when at least one of the selected fields must be present. Parenthesise comparisons when mixing conditions.

````python
df = student_data()
mask = df["id"].notnull() & df["name"].notnull()
required = df[["id", "name"]].notnull().all(axis=1)
print(df.loc[required])
assert mask.equals(required)
assert required.sum() == 8
````

## Count present and missing cells

Summing Boolean masks counts True values. Sum once for column counts and twice for total cells. These are cell counts, not row counts. See <a href="https://www.plus2net.com/python/pandas-dataframe-count.php">DataFrame.count()</a> and <a href="https://www.plus2net.com/python/pandas-dataframe-sum.php">sum()</a>.

````python
df = student_data()
print("Present by column:")
print(df.notnull().sum())
print("Missing by column:")
print(df.isna().sum())
print("Total present cells:", df.notnull().sum().sum())
print("Total missing cells:", df.isna().sum().sum())
assert df.notnull().sum().tolist() == [9, 10, 11, 10, 10]
assert df.notnull().sum().sum() == 50
assert df.notnull().sum().sum() + df.isna().sum().sum() == df.size
````

## Blank text is present until normalised

Empty strings, whitespace, zero and False are not automatically missing. Define which blanks represent absent input, then normalise only the appropriate text fields.

````python
text = pd.Series(["Pen", "", "  ", None], dtype="string")
print("Before normalising blanks:")
print(text.notnull())
trimmed = text.str.strip()
normalised = trimmed.mask(trimmed.eq(""), pd.NA)
print("After normalising blanks:")
print(normalised.notnull())
print("Zero and False are present:", pd.Series([0, False], dtype="object").notnull().tolist())
assert normalised.notnull().tolist() == [True, False, False, False]
````

**Expected output**

```text
Before normalising blanks:
0     True
1     True
2     True
3    False
dtype: bool
After normalising blanks:
0     True
1    False
2    False
3    False
dtype: bool
Zero and False are present: [True, True]
```

## Present numeric data may still be invalid

Infinity is not missing, and bad numeric text is present before parsing. For usable amounts, parse values and separately require finite, non-negative numbers. notnull() alone is not a data-validation rule.

````python
raw_amount = pd.Series(["12.50", "bad", None, "inf", "-3"], dtype="string")
amount = pd.to_numeric(raw_amount, errors="coerce")
finite = pd.Series(np.isfinite(amount.to_numpy(dtype=float, na_value=np.nan)), index=amount.index)
usable = amount.notnull() & finite & amount.ge(0).fillna(False)
print(pd.DataFrame({"raw": raw_amount, "parsed": amount, "present": amount.notnull(), "usable": usable}))
assert usable.tolist() == [True, False, False, False, False]
````

**Expected output**

```text
     raw  parsed  present  usable
0  12.50    12.5     True    True
1    bad    <NA>    False   False
2   <NA>    <NA>    False   False
3    inf     inf     True   False
4     -3    -3.0     True   False
```

## Practical workflow: validate an order extract

This synthetic rule requires a non-blank product and a finite non-negative amount. Keep rejected rows rather than replacing uncertain amounts with zero. The raw source remains available for review.

````python
raw = pd.DataFrame({"order_id": [101, 102, 103, 104, 105], "product": [" Pen ", "", None, "Book", "Mug"], "amount": ["5", "7", "9", "bad", "inf"]})
product = raw["product"].astype("string").str.strip()
product = product.mask(product.eq(""), pd.NA)
amount = pd.to_numeric(raw["amount"], errors="coerce")
finite = pd.Series(np.isfinite(amount.to_numpy(dtype=float, na_value=np.nan)), index=raw.index)
keep = product.notnull() & amount.notnull() & finite & amount.ge(0).fillna(False)
clean = raw.loc[keep].copy()
clean["product"] = product.loc[keep]
clean["amount"] = amount.loc[keep]
review = raw.loc[~keep].copy()
print(clean)
print("Review rows:")
print(review)
assert clean["order_id"].tolist() == [101]
````

**Expected output**

```text
   order_id product  amount
0       101     Pen     5.0
Review rows:
   order_id product amount
1       102              7
2       103     NaN      9
3       104    Book    bad
4       105     Mug    inf
```

## Validate the result and record review reasons

Reason counts may overlap. Compare total retained and review rows separately. Filling or removing missing data is a later policy decision; see <a href="https://www.plus2net.com/python/pandas-dataframe-fillna.php">fillna()</a>, <a href="https://www.plus2net.com/python/pandas-dataframe-dropna.php">dropna()</a> and <a href="https://www.plus2net.com/python/pandas-dataframe-dropna-thresh.php">completeness thresholds</a>.

````python
print("Missing or blank products:", int(product.isna().sum()))
print("Missing or unparseable amounts:", int(amount.isna().sum()))
print("Non-finite parsed amounts:", int((amount.notna() & ~finite).sum()))
assert len(raw) == len(clean) + len(review)
assert clean[["product", "amount"]].notnull().all().all()
print(clean.to_csv(index=False).rstrip())
# Optional Colab exports:
# clean.to_csv("usable_orders.csv", index=False)
# review.to_csv("orders_for_review.csv", index=False)
````


## Common questions

<strong>Is notnull() different from notna()?</strong> They are aliases. <strong>Does it find missing values?</strong> It marks present values True; use <a href="https://www.plus2net.com/python/pandas-dataframe-isnull.php">isnull()/isna()</a> for the missing mask. <strong>Does it modify data?</strong> No. <strong>Why is an empty string True?</strong> It is not a missing sentinel. <strong>Can I test NaN with ==?</strong> Use isna()/notna() instead. <strong>Is a row with any present field complete?</strong> No; use all() over required fields. <strong>Does notnull() mean valid?</strong> No; apply type, range and format checks separately.

## Exercise: require both a name and a mark

Create names Ravi, missing, Alex and marks 75, 80, missing. Retain rows where both fields are present, and count missing cells. Expect one retained row and two missing cells.

## Exercise solution

all(axis=1) over the required fields implements the AND rule.

````python
practice = pd.DataFrame({"name": ["Ravi", None, "Alex"], "mark": [75, 80, None]})
mask = practice[["name", "mark"]].notnull().all(axis=1)
print(practice.loc[mask])
print("Missing cells:", practice.isna().sum().sum())
assert mask.tolist() == [True, False, False]
assert practice.isna().sum().sum() == 2
````

**Expected output**

```text
   name  mark
0  Ravi  75.0
Missing cells: 2
```
