## 1. Above-the-Fold Basic N Smallest Selection

Select top 2 smallest rows ordered by price.

```python
import pandas as pd

df = pd.DataFrame({
    'Product': ['A', 'B', 'C', 'D'],
    'Price': [15.5, 10.0, 25.0, 8.5]
})

result = df.nsmallest(2, 'Price')
print(result)

# Self-validating assertion
assert list(result['Price']) == [8.5, 10.0]
```

## 2. Multi-Column Tie Breaking

Pass a list of column names to break ties sequentially when primary column values match.

```python
df_items = pd.DataFrame({
    'Price': [10, 10, 15, 5],
    'Shipping': [5, 2, 1, 4]
})

result_tie = df_items.nsmallest(2, ['Price', 'Shipping'])
print(result_tie)

# Self-validating assertion
assert result_tie.loc[1, 'Shipping'] == 2
```
