## 1. Basic N Smallest Selection

Select top 2 smallest rows ordered by price.

````python
import pandas as pd

df = pd.DataFrame({
    'Product': ['A', 'B', 'C', 'D'],
    'Price': [15.5, 10.0, 25.0, 8.5]
})

result = df.nsmallest(2, 'Price')
print(result)

# Self-validating assertion
assert list(result['Price']) == [8.5, 10.0]
````

**Expected output**

```text
  Product  Price
3       D    8.5
1       B   10.0
```
