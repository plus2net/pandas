## 1. Environment & Version Verification
Check your installed Pandas version.

## 2. Core Data Structure 1: Pandas Series (1D Labeled Array)
A Series represents a single column of data with an explicit index.

```python
# Create a Series tracking daily peak temperatures
daily_temps = pd.Series([28.5, 31.0, 26.2, 29.8, 32.4], index=["Mon", "Tue", "Wed", "Thu", "Fri"], name="Temperature_C")
print("Daily Temperature Series:")
print(daily_temps)
print("\nAccess by label ('Wed'):", daily_temps["Wed"])
print(f"Average weekly temp: {daily_temps.mean():.2f}°C")
print("Max temp day:", daily_temps.idxmax(), "with", daily_temps.max(), "°C")
```

## 3. Core Data Structure 2: Pandas DataFrame (2D Tabular Structure)
A DataFrame combines multiple aligned Series into a spreadsheet-like grid.

```python
student_data = {
    "id": [1, 2, 3, 4, 5],
    "name": ["Ravi", "Raju", "Alex", "Sophia", "Priya"],
    "class": ["Four", "Three", "Four", "Five", "Four"],
    "mark": [75, 60, 85, 92, 78],
    "gender": ["male", "male", "male", "female", "female"]
}

df_students = pd.DataFrame(student_data)
print("Students DataFrame:")
display(df_students)
```

## 4. Five Essential Diagnostic Inspection Tools
Inspect structure, types, non-null counts, and dimensions.

```python
print("--- 1. First 3 Rows (head) ---")
print(df_students.head(3))

print("\n--- 2. Last 2 Rows (tail) ---")
print(df_students.tail(2))

print("\n--- 3. Shape (Rows, Columns) ---")
print("Shape:", df_students.shape)

print("\n--- 4. Column Names ---")
print("Columns:", list(df_students.columns))

print("\n--- 5. Technical Summary (info) ---")
df_students.info()
```

Inspect top-level functions and attributes provided by Pandas.

## 6. Hands-on Exercise: Department Salary & Performance Analysis
Filter high-earning staff, compute group averages, and verify results with assertions.

```python
staff_data = {
    "emp_id": [101, 102, 103, 104, 105],
    "name": ["Asha", "John", "Meera", "David", "Karan"],
    "department": ["IT", "Sales", "IT", "HR", "Sales"],
    "salary": [75000, 62000, 84000, 58000, 71000]
}

df_staff = pd.DataFrame(staff_data)

# Compute mean salary by department
dept_avg = df_staff.groupby("department")["salary"].mean().round(2)
print("Average Salary by Department:")
print(dept_avg)

# Filter employees earning > 70,000
high_earners = df_staff[df_staff["salary"] > 70000]
print("\nHigh Earners (> $70,000):")
print(high_earners[["name", "department", "salary"]])

# Assertions
assert len(high_earners) == 3, f"Expected 3 high earners, got {len(high_earners)}"
assert dept_avg["IT"] == 79500.00, f"Expected IT average 79500, got {dept_avg['IT']}"
assert "Meera" in high_earners["name"].values, "Meera should be among high earners"
print("\nAll assertions passed! Exercise complete.")
```
