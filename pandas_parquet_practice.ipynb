```python
import pandas as pd
import numpy as np
import os
import time

# Set random seed
np.random.seed(42)

# Create sample DataFrame with 100,000 rows
n = 100000
df = pd.DataFrame({
    'id': np.arange(1, n + 1),
    'category': np.random.choice(['Alpha', 'Beta', 'Gamma', 'Delta'], n),
    'value': np.random.uniform(10.0, 1000.0, n),
    'timestamp': pd.date_range('2026-01-01', periods=n, freq='s')
})

print("DataFrame Shape:", df.shape)
df.head()
```

```python
# Benchmark CSV Export & Import
start = time.time()
df.to_csv('test.csv', index=False)
csv_write_time = time.time() - start

start = time.time()
df_csv = pd.read_csv('test.csv')
csv_read_time = time.time() - start

csv_size = os.path.getsize('test.csv') / (1024 * 1024)

print(f"CSV Write: {csv_write_time:.3f}s | Read: {csv_read_time:.3f}s | Size: {csv_size:.2f} MB")
```

```python
# Benchmark Parquet (Snappy Compression)
start = time.time()
df.to_parquet('test.parquet', compression='snappy', index=False)
pq_write_time = time.time() - start

start = time.time()
df_pq = pd.read_parquet('test.parquet')
pq_read_time = time.time() - start

pq_size = os.path.getsize('test.parquet') / (1024 * 1024)

print(f"Parquet Write: {pq_write_time:.3f}s | Read: {pq_read_time:.3f}s | Size: {pq_size:.2f} MB")
```

```python
# Benchmark Feather Format
start = time.time()
df.to_feather('test.feather')
ft_write_time = time.time() - start

start = time.time()
df_ft = pd.read_feather('test.feather')
ft_read_time = time.time() - start

ft_size = os.path.getsize('test.feather') / (1024 * 1024)

print(f"Feather Write: {ft_write_time:.3f}s | Read: {ft_read_time:.3f}s | Size: {ft_size:.2f} MB")
```

```python
# Compare all results in a summary DataFrame
results = pd.DataFrame({
    'Format': ['CSV', 'Parquet (Snappy)', 'Feather'],
    'Write Time (s)': [csv_write_time, pq_write_time, ft_write_time],
    'Read Time (s)': [csv_read_time, pq_read_time, ft_read_time],
    'File Size (MB)': [csv_size, pq_size, ft_size]
})
results
```

```python
# Practice Exercise Verification
df_tx = pd.DataFrame({
    "item_id": np.random.randint(1000, 9999, size=50000),
    "price": np.random.round(np.random.uniform(5.0, 250.0, size=50000), 2),
    "transaction_date": pd.date_range(start="2026-01-01", periods=50000, freq="s")
})

df_tx.to_parquet("transactions.parquet", compression="snappy", index=False)
df_sub = pd.read_parquet("transactions.parquet", columns=["item_id", "price"])

print("Exercise Output Shape:", df_sub.shape)
assert df_sub.shape == (50000, 2)
print("Exercise Verified Successfully!")
```
