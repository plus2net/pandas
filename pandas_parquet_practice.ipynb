### Step 2: Basic Reading & Writing Parquet Files
Generate a synthetic dataset and save it to Parquet format.

```python
import pandas as pd
import numpy as np

# Generate synthetic DataFrame
df = pd.DataFrame({
    "id": np.arange(1, 100001),
    "category": np.random.choice(["A", "B", "C", "D"], 100000),
    "sales": np.random.uniform(10.0, 500.0, 100000),
    "timestamp": pd.date_range(start="2026-01-01", periods=100000, freq="min")
})

# Save to Parquet
df.to_parquet("data.parquet", index=False)

# Reload Parquet
df_loaded = pd.read_parquet("data.parquet")
print(df_loaded.info())
```

### Step 3: Compression Engines (Snappy, Zstd, Gzip)
Compare different compression algorithms.

```python
# Save with zstd compression
df.to_parquet("data_zstd.parquet", compression="zstd", index=False)
print("Saved data_zstd.parquet successfully!")
```

### Step 4: Ultra-Fast Session Caching with Feather
Export and re-import using Apache Arrow Feather.

```python
# Export to Feather format
df.to_feather("cache.feather")

# Read Feather format
df_cached = pd.read_feather("cache.feather")
print("Cached Shape:", df_cached.shape)
```

### Step 6: Practical Troubleshooting & Type Casting
Sanitize mixed string/object columns before exporting to Parquet.

```python
# Ensure clean object dtypes prior to export
df_student_csv['name'] = df_student_csv['name'].astype(str)
df_student_csv.to_parquet('student_clean.parquet', index=False)
print("Cleaned Parquet file generated successfully!")
```

### Step 7: Exercise - Columnar Storage Conversion
Create 50,000 synthetic transaction records, save to Parquet with snappy compression, and load selective columns.

```python
np.random.seed(42)
df_tx = pd.DataFrame({
    "item_id": np.random.randint(1000, 9999, size=50000),
    "price": np.random.round(np.random.uniform(5.0, 250.0, size=50000), 2),
    "transaction_date": pd.date_range(start="2026-01-01", periods=50000, freq="s")
})

# Export to Parquet
df_tx.to_parquet("transactions.parquet", compression="snappy", index=False)

# Read selective columns
df_sub = pd.read_parquet("transactions.parquet", columns=["item_id", "price"])

print(f"Subset Shape: {df_sub.shape}")
print(df_sub.head())
assert df_sub.shape == (50000, 2)
print("Exercise assertions passed successfully!")
```
