# Pandas: Calculate Percentage Change with pct_change()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-pct-change.php](https://www.plus2net.com/python/pandas-pct-change.php)

Run cells in order in Google Colab.

## Create numerical sample series

```python
import pandas as pd

prices = pd.Series([100, 110, 105, 126], name="PRICE")
print(prices)
assert prices.shape == (4,)
```

## Basic percentage change calculation

```python
chg = prices.pct_change()
print(chg)
assert chg.iloc[1] == 0.1
```

## Configuring lag periods

```python
chg_lag = prices.pct_change(periods=2)
print(chg_lag)
assert chg_lag.iloc[2] == 0.05
```

## Exercise solution

```python
rev = pd.Series([1000, 1200, 1500], name="REV")
answer = rev.pct_change()
print(answer)
assert answer.iloc[2] == 0.25
```
