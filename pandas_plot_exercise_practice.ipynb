### 1. Dataset Setup
Load sample student records with classes and marks.


```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "id": range(1, 11),
    "name": ["John", "Ravi", "Sara", "Alex", "Nina", "Mike", "Tara", "Leo", "Maya", "Omar"],
    "class1": ["Four", "Three", "Four", "Five", "Three", "Four", "Five", "Three", "Four", "Five"],
    "mark": [75, 82, 90, 65, 88, 70, 95, 60, 85, 78]
}
df = pd.DataFrame(data)
print(df)
assert len(df) == 10
```

### 2. Exercise 1: Line Chart of Class Counts


```python
counts = df["class1"].value_counts().sort_index()
print(counts)
assert counts["Four"] == 4

counts.plot.line(marker="o", grid=True, title="Class Student Enrollment Counts")
plt.ylabel("Student Count")
plt.show()
```

### 3. Exercise 2 & 3: Mean and Multi-Bar Charts


```python
mean_df = df.groupby("class1")["mark"].mean()
print("Mean marks:", mean_df.to_dict())

agg_df = df.groupby("class1")["mark"].agg(["mean", "max"])
print(agg_df)
assert agg_df.loc["Four", "max"] == 90

agg_df.plot.bar(title="Mean vs Maximum Mark by Class", figsize=(6.5, 4))
plt.ylabel("Marks")
plt.show()
```

### 4. Exercise 4 & 5: Pie and Density Plots


```python
counts.plot.pie(autopct="%1.0f%%", title="Class Student Share")
plt.ylabel("")
plt.show()

df["mark"].plot.density(title="Student Mark Probability Density", bw_method=0.3)
plt.xlabel("Marks")
plt.show()
```
