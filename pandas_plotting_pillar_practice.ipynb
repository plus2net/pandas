### 1. Unified Pandas Plotting Syntax
Demonstrate both `df.plot(kind=...)` and `df.plot.<kind>()` accessors.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Revenue": [12000, 15000, 14000, 18000, 21000],
    "Expenses": [8000, 9500, 8700, 11000, 12500]
}
df = pd.DataFrame(data)
print(df)

df.plot.line(x="Month", y=["Revenue", "Expenses"], marker="o", title="Financial Trends")
plt.show()
```

### 2. Multi-Plot Subplot Dashboard
Combine multiple plot formats onto a unified analytical canvas.


```python
months = [f"2026-{m:02d}" for m in range(1, 7)]
sales_north = [45000, 48000, 52000, 58000, 61000, 67000]
sales_south = [32000, 34000, 37000, 39000, 44000, 48000]

df_sales = pd.DataFrame({
    "Month": months,
    "North_Region": sales_north,
    "South_Region": sales_south
})
df_sales["Total_Sales"] = df_sales["North_Region"] + df_sales["South_Region"]

print(df_sales)
assert df_sales["Total_Sales"].iloc[-1] == 115000

fig, axes = plt.subplots(2, 2, figsize=(11, 8))

df_sales.plot.line(x="Month", y="Total_Sales", marker="s", color="#1f77b4", ax=axes[0, 0], title="Total Monthly Revenue")
df_sales.plot.bar(x="Month", y=["North_Region", "South_Region"], ax=axes[0, 1], title="Regional Performance")
df_sales.plot.area(x="Month", y=["South_Region", "North_Region"], alpha=0.6, ax=axes[1, 0], title="Cumulative Volume")

total_by_region = pd.Series({
    "North": df_sales["North_Region"].sum(),
    "South": df_sales["South_Region"].sum()
})
total_by_region.plot.pie(autopct="%1.1f%%", ax=axes[1, 1], title="H1 Regional Share")
axes[1, 1].set_ylabel("")

plt.tight_layout()
plt.show()
```
