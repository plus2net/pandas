# Pandas: Calculate Product of Values with prod()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-prod.php](https://www.plus2net.com/python/pandas-prod.php)

Run cells in order in Google Colab.

## Create numerical sample DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "FACTOR_A": [1, 2, 3, 4],
    "FACTOR_B": [5, 5, 2, 2]
})
print(df)
assert df.shape == (4, 2)
```

## Basic product calculation

```python
res = df.prod()
print(res)
assert res["FACTOR_A"] == 24 and res["FACTOR_B"] == 100
```

## Row-wise product calculation (axis=1)

```python
row_prod = df.prod(axis=1)
print(row_prod)
assert row_prod.iloc[0] == 5 and row_prod.iloc[1] == 10
```

## Exercise solution

```python
factors = pd.Series([1.1, 1.2, 1.05], name="GROWTH")
answer = factors.prod()
print(f"Answer: {answer:.3f}")
assert round(answer, 3) == 1.386
```
