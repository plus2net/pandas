## Project 1: Directory Metadata Cataloging to Excel

```python
import pandas as pd
import os
import tempfile
from datetime import datetime

# Simulate files in a directory
temp_dir = tempfile.mkdtemp()
sample_files = ["index.php", "about.php", "contact.php", "style.css", "readme.txt"]
for sf in sample_files:
    with open(os.path.join(temp_dir, sf), "w") as fp:
        fp.write("Sample content for " + sf)

records = []
for file_name in os.listdir(temp_dir):
    full_path = os.path.join(temp_dir, file_name)
    if os.path.isfile(full_path):
        name, ext = os.path.splitext(file_name)
        if ext == ".php":
            mtime = os.path.getmtime(full_path)
            mod_date = datetime.fromtimestamp(mtime).strftime("%Y-%m-%d")
            records.append({"f_name": file_name, "dt": mod_date, "size": os.path.getsize(full_path)})

df_catalog = pd.DataFrame(records)
print("Cataloged PHP files:")
print(df_catalog)
assert len(df_catalog) == 3
assert set(df_catalog["f_name"]) == {"index.php", "about.php", "contact.php"}
```

## Project 2: Temporal File Modification Reporting

```python
df_mock = pd.DataFrame({
    "f_name": ["page1.php", "page2.php", "page3.php", "page4.php"],
    "dt": pd.to_datetime(["2023-07-05", "2023-07-20", "2023-08-01", "2023-08-15"]),
    "size": [1200, 3400, 2100, 4500]
})
jul_updates = df_mock[df_mock["dt"].dt.strftime("%Y-%m") == "2023-07"]
print("July 2023 Updates:")
print(jul_updates)
assert len(jul_updates) == 2
```

## Project 3: Website Link Integrity & Orphan Page Detection

```python
import re

html_sources = {
    "home.php": "<a href='about.php'>About</a> <a href='contact.php'>Contact</a>",
    "about.php": "<a href='home.php'>Home</a>",
    "contact.php": "<a href='home.php'>Home</a>",
    "orphan.php": "No one links to this page"
}
href_pat = re.compile(r"href=[^a-zA-Z0-9]?([a-zA-Z0-9_.-]+\.php)")
linked = set()
for content in html_sources.values():
    linked.update(href_pat.findall(content))

orphans = [f for f in html_sources.keys() if f not in linked]
print("Detected orphan pages:", orphans)
assert orphans == ["orphan.php"]
```
