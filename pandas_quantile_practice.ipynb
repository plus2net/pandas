# Pandas: Calculate Quantiles & Percentiles with quantile()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-quantile.php](https://www.plus2net.com/python/pandas-quantile.php)

Run cells in order in Google Colab.

## Create numerical sample Series

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40, 50], name="SCORES")
print(s)
assert s.shape == (5,)
```

## Single quantile calculation

```python
q50 = s.quantile(0.5)
print(f"Median (q=0.5): {q50}")
assert q50 == 30.0
```

## Computing multiple quantiles simultaneously

```python
q_multi = s.quantile([0.25, 0.75])
print(q_multi)
assert q_multi.loc[0.75] == 40.0
```

## Exercise solution

```python
vals = pd.Series([100, 200, 300, 400, 500], name="VALS")
answer = vals.quantile(0.9)
print(answer)
assert answer == 460.0
```
