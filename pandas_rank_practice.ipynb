# Pandas: Rank Numerical Data with rank()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-rank.php](https://www.plus2net.com/python/pandas-rank.php)

Run cells in order in Google Colab.

## Create numerical sample dataset with ties

```python
import pandas as pd

df = pd.DataFrame({
    "STUDENT": ["Ravi", "Raju", "Alex", "John", "Sara"],
    "SCORE": [80, 90, 80, 95, 70]
})
print(df)
assert df.shape == (5, 2)
```

## Basic ranking (ascending vs descending)

```python
df["RANK_DESC"] = df["SCORE"].rank(ascending=False)
print(df)
assert df.loc[3, "RANK_DESC"] == 1.0
```

## Tie-breaking methods (method parameter)

```python
df["MIN_RANK"] = df["SCORE"].rank(ascending=False, method="min")
df["DENSE_RANK"] = df["SCORE"].rank(ascending=False, method="dense")
print(df[["STUDENT", "SCORE", "MIN_RANK", "DENSE_RANK"]])
assert df.loc[0, "MIN_RANK"] == 3.0 and df.loc[4, "DENSE_RANK"] == 4.0
```

## Percentile ranking (pct=True)

```python
df["PCT_RANK"] = df["SCORE"].rank(pct=True)
print(df[["STUDENT", "SCORE", "PCT_RANK"]])
assert df.loc[3, "PCT_RANK"] == 1.0
```

## Exercise solution

```python
sales = pd.Series([500, 700, 500, 900], name="SALES")
answer = sales.rank(ascending=False, method="min")
print(answer)
assert answer[3] == 1.0
```
