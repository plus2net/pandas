## 1. Basic Scatter Plot (X vs Y)

Unlike line plots, `df.plot.scatter()` strictly requires both an `x=` and a `y=` coordinate column.

````python
import pandas as pd
import matplotlib.pyplot as plt

my_dict = {
    "visit": [50, 53, 55, 57, 59, 60, 61],
    "sale": [8, 9, 11, 12, 14, 16, 17]
}
df = pd.DataFrame(data=my_dict)
print(df)

df.plot.scatter(x="visit", y="sale", title="Sale Vs Visit", figsize=(6, 3))
plt.show()
assert len(df) == 7
assert df["visit"].sum() == 395
````

**Expected output**

```text
   visit  sale
0     50     8
1     53     9
2     55    11
3     57    12
4     59    14
5     60    16
6     61    17
```

## 2. Colormap Gradient (3rd Dimension)

Pass a column name to `c=` to color points according to values with an automatic colorbar.

````python
df.plot.scatter(x="visit", y="sale", c="sale", colormap="viridis", title="Sales Gradient", figsize=(7, 4))
plt.show()
assert df["sale"].sum() == 87
````

## 3. Bubble Chart with Dynamic Point Sizes

Pass a scaled column or array to `s=` to size markers proportionally to volume.

````python
df.plot.scatter(x="visit", y="sale", s=df["sale"] * 20, title="Bubble Chart: Sales Volume", alpha=0.6, figsize=(7, 4))
plt.show()
````

## 4. Practical Workflow: Marketing Ad Spend vs Conversions vs Revenue

Analyze digital ad performance by mapping Spend to X, Conversions to Y, and Revenue to the colorbar scale.

````python
campaigns = pd.DataFrame({
    "Campaign": ["Email", "Search", "Social", "Display", "Affiliate"],
    "Ad_Spend": [500, 1500, 1200, 800, 600],
    "Conversions": [45, 130, 95, 50, 40],
    "Revenue": [4500, 14500, 10200, 4800, 3900]
})

assert len(campaigns) == 5
assert campaigns["Ad_Spend"].sum() == 4600
assert campaigns["Revenue"].sum() == 37900
print("Total Ad Spend:", campaigns["Ad_Spend"].sum())
print("Total Revenue:", campaigns["Revenue"].sum())

ax = campaigns.plot.scatter(
    x="Ad_Spend",
    y="Conversions",
    c="Revenue",
    colormap="viridis",
    s=100,
    figsize=(8, 5),
    title="Campaign Spend vs Conversions (Color = Revenue)",
    grid=True
)
plt.show()
````

**Expected output**

```text
Total Ad Spend: 4600
Total Revenue: 37900
```

## Exercise: Ad Impressions vs Click Performance

Create a DataFrame representing 5 online ads with columns: `Ad_ID` (`[1, 2, 3, 4, 5]`), `Impressions` (`[1000, 2500, 4000, 6000, 8500]`), and `Clicks` (`[25, 60, 110, 175, 260]`). Plot a scatter chart of Impressions on the X-axis vs Clicks on the Y-axis with gridlines, title the plot *"Ad Clicks vs Impressions"*, and verify total clicks.

## Exercise Solution

Verify your implementation with assertions below:

````python
ads = pd.DataFrame({
    "Ad_ID": [1, 2, 3, 4, 5],
    "Impressions": [1000, 2500, 4000, 6000, 8500],
    "Clicks": [25, 60, 110, 175, 260]
})

total_clicks = ads["Clicks"].sum()
assert total_clicks == 630
print("Total Clicks:", total_clicks)

ads.plot.scatter(
    x="Impressions",
    y="Clicks",
    title="Ad Clicks vs Impressions",
    grid=True,
    color="teal",
    s=60,
    figsize=(6, 3)
)
plt.show()
````

**Expected output**

```text
Total Clicks: 630
```
