# Pandas: Calculate Standard Error of Mean with sem()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-sem.php](https://www.plus2net.com/python/pandas-sem.php)

Run cells in order in Google Colab.

## Create numerical sample DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "TRIAL_A": [10, 12, 14, 16, 18],
    "TRIAL_B": [20, 25, 30, 35, 40]
})
print(df)
assert df.shape == (5, 2)
```

## Basic SEM calculation

```python
sem_res = df.sem()
print(sem_res)
assert round(sem_res["TRIAL_A"], 4) == 1.4142
```

## Standard error on a single Series

```python
s_sem = df["TRIAL_A"].sem()
print(f"Standard Error: {s_sem:.4f}")
assert round(s_sem, 4) == 1.4142
```

## Exercise solution

```python
vals = pd.Series([2, 4, 4, 4, 5, 5, 7, 9], name="MEASURE")
answer = vals.sem()
print(f"Answer: {answer:.4f}")
assert round(answer, 4) == 0.7777
```
