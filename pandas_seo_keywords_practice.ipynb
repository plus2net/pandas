# SEO Reports & Keyword Analysis Using Pandas

Interactive companion notebook for [SEO Reports & Keyword Analysis](https://www.plus2net.com/python/pandas-seo-keywords.php). Covers GA4 data cleaning, regex query matching, and keyword token counts.

## Search Query Filtering with OR and AND (Lookahead) Regex

```python
import pandas as pd

queries_df = pd.DataFrame({
    "queries": [
        "tkinter button color",
        "pandas dataframe plot",
        "tkinter label color background",
        "python pandas date range",
        "tkinter combobox select event"
    ],
    "position": [6.5, 3.2, 8.1, 12.4, 7.8]
})

keywords = ["tkinter", "color"]
# AND matching
and_regex = "".join([f"(?=.*{w})" for w in keywords])
and_matches = queries_df[queries_df["queries"].str.contains(and_regex, case=False, na=False)]
print("AND Matches:")
print(and_matches)
assert len(and_matches) == 2
assert set(and_matches.index) == {0, 2}
```

## Keyword Frequency Extraction with stack() & value_counts()

```python
word_counts = queries_df["queries"].str.split(expand=True).stack().value_counts()
print("Top Keyword Frequencies:")
print(word_counts.head(5))
assert word_counts["tkinter"] == 3
assert word_counts["color"] == 2
```
