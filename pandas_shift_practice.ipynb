# Pandas: Shift Index Data with shift()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-shift.php](https://www.plus2net.com/python/pandas-shift.php)

Run cells in order in Google Colab.

## Create numerical sample series

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40], name="VAL")
print(s)
assert s.shape == (4,)
```

## Basic shifting (lagging data)

```python
shifted_down = s.shift(1)
print(shifted_down)
assert pd.isna(shifted_down.iloc[0]) and shifted_down.iloc[1] == 10.0
```

## Negative shifting (leading data)

```python
shifted_up = s.shift(-1)
print(shifted_up)
assert shifted_up.iloc[0] == 20.0 and pd.isna(shifted_up.iloc[3])
```

## Filling missing values with fill_value

```python
filled_shift = s.shift(1, fill_value=0)
print(filled_shift)
assert filled_shift.iloc[0] == 0
```

## Exercise solution

```python
sales = pd.Series([100, 200, 300], name="SALES")
answer = sales.shift(1, fill_value=0)
print(answer)
assert answer.iloc[1] == 100
```
