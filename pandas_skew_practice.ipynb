# Pandas: Calculate Skewness with skew()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-skew.php](https://www.plus2net.com/python/pandas-skew.php)

Run cells in order in Google Colab.

## Create numerical sample dataset

```python
import pandas as pd

s = pd.Series([10, 12, 15, 18, 100], name="VAL")
print(s)
assert s.shape == (5,)
```

## Basic skewness calculation

```python
s_skew = s.skew()
print(f"Skewness: {s_skew:.4f}")
assert s_skew > 2.0
```

## Understanding positive vs negative skewness

```python
df = pd.DataFrame({
    "RIGHT_SKEW": [1, 2, 2, 3, 10],
    "LEFT_SKEW": [1, 8, 9, 9, 10]
})
df_skew = df.skew()
print(df_skew)
assert df_skew["RIGHT_SKEW"] > 0 and df_skew["LEFT_SKEW"] < 0
```

## Exercise solution

```python
scores = pd.Series([100, 95, 90, 85, 20], name="SCORES")
answer = scores.skew()
print(answer)
assert answer < 0
```
