# Pandas DataFrame `sort_index()` Method Practice

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-sort_index.php](https://www.plus2net.com/python/pandas-sort_index.php)

## 1. Ascending Index Sorting

Sort unsorted index labels (`'b'`, `'c'`, `'a'`) alphabetically.

````python
import pandas as pd

df = pd.DataFrame({'VAL': [10, 20, 30]}, index=['b', 'c', 'a'])
result = df.sort_index()
print(result)

# Self-validating assertion
assert result.index.tolist() == ['a', 'b', 'c']
assert result.loc['a', 'VAL'] == 30
````

**Expected output**

```text
   VAL
a   30
b   10
c   20
```

## 2. Column Header Sorting (`axis=1`)

Alphabetize DataFrame column names across columns.

````python
df_data = pd.DataFrame({'Z_COL': [1, 2], 'A_COL': [3, 4]})
df_sorted_cols = df_data.sort_index(axis=1)
print(df_sorted_cols)

# Self-validating assertion
assert list(df_sorted_cols.columns) == ['A_COL', 'Z_COL']
````

**Expected output**

```text
   A_COL  Z_COL
0      3      1
1      4      2
```

## 3. Multi-Index Level Sorting

Sort MultiIndex DataFrames hierarchically across index levels.

````python
tuples = [('b', 2), ('a', 2), ('a', 1)]
index = pd.MultiIndex.from_tuples(tuples, names=['LETTER', 'NUM'])
df_multi = pd.DataFrame({'SCORE': [10, 20, 30]}, index=index)

sorted_multi = df_multi.sort_index()
print(sorted_multi)

# Self-validating assertion
assert sorted_multi.index[0] == ('a', 1)
assert sorted_multi.index[2] == ('b', 2)
````

**Expected output**

```text
            SCORE
LETTER NUM       
a      1       30
       2       20
b      2       10
```
