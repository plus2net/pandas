# Pandas: Pivot Columns to MultiIndex Rows with stack()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-stack.php](https://www.plus2net.com/python/pandas-stack.php)

Run cells in order in Google Colab.

## Create multi-header sample dataset

```python
import pandas as pd
cols = pd.MultiIndex.from_tuples([("MATH", "Term1"), ("MATH", "Term2"), ("ENGLISH", "Term1"), ("ENGLISH", "Term2")])
df = pd.DataFrame([[85, 90, 78, 82], [88, 92, 80, 85]], index=["Ravi", "Raju"], columns=cols)
print(df)
assert df.shape == (2, 4)
```

## Basic stack operation

```python
stacked = df.stack()
print(stacked)
assert stacked.index.nlevels == 2
```

## Stacking specific levels

```python
stacked_l0 = df.stack(level=0)
print(stacked_l0)
assert "MATH" in stacked_l0.index.get_level_values(1)
```

## Handling missing values with dropna

```python
df_nan = df.copy()
df_nan.iloc[0, 0] = None
stacked_with_nan = df_nan.stack(future_stack=True)
print(stacked_with_nan)
assert pd.isna(stacked_with_nan.loc[("Ravi", "Term1"), "MATH"])
```

## Exercise solution

```python
ex_df = pd.DataFrame([[90, 95], [80, 85]], index=["Ravi", "Raju"], columns=["Math", "Science"])
answer = ex_df.stack()
print(answer)
assert len(answer) == 4 and answer.index.nlevels == 2
```
