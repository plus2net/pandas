### 1. In-Memory Student Benchmark Dataset
Create the standard 35-row student DataFrame in Python.


```python
import pandas as pd

student_data = {
    "id": list(range(1, 36)),
    "name": [
        "John Deo", "Max Ruin", "Arnold", "Krish Star", "John Mike",
        "Alex John", "My John Rob", "Asruid", "Tes Qry", "Big John",
        "Ronald", "Recky", "Kty", "Bigy", "Tade Row", "Gimmy",
        "Tumyu", "Honny", "Tinny", "Jackly", "Babby John", "Reggid",
        "Herod", "Tiddy Now", "Giff Tow", "Crelea", "Big Nose",
        "RorFail", "Duke", "Dean", "Eack", "Blake", "David",
        "Ram", "Jane"
    ],
    "class1": [
        "Four", "Three", "Three", "Four", "Four",
        "Four", "Five", "Five", "Six", "Four",
        "Six", "Six", "Seven", "Seven", "Four",
        "Four", "Six", "Five", "Nine", "Nine",
        "Four", "Seven", "Eight", "Seven", "Seven",
        "Seven", "Three", "Five", "Six", "Five",
        "Six", "Four", "Eight", "Three", "Four"
    ],
    "mark": [
        75, 85, 55, 60, 60,
        55, 78, 85, 78, 55,
        89, 94, 88, 88, 88,
        88, 54, 75, 18, 65,
        69, 55, 79, 78, 88,
        79, 81, 59, 55, 74,
        65, 88, 89, 78, 85
    ],
    "gender": [
        "female", "male", "male", "female", "female",
        "male", "male", "male", "male", "female",
        "female", "female", "female", "female", "male",
        "male", "male", "male", "male", "female",
        "female", "female", "male", "male", "male",
        "male", "female", "male", "female", "female",
        "female", "female", "female", "male", "female"
    ]
}
df_student = pd.DataFrame(student_data)
print(df_student.head())
assert len(df_student) == 35
```

### 2. Statistical Metrics and Summaries
Check row counts, average marks, and class distributions.


```python
print("Total Records:", len(df_student))
print("Mean Mark:", df_student["mark"].mean())
print("Class Counts:\n", df_student["class1"].value_counts())

assert df_student["mark"].mean() == 73.0
assert len(df_student[df_student["gender"] == "female"]) == 18
```

### 3. Hands-on Exercise: Honor Roll Filter
Filter students with mark >= 80 and export to CSV.


```python
honor_roll = df_student[df_student["mark"] >= 80].sort_values(by="mark", ascending=False)
print("Honor Roll Count:", len(honor_roll))
print(honor_roll[["name", "class1", "mark"]].head(3))

assert len(honor_roll) == 13
assert honor_roll["mark"].min() >= 80
print("All assertions passed cleanly!")
```
