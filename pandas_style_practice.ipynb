# Pandas DataFrame Style: Conditional Formatting & HTML Styling

Interactive companion notebook for [Pandas Style](https://www.plus2net.com/python/pandas-style.php). Covers formatters, min/max highlights, gradients, and custom HTML table CSS.

## Formatters & Highlight Min / Max

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "name": ["John Deo", "Max Ruin", "Arnold", "Krish Star"],
    "class": ["Four", "Three", "Three", "Four"],
    "mark": [75, 85, 55, 60]
})
styled = (df.style
    .format({"name": lambda x: x.lower()})
    .highlight_max(color="lightgreen", subset=["mark"])
    .highlight_min(color="#f8d7da", subset=["mark"]))
assert df["mark"].max() == 85
assert df["mark"].min() == 55
print("Style object created successfully:", styled)
```

## Background Gradient & Conditional Row Formatting

```python
def highlight_passing(row):
    if row["mark"] >= 60:
        return ["background-color: #fff3cd; font-weight: bold"] * len(row)
    else:
        return ["background-color: white"] * len(row)

styled_rows = df.style.apply(highlight_passing, axis=1)
html_table = styled_rows.to_html()
assert "<table" in html_table
print("Generated HTML table length:", len(html_table))
```
