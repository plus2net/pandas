### Step 2: Basic to_feather() Serialization

```python
import pandas as pd
import numpy as np

df_sales = pd.DataFrame({
    "transaction_id": np.arange(1001, 1006),
    "store_code": ["S101", "S102", "S101", "S103", "S102"],
    "amount": [120.50, 450.00, 85.25, 920.10, 310.00]
})

# Write to feather
df_sales.to_feather("sales.feather")

# Reload from feather
df_reloaded = pd.read_feather("sales.feather")
print(df_reloaded)
```

### Step 3: Compression Algorithms (LZ4 vs Zstd)

```python
df_sales.to_feather("sales_lz4.feather", compression="lz4")
df_sales.to_feather("sales_zstd.feather", compression="zstd", compression_level=6)
print("Compressed Feather files written successfully!")
```

### Step 5: Exercise Verification

```python
np.random.seed(42)
df_users = pd.DataFrame({
    "user_id": np.arange(50001, 70001),
    "score": np.random.uniform(50.0, 100.0, size=20000),
    "status": np.random.choice(["Active", "Inactive"], size=20000)
})

df_export = df_users.set_index("user_id").reset_index()
df_export.to_feather("users.feather", compression="zstd")

df_reloaded = pd.read_feather("users.feather")
print(df_reloaded.head())
assert df_reloaded.shape == (20000, 3)
print("Exercise assertions passed successfully!")
```
