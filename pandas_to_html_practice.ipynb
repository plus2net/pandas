## Create an HTML Table String

Without a destination, <code>to_html()</code> returns a string containing a table, not a complete HTML document. By default it includes column headings and row index labels. These student scores use the same small dataset throughout the basic examples.

````python
import pandas as pd
from pathlib import Path
from io import StringIO

folder = Path("pandas_html_practice")
folder.mkdir(exist_ok=True)

def student_data():
    return pd.DataFrame({"NAME": ["Ravi", "Raju", "Alex"], "ID": [1, 2, 3],
                         "MATH": [30, 40, 50], "ENGLISH": [20, 30, 41]})

df = student_data()
table_html = df.to_html()
print(table_html)
assert isinstance(table_html, str)
assert table_html.count("<tr") == 4
````

## Save a Table to a File or Buffer

With a path or writable buffer, to_html writes the table and returns None. Create the parent directory first. This relative path works in Colab and on a local computer; a Colab runtime cannot write directly to your Windows D drive. Read the UTF-8 file back to verify it.

````python
df = student_data()
path = folder / "students.html"
result = df.to_html(path, index=False, encoding="utf-8")
assert result is None
saved = path.read_text(encoding="utf-8")
assert "Ravi" in saved and "<table" in saved
buffer = StringIO()
df.to_html(buffer, index=False)
assert buffer.getvalue() == saved
print("Saved:", path.as_posix())
print("Return value:", result)
````

**Expected output**

```text

Return value: None
```

## classes, table_id and border

classes adds class names to the generated table; it does not load Bootstrap or define CSS. table_id adds an ID that you can target. border=0 omits the legacy border attribute. Use a stylesheet for presentation.

````python
df = student_data()
table_html = df.to_html(index=False, classes=["table", "table-striped"],
                        table_id="student-table", border=0)
print(table_html.splitlines()[0])
assert 'class="dataframe table table-striped"' in table_html
assert 'id="student-table"' in table_html
assert 'border=' not in table_html
````

## Column Width and Header Alignment

col_space is a minimum width, not a maximum width. A dictionary sets widths for individual columns. justify aligns column headers; use CSS to align body cells. Long text may still make a table wider, so use a scrolling wrapper.

## float_format, decimal and Column Formatters

Display precision does not round the stored values. float_format applies to floating values; formatters supplies functions for named columns. decimal changes the decimal separator in numeric rendering, while a custom formatter controls its own returned string.

````python
df = pd.DataFrame({"product": ["Notebook", "Pen"], "price": [12.3456, 3.5]})
print(df.to_html(index=False, float_format=lambda value: f"{value:.2f}"))
print("Comma decimal:")
print(df.to_html(index=False, decimal=","))
currency = df.to_html(index=False, formatters={"price": lambda value: f"INR {value:,.2f}"})
assert "INR 12.35" in currency
assert df.loc[0, "price"] == 12.3456
````

## Keep Text Escaped in Exported Tables

escape=True is the default: angle brackets and ampersands in data are rendered as text. escape=False allows markup in values and should be reserved for content you control. Do not disable escaping simply to make a table look better.

````python
df = pd.DataFrame({"note": ["A & B", "<strong>Priority</strong>"]})
safe_html = df.to_html(index=False, escape=True)
print(safe_html)
assert "A &amp; B" in safe_html
assert "&lt;strong&gt;Priority&lt;/strong&gt;" in safe_html
trusted_html = df.to_html(index=False, escape=False)
assert "<strong>Priority</strong>" in trusted_html
````

## Practical Sales Report: Select, Aggregate and Check

This synthetic order dataset contains one missing revenue value. Keep the known subtotal separate from the number of incomplete records. The <a href="https://www.plus2net.com/python/pandas-dataframe-groupby.php">groupby()</a> report uses min_count=1 so a group with no known revenue stays missing. Reconcile the report before generating HTML.

````python
def sales_data():
    return pd.DataFrame({"order_id": [101, 102, 103, 104, 105],
                         "category": ["Books", "Books", "Stationery", "Stationery", "Gifts"],
                         "revenue": [120.5, 80.0, 30.25, 20.0, float("nan")]})

sales = sales_data()
summary = sales.groupby("category", sort=True).agg(
    records=("order_id", "size"), known_values=("revenue", "count"),
    known_revenue=("revenue", lambda values: values.sum(min_count=1)))
summary["missing_values"] = summary["records"] - summary["known_values"]
summary = summary.reset_index()
print(summary.to_string(index=False))
assert summary["records"].sum() == len(sales)
assert summary["missing_values"].sum() == 1
assert summary["known_revenue"].sum(min_count=1) == sales["revenue"].sum(min_count=1) == 250.75
````

**Expected output**

```text
  category  records  known_values  known_revenue  missing_values
     Books        2             2         200.50               0
     Gifts        1             0            NaN               1
Stationery        2             2          50.25               0
```

## Build a Complete HTML Document with CSS

A table fragment needs a document title, character encoding and styles for a standalone report. CSS centers the table and controls body-cell alignment; justify alone affects headers. The scrolling wrapper helps on narrow screens. Include a caption so readers know what the table contains. Keep the missing-value label visible.

````python
table_html = summary.to_html(index=False, table_id="sales-summary",
                             classes=["report", "center"], border=0,
                             na_rep="Not recorded", float_format=lambda value: f"{value:.2f}")
opening_end = table_html.index(">") + 1
table_html = (table_html[:opening_end] +
              "<caption>Revenue by category — synthetic practice data</caption>" +
              table_html[opening_end:])
css = """body {font-family: Arial, sans-serif; margin: 2rem; color: #222;}
.table-wrap {overflow-x: auto;}
.center {margin-left: auto; margin-right: auto;}
.report {border-collapse: collapse; min-width: 36rem;}
.report th, .report td {padding: .65rem; border: 1px solid #ccc;}
.report th {background: #eaf1fb;}
.report td {text-align: right;}
.report td:first-child {text-align: left;}
.report tbody tr:nth-child(even) {background: #f5f7fa;}
.report caption {text-align: left; font-weight: bold; margin-bottom: .75rem;}"""
document = ("<!DOCTYPE html><html lang='en'><head><meta charset='UTF-8'>"
            "<meta name='viewport' content='width=device-width, initial-scale=1'>"
            "<title>Practice Sales Report</title><style>" + css + "</style></head><body>"
            "<h1>Practice Sales Report</h1>"
            "<p>Known revenue subtotal: 250.75. One record needs review; this is not a complete revenue total.</p>"
            "<div class='table-wrap'>" + table_html + "</div></body></html>")
path = folder / "sales_report.html"
path.write_text(document, encoding="utf-8")
restored = path.read_text(encoding="utf-8")
assert "Not recorded" in restored and "250.75" in restored
assert "<caption>" in restored and "overflow-x: auto" in restored
print("Saved:", path.as_posix())
print("Known revenue subtotal:", summary["known_revenue"].sum(min_count=1))
````

**Expected output**

```text
Saved: pandas_html_practice/sales_report.html
Known revenue subtotal: 250.75
```

## Check the Exported HTML Structure

Check the saved file, not only the DataFrame. This small standard-library parser counts body rows, verifies that cells were not truncated and reads a known numeric value. It confirms the data structure; visual layout still needs a browser preview.

````python
from html.parser import HTMLParser

class TableCheck(HTMLParser):
    def __init__(self):
        super().__init__()
        self.in_body = False
        self.in_cell = False
        self.rows = 0
        self.values = []
    def handle_starttag(self, tag, attrs):
        if tag == "tbody":
            self.in_body = True
        if tag == "tr" and self.in_body:
            self.rows += 1
        if tag == "td" and self.in_body:
            self.in_cell = True
    def handle_endtag(self, tag):
        if tag == "tbody":
            self.in_body = False
        if tag == "td":
            self.in_cell = False
    def handle_data(self, data):
        if self.in_cell:
            self.values.append(data.strip())

check = TableCheck()
check.feed((folder / "sales_report.html").read_text(encoding="utf-8"))
assert check.rows == len(summary) == 3
assert len(check.values) == summary.shape[0] * summary.shape[1]
assert "200.50" in check.values and "Not recorded" in check.values
print("Verified report rows:", check.rows)
print("Verified body cells:", len(check.values))
````

**Expected output**

```text
Verified report rows: 3
Verified body cells: 15
```

<strong>buf:</strong> no destination returns a string; a file or buffer returns None. <strong>notebook=True:</strong> changes the generated markup for notebook use, but does not display it automatically. <strong>encoding:</strong> specify UTF-8 when writing a file. <strong>Styles missing:</strong> define CSS or load the intended framework. <strong>Unexpected ellipses:</strong> remove max_rows/max_cols for a complete export. <strong>Wrong decimal text:</strong> review float_format, decimal and custom formatters. <strong>Missing values:</strong> na_rep is presentation, not replacement. <strong>More complex styling:</strong> DataFrame.style.to_html supports cell styling through the separate Styler API; do not assume its options are identical to DataFrame.to_html.
