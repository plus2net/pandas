## 1. Basic String Conversion

Convert string Series entries into numerical floats.

````python
import pandas as pd

s = pd.Series(['1.5', '2.0', '3'])
converted = pd.to_numeric(s)
print(converted)

# Self-validating assertions
assert converted.dtype == 'float64'
assert converted[0] == 1.5
````

**Expected output**

```text
0    1.5
1    2.0
2    3.0
dtype: float64
```

## 2. Dirty Data Coercion (`errors='coerce'`)

Convert corrupt string tokens directly to `NaN` without crashing.

````python
s_dirty = pd.Series(['10.5', 'ERROR', '30', '  '])
cleaned = pd.to_numeric(s_dirty, errors='coerce')
print(cleaned)

# Self-validating assertions
assert cleaned[0] == 10.5
assert pd.isna(cleaned[1])
````

**Expected output**

```text
0    10.5
1     NaN
2    30.0
3     NaN
dtype: float64
```
