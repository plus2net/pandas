## 1. Basic Numeric Conversion

Convert a string Series into numeric float dtypes.

````python
import pandas as pd

s = pd.Series(['1.5', '2.0', '3'])
converted = pd.to_numeric(s)
print(converted)

# Self-validating assertion
assert converted.dtype == 'float64'
assert converted[0] == 1.5
````

**Expected output**

```text
0    1.5
1    2.0
2    3.0
dtype: float64
```
