## 1. Above-the-Fold Basic String Conversion

Convert string Series entries into numerical floats.

```python
import pandas as pd

s = pd.Series(['1.5', '2.0', '3'])
converted = pd.to_numeric(s)
print(converted)

# Self-validating assertions
assert converted.dtype == 'float64'
assert converted[0] == 1.5
```

## 2. Dirty Data Coercion (`errors='coerce'`)

Convert corrupt string tokens directly to `NaN` without crashing.

```python
s_dirty = pd.Series(['10.5', 'ERROR', '30', '  '])
cleaned = pd.to_numeric(s_dirty, errors='coerce')
print(cleaned)

# Self-validating assertions
assert cleaned[0] == 10.5
assert pd.isna(cleaned[1])
```

## 3. Memory Optimization via `downcast`

Automatically reduce bit-width of numeric columns.

```python
s_small = pd.Series(['1', '5', '10'])
s_std = pd.to_numeric(s_small)
print('Standard Dtype:', s_std.dtype)

s_down = pd.to_numeric(s_small, downcast='integer')
print('Downcasted Dtype:', s_down.dtype)

# Self-validating assertions
assert s_down.dtype == 'int8'
```

## 4. Batch Converting DataFrame Columns

Run `to_numeric()` across multiple DataFrame columns simultaneously.

```python
df_raw = pd.DataFrame({
    'COL_1': ['1', '2', '3'],
    'COL_2': ['4.5', 'invalid', '6.5']
})

df_numeric = df_raw.apply(pd.to_numeric, errors='coerce')
print(df_numeric.dtypes)

# Self-validating assertions
assert pd.isna(df_numeric.loc[1, 'COL_2'])
```
