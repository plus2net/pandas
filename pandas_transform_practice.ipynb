## 1. Quick Square Root Transformation

Apply element-wise square root transformation using a custom lambda.

````python
import pandas as pd

df = pd.DataFrame({'A': [1, 4], 'B': [9, 16]})
result = df.transform(lambda x: x ** 0.5)
print(result)

# Self-validating assertion
assert result.loc[0, 'A'] == 1.0
assert result.loc[1, 'B'] == 4.0
````

**Expected output**

```text
     A    B
0  1.0  3.0
1  2.0  4.0
```

## 2. Multi-Function Transformations

Apply multiple NumPy mathematical functions (`sqrt`, `exp`) across columns.

````python
import numpy as np

df_vals = pd.DataFrame({'NUM': [1, 2, 3]})
result_multi = df_vals.transform([np.sqrt, np.exp])
print(result_multi)

# Self-validating assertion
assert 'sqrt' in result_multi.columns.get_level_values(1)
assert 'exp' in result_multi.columns.get_level_values(1)
````

**Expected output**

```text
       NUM          
      sqrt       exp
0  1.000000  2.718282
1  1.414214  7.389056
2  1.732051 20.085537
```

## 3. GroupBy Department Salary Broadcasting

Calculate department average salary and broadcast back to original row dimensions.

````python
df_sales = pd.DataFrame({
    'DEPT': ['IT', 'IT', 'HR', 'HR'],
    'SALARY': [70000, 80000, 50000, 60000]
})

df_sales['DEPT_AVG'] = df_sales.groupby('DEPT')['SALARY'].transform('mean')
print(df_sales)

# Self-validating assertion
assert df_sales.loc[0, 'DEPT_AVG'] == 75000.0
assert df_sales.loc[2, 'DEPT_AVG'] == 55000.0
````

**Expected output**

```text
  DEPT  SALARY  DEPT_AVG
0   IT   70000   75000.0
1   IT   80000   75000.0
2   HR   50000   55000.0
3   HR   60000   55000.0
```
