# Pandas: Pivot MultiIndex Rows to Columns with unstack()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-unstack.php](https://www.plus2net.com/python/pandas-unstack.php)

Run cells in order in Google Colab.

## Create MultiIndex row dataset

```python
import pandas as pd
idx = pd.MultiIndex.from_tuples([("Ravi", "Term1"), ("Ravi", "Term2"), ("Raju", "Term1"), ("Raju", "Term2")], names=["NAME", "TERM"])
df = pd.DataFrame({"MATH": [85, 90, 88, 92], "ENGLISH": [78, 82, 80, 85]}, index=idx)
print(df)
assert df.index.nlevels == 2
```

## Basic unstack operation

```python
unstacked = df.unstack()
print(unstacked)
assert unstacked.columns.nlevels == 2
```

## Unstacking specific levels

```python
unstacked_name = df.unstack(level="NAME")
print(unstacked_name)
assert "Ravi" in unstacked_name.columns.get_level_values(1)
```

## Filling missing values with fill_value

```python
sparse_idx = pd.MultiIndex.from_tuples([("Ravi", "Term1"), ("Raju", "Term2")], names=["NAME", "TERM"])
sparse_df = pd.DataFrame({"MATH": [85, 92]}, index=sparse_idx)
filled = sparse_df.unstack(fill_value=0)
print(filled)
assert filled.loc["Ravi", ("MATH", "Term2")] == 0
```

## Exercise solution

```python
ex_idx = pd.MultiIndex.from_tuples([("Ravi", "Math"), ("Ravi", "Science"), ("Raju", "Math"), ("Raju", "Science")])
ex_s = pd.Series([90, 95, 80, 85], index=ex_idx)
answer = ex_s.unstack()
print(answer)
assert list(answer.columns) == ["Math", "Science"]
```
