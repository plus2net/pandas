# Validating Email Addresses in Pandas DataFrames

Interactive companion notebook for [Pandas Validate Email](https://www.plus2net.com/python/pandas-validate_email.php). Covers vectorized regex matching and data sanitation.

## Vectorized Regex Email Validation

```python
import pandas as pd

df_subscribers = pd.DataFrame({
    "name": ["Ravi", "Alice", "Alex", "Ron", "InvalidUser", "Admin"],
    "raw_email": [
        " Ravi ",
        "  alice@example.com  ",
        "alex.invalid@",
        "ron.23@my_domain.com",
        "bad@domain..com",
        "admin@plus2net.com"
    ]
})
df_subscribers["clean_email"] = df_subscribers["raw_email"].str.strip().str.lower()
email_pattern = r"^[a-zA-Z0-9_.+-]+@[a-zA-Z0-9_-]+(?:\.[a-zA-Z0-9_-]+)+$"
df_subscribers["is_valid"] = df_subscribers["clean_email"].str.match(email_pattern, na=False)
valid_records = df_subscribers[df_subscribers["is_valid"]]
print("Valid Subscribers:")
print(valid_records[["name", "clean_email"]])
assert len(valid_records) == 3
assert set(valid_records["name"]) == {"Alice", "Ron", "Admin"}
```
