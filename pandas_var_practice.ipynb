# Pandas: Calculate Variance with var()

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-var.php](https://www.plus2net.com/python/pandas-var.php)

Run cells in order in Google Colab.

## Create numerical sample DataFrame

```python
import pandas as pd

df = pd.DataFrame({
    "COL_X": [2, 4, 6, 8, 10],
    "COL_Y": [5, 10, 15, 20, 25]
})
print(df)
assert df.shape == (5, 2)
```

## Basic variance calculation

```python
var_res = df.var()
print(var_res)
assert var_res["COL_X"] == 10.0
```

## Variance on a single Series

```python
s_var = df["COL_X"].var()
print(f"Variance: {s_var}")
assert s_var == 10.0
```

## Exercise solution

```python
vals = pd.Series([1, 3, 5, 7, 9], name="VALS")
answer = vals.var()
print(answer)
assert answer == 10.0
```
