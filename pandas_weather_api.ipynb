# Extracting Live Weather API Data into a Pandas DataFrame

**Tutorial Guide:** Learn more on the full tutorial page at [https://www.plus2net.com/python/pandas-weather-api.php](https://www.plus2net.com/python/pandas-weather-api.php)

Run cells in order in Google Colab. Uses free public Open-Meteo API (no key required).

## 1. Weather API to DataFrame

Fetch live hourly meteorological forecasts from Open-Meteo and convert into a Pandas DataFrame.

```python
import requests
import pandas as pd

url = "https://api.open-meteo.com/v1/forecast?latitude=52.52&longitude=13.41&hourly=temperature_2m,relative_humidity_2m"
response = requests.get(url)

if response.status_code == 200:
    data = response.json()
    hourly = data['hourly']
    df = pd.DataFrame({
        'Time': pd.to_datetime(hourly['time']),
        'Temperature_C': hourly['temperature_2m'],
        'Humidity': hourly['relative_humidity_2m']
    })
    print(df.head())
    assert not df.empty
    assert 'Temperature_C' in df.columns
else:
    print(f"Error: {response.status_code}")
```

## 2. Datetime Indexing & Resampling

Set the timestamp as the DataFrame index and resample into daily statistics.

```python
df.set_index('Time', inplace=True)
daily_avg = df.resample('D').mean()
print(daily_avg.head())
assert len(daily_avg) > 0
```

## 3. Saving Weather Data

Export the DataFrame to CSV and SQLite database.

```python
import sqlite3

df.to_csv("weather_forecast.csv")
conn = sqlite3.connect("weather.db")
df.to_sql("hourly_forecast", conn, if_exists="replace", index=True)
conn.close()
print("Weather data successfully saved and verified!")
assert not df.empty
```
